## Definire nuove funzioni

Una **definizione di funzione** (*function definition*) specifica il nome di una nuova funzione e la sequenza di istruzioni che vengono eseguite quando la funzione viene chiamata. Ecco un esempio:

````python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")
````

`def` è una parola chiave che indica che questa è una definizione di funzione.
Il nome della funzione è `print_lyrics`.
Qualsiasi nome valido per una variabile è valido anche per una funzione.

Le parentesi vuote dopo il nome indicano che questa funzione non riceve argomenti.

La prima riga della definizione di funzione si chiama **intestazione** (*header*) – il resto si chiama **corpo** (*body*).
L'intestazione deve terminare con i due punti e il corpo deve essere indentato (rientrato). Per convenzione, l'indentazione è sempre di quattro spazi.
Il corpo di questa funzione è formato da due istruzioni print; in generale, il corpo di una funzione può contenere un numero qualsiasi di istruzioni di qualsiasi tipo.

Definire una funzione crea un **oggetto funzione** (*function object*), che possiamo visualizzare così.

````python
print_lyrics
````

L'output indica che `print_lyrics` è una funzione che non riceve argomenti.
`__main__` è il nome del modulo che contiene `print_lyrics`.

Ora che abbiamo definito una funzione, possiamo chiamarla nello stesso modo in cui chiamiamo le funzioni predefinite.

````python
print_lyrics()
````

Quando la funzione viene eseguita, esegue le istruzioni del corpo, che visualizzano i primi due versi di "The Lumberjack Song" (la canzone del taglialegna).

## Parametri

Alcune delle funzioni che abbiamo visto richiedono argomenti; per esempio, quando chiami `abs` passi un numero come argomento.
Alcune funzioni ricevono più di un argomento; per esempio, `math.pow` ne riceve due, la base e l'esponente.

Ecco la definizione di una funzione che riceve un argomento.

````python
def print_twice(string):
    print(string)
    print(string)
````

Il nome di variabile tra parentesi è un **parametro** (*parameter*).
Quando la funzione viene chiamata, il valore dell'argomento viene assegnato al parametro.
Per esempio, possiamo chiamare `print_twice` così.

````python
print_twice('Dennis Moore, ')
````

Eseguire questa funzione ha lo stesso effetto di assegnare l'argomento al parametro e poi eseguire il corpo della funzione, così.

````python
string = 'Dennis Moore, '
print(string)
print(string)
````

Puoi anche usare una variabile come argomento.

````python
line = 'Dennis Moore, '
print_twice(line)
````

In questo esempio, il valore di `line` viene assegnato al parametro `string`.

## Chiamare funzioni

Una volta definita una funzione, puoi usarla all'interno di un'altra funzione.
Per dimostrarlo, scriveremo delle funzioni che stampano il testo di "The Spam Song" (<https://www.songfacts.com/lyrics/monty-python/the-spam-song>).

> Spam, Spam, Spam, Spam,
> Spam, Spam, Spam, Spam,
> Spam, Spam,
> (Lovely Spam, Wonderful Spam!)
> Spam, Spam,

Inizieremo con la funzione seguente, che riceve due parametri.

````python
def repeat(word, n):
    print(word * n)
````

Possiamo usare questa funzione per stampare il primo verso della canzone, così.

````python
spam = 'Spam, '
repeat(spam, 4)
````

Per visualizzare i primi due versi, possiamo definire una nuova funzione che usa `repeat`.

````python
def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
````

E poi chiamarla così.

````python
first_two_lines()
````

Per visualizzare gli ultimi tre versi, possiamo definire un'altra funzione, che usa anch'essa `repeat`.

````python
def last_three_lines():
    repeat(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repeat(spam, 2)
````

````python
last_three_lines()
````

Infine, possiamo mettere tutto insieme con una funzione che stampa l'intera strofa.

````python
def print_verse():
    first_two_lines()
    last_three_lines()
````

````python
print_verse()
````

Quando eseguiamo `print_verse`, questa chiama `first_two_lines`, che chiama `repeat`, che chiama `print`.
Sono un bel po' di funzioni.

Naturalmente avremmo potuto fare la stessa cosa con meno funzioni, ma lo scopo di questo esempio è mostrare come le funzioni possono lavorare insieme.

## Ripetizione

Se vogliamo visualizzare più di una strofa, possiamo usare un'istruzione `for`.
Ecco un esempio semplice.

````python
for i in range(2):
    print(i)
````

La prima riga è un'intestazione che termina con i due punti.
La seconda riga è il corpo, che deve essere indentato.

L'intestazione inizia con la parola chiave `for`, una nuova variabile chiamata `i` e un'altra parola chiave, `in`.
Usa la funzione `range` per creare una sequenza di due valori, che sono `0` e `1`.
In Python, quando iniziamo a contare, di solito partiamo da `0`.

Quando l'istruzione `for` viene eseguita, assegna il primo valore di `range` a `i` e poi esegue la funzione `print` nel corpo, che visualizza `0`.

Quando arriva alla fine del corpo, torna indietro all'intestazione: ecco perché questa istruzione si chiama **ciclo** (*loop*).
Al secondo giro del ciclo, assegna a `i` il valore successivo di `range` e lo visualizza.
Poi, dato che quello era l'ultimo valore di `range`, il ciclo termina.

Ecco come possiamo usare un ciclo `for` per stampare due strofe della canzone.

````python
for i in range(2):
    print("Verse", i)
    print_verse()
    print()
````

Puoi mettere un ciclo `for` all'interno di una funzione.
Per esempio, `print_n_verses` riceve un parametro chiamato `n`, che deve essere un intero, e visualizza il numero di strofe indicato.

````python
def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
````

In questo esempio non usiamo `i` nel corpo del ciclo, ma nell'intestazione deve comunque esserci un nome di variabile.

## Variabili e parametri sono locali

Quando crei una variabile all'interno di una funzione, questa è **locale** (*local*), il che
significa che esiste solo all'interno della funzione.
Per esempio, la funzione seguente riceve due argomenti, li concatena e stampa il risultato due volte.

````python
def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
````

Ecco un esempio che la usa:

````python
line1 = 'Always look on the '
line2 = 'bright side of life.'
cat_twice(line1, line2)
````

Quando `cat_twice` viene eseguita, crea una variabile locale chiamata `cat`, che viene distrutta quando la funzione termina.
Se proviamo a visualizzarla, otteniamo un `NameError`:

````python
%%expect NameError

print(cat)
````

Fuori dalla funzione, `cat` non è definita.

Anche i parametri sono locali.
Per esempio, fuori da `cat_twice` non esistono `part1` né `part2`.

## Diagrammi dello stack

Per tenere traccia di quali variabili si possono usare e dove, a volte è utile disegnare un **diagramma dello stack** (*stack diagram*).
Come i diagrammi di stato, i diagrammi dello stack mostrano il valore di ogni variabile, ma mostrano anche la funzione a cui appartiene ciascuna variabile.

Ogni funzione è rappresentata da un **frame** (riquadro).
Un frame è un rettangolo con il nome di una funzione all'esterno e i parametri e le variabili locali della funzione all'interno.

Ecco il diagramma dello stack per l'esempio precedente.

````python
from diagram import make_frame, Stack

d1 = dict(line1=line1, line2=line2)
frame1 = make_frame(d1, name='__main__', dy=-0.3, loc='left')

d2 = dict(part1=line1, part2=line2, cat=line1+line2)
frame2 = make_frame(d2, name='cat_twice', dy=-0.3, 
                    offsetx=0.03, loc='left')

d3 = dict(string=line1+line2)
frame3 = make_frame(d3, name='print_twice', 
                    offsetx=0.04, offsety=-0.3, loc='left')

d4 = {"?": line1+line2}
frame4 = make_frame(d4, name='print', 
                    offsetx=-0.22, offsety=0, loc='left')

stack = Stack([frame1, frame2, frame3, frame4], dy=-0.8)
````

````python
from diagram import diagram, adjust


width, height, x, y = [3.77, 2.9, 1.1, 2.65]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)

import matplotlib.pyplot as plt
plt.savefig('chap03_stack_diagram.png', dpi=300)
````

I frame sono disposti in una pila (*stack*) che indica quale funzione ha chiamato
quale, e così via. Leggendo dal basso, `print` è stata chiamata da `print_twice`, che è stata chiamata da `cat_twice`, che è stata chiamata da `__main__` – che è un nome speciale per il frame più in alto.
Quando crei una variabile fuori da qualsiasi funzione, essa appartiene a `__main__`.

Nel frame di `print`, il punto interrogativo indica che non conosciamo il nome del parametro.

## Traceback

Quando si verifica un errore di runtime in una funzione, Python visualizza il nome della funzione che era in esecuzione, il nome della funzione che l'ha chiamata, e così via, risalendo lo stack.
Per vedere un esempio, definirò una versione di `print_twice` che contiene un errore – cerca di stampare `cat`, che è una variabile locale di un'altra funzione.

````python
def print_twice(string):
    print(cat)            # NameError
    print(cat)
````

Ecco cosa succede quando eseguiamo `cat_twice`.

````python
# This cell tells Jupyter to provide detailed debugging information
# when a runtime error occurs, including a traceback.

%xmode Verbose
````

````python
%%expect NameError

cat_twice(line1, line2)
````

Il messaggio di errore include un **traceback**, che mostra la funzione che era in esecuzione quando si è verificato l'errore, la funzione che l'ha chiamata, e così via.
In questo esempio, mostra che `cat_twice` ha chiamato `print_twice` e che l'errore si è verificato in `print_twice`.

L'ordine delle funzioni nel traceback è lo stesso dell'ordine dei frame nel diagramma dello stack.
La funzione che era in esecuzione si trova in fondo.

## Perché le funzioni?

Potrebbe non essere ancora chiaro perché valga la pena dividere un programma in
funzioni.
I motivi sono diversi:

-   Creare una nuova funzione ti dà l'occasione di dare un nome a un gruppo di
    istruzioni, il che rende il programma più facile da leggere e da correggere.

-   Le funzioni possono rendere un programma più piccolo eliminando il codice ripetuto.
    In seguito, se fai una modifica, devi farla in un solo punto.

-   Dividere un programma lungo in funzioni ti permette di correggere le parti
    una alla volta e poi assemblarle in un insieme funzionante.

-   Le funzioni ben progettate sono spesso utili per molti programmi. Una volta
    scritta e corretta una funzione, puoi riutilizzarla.

## Debugging

Il debugging può essere frustrante, ma è anche stimolante, interessante e a volte persino divertente.
Ed è una delle abilità più importanti che puoi imparare.

Per certi versi il debugging è come un lavoro da investigatore.
Hai degli indizi e devi dedurre gli eventi che hanno portato ai
risultati che vedi.

Il debugging è anche come la scienza sperimentale.
Una volta che hai un'idea di cosa sta andando storto, modifichi il programma e riprovi.
Se la tua ipotesi era corretta, puoi prevedere il risultato della modifica e fai un passo avanti verso un programma funzionante.
Se la tua ipotesi era sbagliata, devi formularne una nuova.

Per alcune persone, programmare e fare debugging sono la stessa cosa; cioè, programmare è il processo di correggere gradualmente un programma finché non fa ciò che vuoi.
L'idea è che dovresti partire da un programma funzionante e fare piccole modifiche, correggendole man mano.

Se ti accorgi di passare molto tempo a fare debugging, spesso è segno che stai scrivendo troppo codice prima di iniziare a testarlo.
Se fai passi più piccoli, potresti scoprire di riuscire ad andare più veloce.

## Glossario

**definizione di funzione** (*function definition*):
Un'istruzione che crea una funzione.

**intestazione** (*header*):
La prima riga di una definizione di funzione.

**corpo** (*body*):
La sequenza di istruzioni all'interno di una definizione di funzione.

**oggetto funzione** (*function object*):
Un valore creato da una definizione di funzione.
Il nome della funzione è una variabile che fa riferimento a un oggetto funzione.

**parametro** (*parameter*):
Un nome usato all'interno di una funzione per fare riferimento al valore passato come argomento.

**ciclo** (*loop*):
Un'istruzione che esegue una o più istruzioni, spesso ripetutamente.

**variabile locale** (*local variable*):
Una variabile definita all'interno di una funzione, accessibile solo all'interno della funzione.

**diagramma dello stack** (*stack diagram*):
Una rappresentazione grafica di una pila di funzioni, delle loro variabili e dei valori a cui fanno riferimento.

**frame**:
Un riquadro in un diagramma dello stack che rappresenta una chiamata di funzione.
Contiene le variabili locali e i parametri della funzione.

**traceback**:
Un elenco delle funzioni in esecuzione, stampato quando si verifica un'eccezione.
