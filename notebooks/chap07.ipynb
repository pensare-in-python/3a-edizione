Nota che ho cambiato il nome della variabile da `i` a `letter` (lettera), che fornisce più informazioni sul valore a cui fa riferimento.
La variabile definita in un ciclo `for` si chiama **variabile di ciclo** (*loop variable*).

Ora che possiamo scorrere le lettere di una parola, possiamo verificare se contiene la lettera "e".

```python
for letter in "Gadsby":
    if letter == 'E' or letter == 'e':
        print('This word has an "e"')
```

Prima di andare avanti, incapsuliamo quel ciclo in una funzione.

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            print('This word has an "e"')
```

E rendiamola una funzione pura che restituisce `True` se la parola contiene una "e" e `False` altrimenti.

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            return True
    return False
```

Possiamo generalizzarla in modo che riceva la parola come parametro.

```python
def has_e(word):
    for letter in word:
        if letter == 'E' or letter == 'e':
            return True
    return False
```

## L'operatore in

La versione di `has_e` che abbiamo scritto in questo capitolo è più complicata del necessario.
Python fornisce un operatore, `in`, che verifica se un carattere compare in una stringa.

```python
word = 'Gadsby'
'e' in word
```

Quindi possiamo riscrivere `has_e` così.

```python
def has_e(word):
    if 'E' in word or 'e' in word:
        return True
    else:
        return False
```

E poiché la condizione dell'istruzione `if` ha un valore booleano, possiamo eliminare l'istruzione `if` e restituire direttamente il booleano.

```python
def has_e(word):
    return 'E' in word or 'e' in word
```

Possiamo semplificare ancora di più questa funzione usando il metodo `lower`, che converte in minuscolo le lettere di una stringa.
Ecco un esempio.

```python
word.lower()
```

`lower` crea una nuova stringa – non modifica la stringa esistente – quindi il valore di `word` resta invariato.

```python
word
```

Ecco come possiamo usare `lower` in `has_e`.

```python
def has_e(word):
    return 'e' in word.lower()
```

```python
has_e('Gadsby')
```

```python
has_e('Emma')
```

## Ricerca

Basandoci su questa versione più semplice di `has_e`, scriviamo una funzione più generale chiamata `uses_any` che riceve un secondo parametro, una stringa di lettere.
Restituisce `True` se la parola usa almeno una delle lettere e `False` altrimenti.

```python
def uses_any(word, letters):
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

Ecco un esempio in cui il risultato è `True`.

```python
uses_any('banana', 'aeiou')
```

E un altro in cui è `False`.

```python
uses_any('apple', 'xyz')
```

`uses_any` converte `word` e `letters` in minuscolo, quindi funziona con qualsiasi combinazione di maiuscole e minuscole.

```python
uses_any('Banana', 'AEIOU')
```

La struttura di `uses_any` è simile a quella di `has_e`.
Scorre le lettere di `word` e le controlla una alla volta.
Se ne trova una che compare in `letters`, restituisce immediatamente `True`.
Se arriva alla fine del ciclo senza trovarne nessuna, restituisce `False`.

Questo schema si chiama **ricerca lineare** (*linear search*).

## Doctest

Nel [capitolo 4](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap04.ipynb) abbiamo usato una docstring per documentare una funzione – cioè per spiegare cosa fa.
È anche possibile usare una docstring per *testare* una funzione.
Ecco una versione di `uses_any` con una docstring che include dei test.

```python
def uses_any(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any('banana', 'aeiou')
    True
    >>> uses_any('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

Ogni test inizia con `>>>`, che in alcuni ambienti Python si usa come prompt per indicare dove l'utente può scrivere codice.
In un doctest, il prompt è seguito da un'espressione, di solito una chiamata di funzione.
La riga successiva indica il valore che l'espressione dovrebbe avere se la funzione funziona correttamente.

Nel primo esempio, `'banana'` usa `'a'`, quindi il risultato dovrebbe essere `True`.
Nel secondo esempio, `'apple'` non usa nessuna delle lettere di `'xyz'`, quindi il risultato dovrebbe essere `False`.

Per eseguire questi test, dobbiamo importare il modulo `doctest` ed eseguire una funzione chiamata `run_docstring_examples`.
Per rendere questa funzione più facile da usare, ho scritto la funzione seguente, che riceve un oggetto funzione come argomento.

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

Non abbiamo ancora visto `globals` e `__name__` – puoi ignorarli.
Ora possiamo testare `uses_any` così.

```python
run_doctests(uses_any)
```

`run_doctests` trova le espressioni nella docstring e le valuta.
Se il risultato è il valore atteso, il test **passa** (*pass*).
Altrimenti **fallisce** (*fail*).

Se tutti i test passano, `run_doctests` non visualizza alcun output – in questo caso, nessuna notizia è una buona notizia.
Per vedere cosa succede quando un test fallisce, ecco una versione sbagliata di `uses_any`.

```python
def uses_any_incorrect(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any_incorrect('banana', 'aeiou')
    True
    >>> uses_any_incorrect('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
        else:
            return False     # INCORRECT!
```

Ed ecco cosa succede quando la testiamo.

```python
run_doctests(uses_any_incorrect)
```

L'output include l'esempio che è fallito, il valore che la funzione avrebbe dovuto produrre e il valore che la funzione ha effettivamente prodotto.

Se non sei sicuro del perché questo test sia fallito, avrai occasione di fare debugging in un esercizio.

## Glossario

**variabile di ciclo** (*loop variable*):
Una variabile definita nell'intestazione di un ciclo `for`.

**oggetto file** (*file object*):
Un oggetto che rappresenta un file aperto e tiene traccia di quali parti del file sono state lette o scritte.

**metodo** (*method*):
Una funzione associata a un oggetto e chiamata usando l'operatore punto.

**aggiornamento** (*update*):
Un'istruzione di assegnazione che dà un nuovo valore a una variabile già esistente, invece di creare una nuova variabile.

**inizializzare** (*initialize*):
Creare una nuova variabile e darle un valore.

**incremento** (*increment*):
L'aumento del valore di una variabile.

**decremento** (*decrement*):
La diminuzione del valore di una variabile.

**contatore** (*counter*):
Una variabile usata per contare qualcosa, di solito inizializzata a zero e poi incrementata.

**ricerca lineare** (*linear search*):
Uno schema di calcolo che esamina una sequenza di elementi e si ferma quando trova ciò che cerca.

**passare** (*pass*):
Se un test viene eseguito e il risultato è quello atteso, il test passa.

**fallire** (*fail*):
Se un test viene eseguito e il risultato non è quello atteso, il test fallisce.
