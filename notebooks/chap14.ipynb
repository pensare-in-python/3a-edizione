## Tipi definiti dal programmatore

Abbiamo usato molti dei tipi predefiniti di Python – ora definiremo un nuovo tipo.
Come primo esempio, creeremo un tipo chiamato `Time` che rappresenta un'ora del giorno.
Un tipo definito dal programmatore si chiama anche **classe** (*class*).
Una definizione di classe appare così:

```python
class Time:
    """Represents a time of day."""
```

L'intestazione indica che la nuova classe si chiama `Time`.
Il corpo è una docstring che spiega a cosa serve la classe.
Definire una classe crea un **oggetto classe** (*class object*).

L'oggetto classe è come una fabbrica per creare oggetti.
Per creare un oggetto `Time`, chiami `Time` come se fosse una funzione.

```python
print(lunch)
```

Creare un nuovo oggetto si chiama **istanziazione** (*instantiation*), e l'oggetto è un'**istanza** (*instance*) della classe.

## Oggetti come valori di ritorno

Le funzioni possono restituire oggetti. Per esempio, `make_time` riceve parametri chiamati `hour`, `minute` e `second`, li memorizza come attributi in un oggetto `Time` e restituisce il nuovo oggetto.

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

Potrebbe sorprendere che i parametri abbiano gli stessi nomi degli attributi, ma è un modo comune di scrivere una funzione come questa.
Ecco come usiamo `make_time` per creare un oggetto `Time`.

```python
time = make_time(11, 59, 1)
print_time(time)
```

## Gli oggetti sono mutabili

Supponiamo che tu stia andando alla proiezione di un film, come *Monty Python e il Sacro Graal*, che inizia alle `9:20 PM` (21:20) e dura `92` minuti, cioè un'ora e `32` minuti.
A che ora finirà il film?

Per prima cosa, creeremo un oggetto `Time` che rappresenta l'orario di inizio.

```python
start = make_time(9, 20, 0)
print_time(start)
```

Per trovare l'orario di fine, possiamo modificare gli attributi dell'oggetto `Time`, aggiungendo la durata del film.

```python
start.hour += 1
start.minute += 32
print_time(start)
```

Il film finirà alle 10:52 PM (22:52).

Incapsuliamo questo calcolo in una funzione e generalizziamola in modo che riceva la durata del film in tre parametri: `hours`, `minutes` e `seconds`.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
```

Ecco un esempio che ne mostra l'effetto.

```python
start = make_time(9, 20, 0)
increment_time(start, 1, 32, 0)
print_time(start)
```

## Funzioni pure

Possiamo usare `copy` per scrivere funzioni pure che non modificano i loro parametri.
Per esempio, ecco una funzione che riceve un oggetto `Time` e una durata in ore, minuti e secondi.
Crea una copia dell'oggetto originale, usa `increment_time` per modificare la copia e la restituisce.

```python
def add_time(time, hours, minutes, seconds):
    total = copy(time)
    increment_time(total, hours, minutes, seconds)
    return total
```

Ecco come la usiamo.

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

Il valore di ritorno è un nuovo oggetto che rappresenta l'orario di fine del film.
E possiamo confermare che `start` è invariato.

```python
print_time(start)
```

`add_time` è una **funzione pura** (*pure function*) perché non modifica nessuno degli oggetti che le vengono passati come argomenti e il suo unico effetto è restituire un valore.

Tutto ciò che si può fare con le funzioni non pure si può fare anche con le funzioni pure.
In effetti, alcuni linguaggi di programmazione permettono solo funzioni pure.
I programmi che usano funzioni pure potrebbero essere meno soggetti a errori, ma le funzioni non pure a volte sono comode e possono essere più efficienti.

In generale, ti suggerisco di scrivere funzioni pure ogni volta che è ragionevole e di ricorrere a funzioni non pure solo se c'è un vantaggio evidente.
Questo approccio si potrebbe chiamare **stile di programmazione funzionale** (*functional programming style*).

## Prototipo e toppa

Nell'esempio precedente, `increment_time` e `add_time` sembrano funzionare, ma se proviamo un altro esempio, vedremo che non sono del tutto corrette.

Supponiamo che tu arrivi al cinema e scopra che il film inizia alle `9:40`, non alle `9:20`.
Ecco cosa succede quando calcoliamo l'orario di fine aggiornato.

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

Il risultato non è un orario valido.
Il problema è che `increment_time` non gestisce i casi in cui il numero di secondi o di minuti supera `60`.

Ecco una versione migliorata che verifica se `second` è maggiore o uguale a `60` – in tal caso incrementa `minute` – poi verifica se `minute` è maggiore o uguale a `60` – in tal caso incrementa `hour`.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds

    if time.second >= 60:
        time.second -= 60
        time.minute += 1

    if time.minute >= 60:
        time.minute -= 60
        time.hour += 1
```

Correggere `increment_time` corregge anche `add_time`, che la usa.
Quindi ora l'esempio precedente funziona correttamente.

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

Ma questa funzione non è ancora corretta, perché gli argomenti potrebbero essere maggiori di `60`.
Per esempio, supponiamo che la durata ci venga data come `92` minuti, invece che come `1` ora e `32` minuti.
Potremmo chiamare `add_time` così.

```python
end = add_time(start, 0, 92, 0)
print_time(end)
```

Il risultato non è un orario valido.
Proviamo quindi un approccio diverso, usando la funzione `divmod`.
Faremo una copia di `start` e la modificheremo incrementando l'attributo `minute`.

```python
end = copy(start)
end.minute = start.minute + 92
end.minute
```

Ora `minute` è `132`, cioè `2` ore e `12` minuti.
Possiamo usare `divmod` per dividere per `60` e ottenere il numero di ore intere e il numero di minuti rimanenti.

```python
carry, end.minute = divmod(end.minute, 60)
carry, end.minute
```

Ora `minute` è corretto, e possiamo aggiungere le ore a `hour`.

```python
end.hour += carry
print_time(end)
```

Il risultato è un orario valido.
Possiamo fare la stessa cosa con `hour` e `second`, e incapsulare l'intero procedimento in una funzione.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
    
    carry, time.second = divmod(time.second, 60)
    carry, time.minute = divmod(time.minute + carry, 60)
    carry, time.hour = divmod(time.hour + carry, 24)
```

Con questa versione di `increment_time`, `add_time` funziona correttamente, anche se gli argomenti superano `60`.

```python
end = add_time(start, 0, 90, 120)
print_time(end)
```

Questa sezione illustra un piano di sviluppo dei programmi che chiamo **prototipo e toppa** (*prototype and patch*).
Siamo partiti da un prototipo semplice che funzionava correttamente con il primo esempio.
Poi lo abbiamo testato con esempi più difficili – quando abbiamo trovato un errore, abbiamo modificato il programma per correggerlo, come quando si mette una toppa a uno pneumatico bucato.

Questo approccio può essere efficace, soprattutto se non hai ancora una comprensione profonda del problema.
Ma le correzioni incrementali possono generare codice inutilmente complicato – perché gestisce molti casi particolari – e poco affidabile – perché è difficile sapere se hai
trovato tutti gli errori.

## Sviluppo guidato dalla progettazione

Un piano alternativo è lo **sviluppo guidato dalla progettazione** (*design-first development*), che comporta più pianificazione prima di realizzare il prototipo. In un processo guidato dalla progettazione, a volte un'intuizione di alto livello sul problema rende la programmazione molto più facile.

In questo caso, l'intuizione è che possiamo pensare a un oggetto `Time` come a un numero di tre cifre in base 60 – detta anche sessagesimale.
L'attributo `second` è la "colonna delle unità", l'attributo `minute` è la "colonna dei sessanta"
e l'attributo `hour` è la "colonna dei tremilaseicento".
Quando abbiamo scritto `increment_time`, stavamo di fatto facendo un'addizione in base 60, ed è per questo che dovevamo fare il riporto da una colonna alla successiva.

Questa osservazione suggerisce un altro approccio all'intero problema – possiamo convertire gli oggetti `Time` in interi e sfruttare il fatto che Python sa fare l'aritmetica con gli interi.

Ecco una funzione che converte un `Time` in un intero.

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

Il risultato è il numero di secondi dall'inizio del giorno.
Per esempio, `01:01:01` è `1` ora, `1` minuto e `1` secondo dall'inizio del giorno, cioè la somma di `3600` secondi, `60` secondi e `1` secondo.

```python
time = make_time(1, 1, 1)
print_time(time)
time_to_int(time)
```

Ed ecco una funzione che va nella direzione opposta – convertendo un intero in un oggetto `Time` – usando la funzione `divmod`.

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

Possiamo testarla riconvertendo l'esempio precedente in un `Time`.

```python
time = int_to_time(3661)
print_time(time)
```

Usando queste funzioni, possiamo scrivere una versione più concisa di `add_time`.

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

La prima riga converte gli argomenti in un oggetto `Time` chiamato `duration`.
La seconda riga converte `time` e `duration` in secondi e li somma.
La terza riga converte la somma in un oggetto `Time` e lo restituisce.

Ecco come funziona.

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

Per certi versi, convertire dalla base 60 alla base 10 e viceversa è più difficile che
gestire direttamente gli orari. La conversione di base è più astratta; la nostra intuizione
per gestire gli orari è migliore.

Ma se abbiamo l'intuizione di trattare gli orari come numeri in base 60 – e investiamo lo sforzo di scrivere le funzioni di conversione `time_to_int` e `int_to_time` – otteniamo un programma più breve, più facile da leggere e da correggere, e più affidabile.

È anche più facile aggiungere funzionalità in seguito. Per esempio, immagina di sottrarre due oggetti `Time` per trovare l'intervallo tra loro.
L'approccio ingenuo è implementare la sottrazione con il prestito.
Usare le funzioni di conversione è più facile e più probabilmente corretto.

Paradossalmente, a volte rendere un problema più difficile – o più generale – lo rende più facile, perché ci sono meno casi particolari e meno occasioni di errore.

## Glossario

**programmazione orientata agli oggetti** (*object-oriented programming*):
Uno stile di programmazione che usa gli oggetti per organizzare codice e dati.

**classe** (*class*):
Un tipo definito dal programmatore. Una definizione di classe crea un nuovo oggetto classe.

**oggetto classe** (*class object*):
Un oggetto che rappresenta una classe – è il risultato di una definizione di classe.

**istanziazione** (*instantiation*):
Il processo di creare un oggetto che appartiene a una classe.

**istanza** (*instance*):
Un oggetto che appartiene a una classe.

**attributo** (*attribute*):
Una variabile associata a un oggetto, chiamata anche variabile d'istanza.

**diagramma degli oggetti** (*object diagram*):
Una rappresentazione grafica di un oggetto, dei suoi attributi e dei loro valori.

**specificatore di formato** (*format specifier*):
In una f-string, uno specificatore di formato determina come un valore viene convertito in stringa.

**funzione pura** (*pure function*):
Una funzione che non modifica i suoi parametri e non ha altri effetti oltre a restituire un valore.

**stile di programmazione funzionale** (*functional programming style*):
Un modo di programmare che usa funzioni pure ogni volta che è possibile.

**prototipo e toppa** (*prototype and patch*):
Un modo di sviluppare programmi partendo da una bozza approssimativa e aggiungendo gradualmente funzionalità e correggendo bug.

**sviluppo guidato dalla progettazione** (*design-first development*):
Un modo di sviluppare programmi con una pianificazione più accurata rispetto a prototipo e toppa.
