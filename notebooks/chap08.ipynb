## Una stringa è una sequenza

Una stringa è una sequenza di caratteri. Un **carattere** (*character*) può essere una lettera (di quasi qualsiasi alfabeto), una cifra, un segno di punteggiatura o uno spazio.

Puoi selezionare un carattere di una stringa con l'operatore parentesi quadre.
Questa istruzione di esempio seleziona il carattere numero 1 di `fruit` e
lo assegna a `letter`:

```python
fruit = 'banana'
letter = fruit[1]
```

L'espressione tra parentesi quadre è un **indice** (*index*), così chiamato perché *indica* quale carattere della sequenza selezionare.
Ma il risultato potrebbe non essere quello che ti aspetti.

```python
letter
```

La lettera con indice `1` è in realtà la seconda lettera della stringa.
Un indice è uno scostamento (*offset*) dall'inizio della stringa, quindi lo scostamento della prima lettera è `0`.

```python
fruit[0]
```

Puoi pensare a `'b'` come alla lettera "zeresima" di `'banana'`.

L'indice tra parentesi quadre può essere una variabile.

```python
i = 1
fruit[i]
```

Oppure un'espressione che contiene variabili e operatori.

```python
fruit[i+1]
```

Ma il valore dell'indice deve essere un intero – altrimenti ottieni un `TypeError`.

```python
%%expect TypeError

fruit[1.5]
```

Come abbiamo visto nel capitolo 1, possiamo usare la funzione predefinita `len` per ottenere la lunghezza di una stringa.

```python
n = len(fruit)
n
```

Per ottenere l'ultima lettera di una stringa, potresti essere tentato di scrivere questo:

```python
%%expect IndexError

fruit[n]
```

Ma questo causa un `IndexError` perché in `'banana'` non c'è nessuna lettera con indice 6. Poiché abbiamo iniziato a contare da `0`, le sei lettere sono numerate da `0` a `5`. Per ottenere l'ultimo carattere, devi sottrarre `1` a `n`:

```python
fruit[n-1]
```

Ma c'è un modo più semplice.
Per ottenere l'ultima lettera di una stringa, puoi usare un indice negativo, che conta all'indietro dalla fine.

```python
fruit[-1]
```

L'indice `-1` seleziona l'ultima lettera, `-2` la penultima, e così via.

## Slice di stringhe

Un segmento di una stringa si chiama **slice** (fetta).
Selezionare una slice è simile a selezionare un carattere.

```python
fruit = 'banana'
fruit[0:3]
```

L'operatore `[n:m]` restituisce la parte della stringa dal carattere `n`-esimo
al carattere `m`-esimo, includendo il primo ma escludendo il secondo.
Questo comportamento è controintuitivo, ma può aiutare immaginare che gli indici puntino *tra* i caratteri, come in questa figura:

```python
from diagram import make_binding, Element, Value

binding = make_binding("fruit", ' b a n a n a ')
elements = [Element(Value(i), None) for i in range(7)]
```

```python
import matplotlib.pyplot as plt
from diagram import diagram, adjust
from matplotlib.transforms import Bbox

width, height, x, y = [1.35, 0.54, 0.23, 0.39]

ax = diagram(width, height)
bbox = binding.draw(ax, x, y)
bboxes = [bbox]

def draw_elts(x, y, elements):
    for elt in elements:
        bbox = elt.draw(ax, x, y, draw_value=False)
        bboxes.append(bbox)

        x1 = (bbox.xmin + bbox.xmax) / 2
        y1 = bbox.ymax + 0.02
        y2 = y1 + 0.14
        handle = plt.plot([x1, x1], [y1, y2], ':', lw=0.5, color='gray')
        x += 0.105
    
draw_elts(x + 0.48, y - 0.25, elements)
bbox = Bbox.union(bboxes)
# adjust(x, y, bbox)
```

Per esempio, la slice `[3:6]` seleziona le lettere `ana`, il che significa che `6` è valido come parte di una slice, ma non come indice.

Se ometti il primo indice, la slice parte dall'inizio della stringa.

```python
fruit[:3]
```

Se ometti il secondo indice, la slice arriva fino alla fine della stringa:

```python
fruit[3:]
```

Se il primo indice è maggiore o uguale al secondo, il risultato è una **stringa vuota** (*empty string*), rappresentata da due apici:

```python
fruit[3:3]
```

Una stringa vuota non contiene caratteri e ha lunghezza 0.

## Le stringhe sono immutabili

Si potrebbe essere tentati di usare l'operatore `[]` a sinistra di
un'assegnazione, con l'intenzione di cambiare un carattere di una stringa, così:

```python
%%expect TypeError

greeting = 'Hello, world!'
greeting[0] = 'J'
```

Il risultato è un `TypeError`.
Nel messaggio di errore, l'"oggetto" (*object*) è la stringa e l'"elemento" (*item*) è il carattere
che abbiamo cercato di assegnare.
Per ora, un **oggetto** è la stessa cosa di un valore, ma affineremo questa definizione più avanti.

Il motivo di questo errore è che le stringhe sono **immutabili** (*immutable*), il che significa che non puoi modificare una stringa esistente.
Il massimo che puoi fare è creare una nuova stringa che sia una variante dell'originale.

```python
new_greeting = 'J' + greeting[1:]
new_greeting
```

Questo esempio concatena una nuova prima lettera a una slice di `greeting`.
Non ha alcun effetto sulla stringa originale.

```python
greeting
```

## Confronto tra stringhe

Gli operatori relazionali funzionano sulle stringhe. Per vedere se due stringhe sono
uguali, possiamo usare l'operatore `==`.

```python
word = 'banana'

if word == 'banana':
    print('All right, banana.')
```

Le altre operazioni relazionali sono utili per mettere le parole in ordine
alfabetico:

```python
def compare_word(word):
    if word < 'banana':
        print(word, 'comes before banana.')
    elif word > 'banana':
        print(word, 'comes after banana.')
    else:
        print('All right, banana.')
```

```python
compare_word('apple')
```

Python non tratta le lettere maiuscole e minuscole come fanno
le persone. Tutte le lettere maiuscole vengono prima di tutte le lettere
minuscole, quindi:

```python
compare_word('Pineapple')
```

Per risolvere questo problema, possiamo convertire le stringhe in un formato standard, per esempio tutte minuscole, prima di eseguire il confronto.
Tienilo a mente se dovessi difenderti da un uomo armato di un Ananas (*Pineapple*).

## Metodi delle stringhe

Le stringhe forniscono metodi che eseguono diverse operazioni utili.
Un metodo è simile a una funzione – riceve argomenti e restituisce un valore – ma la sintassi è diversa.
Per esempio, il metodo `upper` prende una stringa e restituisce una nuova stringa con tutte le lettere maiuscole.

Invece della sintassi da funzione `upper(word)`, usa la sintassi da metodo `word.upper()`.

```python
word = 'banana'
new_word = word.upper()
new_word
```

Questo uso dell'operatore punto specifica il nome del metodo, `upper`, e il nome della stringa a cui applicare il metodo, `word`.
Le parentesi vuote indicano che questo metodo non riceve argomenti.

Una chiamata di metodo si chiama **invocazione** (*invocation*); in questo caso diremmo che stiamo invocando `upper` su `word`.

## Scrivere file

Gli operatori e i metodi delle stringhe sono utili per leggere e scrivere file di testo.
Come esempio, lavoreremo con il testo di *Dracula*, un romanzo di Bram Stoker disponibile su Project Gutenberg (<https://www.gutenberg.org/ebooks/345>).

```python
download('https://github.com/AllenDowney/ThinkPython/raw/v3/pg345.txt');
```

Ho scaricato il libro in un file di testo semplice chiamato `pg345.txt`, che possiamo aprire in lettura così:

L'argomento keyword `encoding='utf-8'` non compare nel libro stampato.
Serve solo su Windows.

```python
reader = open('pg345.txt', encoding='utf-8')
```

Oltre al testo del libro, questo file contiene una sezione all'inizio con informazioni sul libro e una sezione alla fine con informazioni sulla licenza.
Prima di elaborare il testo, possiamo rimuovere questo materiale aggiuntivo trovando le righe speciali all'inizio e alla fine che cominciano con `'***'`.

La funzione seguente riceve una riga e verifica se è una delle righe speciali.
Usa il metodo `startswith`, che verifica se una stringa inizia con una data sequenza di caratteri.

```python
def is_special_line(line):
    return line.startswith('*** ')
```

Possiamo usare questa funzione per scorrere le righe del file e stampare solo le righe speciali.

```python
for line in reader:
    if is_special_line(line):
        print(line.strip())
```

Ora creiamo un nuovo file, chiamato `pg345_cleaned.txt`, che contiene solo il testo del libro.
Per scorrere di nuovo il libro, dobbiamo riaprirlo in lettura.
E, per scrivere un nuovo file, possiamo aprirlo in scrittura.

```python
reader = open('pg345.txt', encoding='utf-8')
writer = open('pg345_cleaned.txt', 'w', encoding='utf-8')
```

`open` riceve un parametro facoltativo che specifica la "modalità" (*mode*) – in questo esempio, `'w'` indica che stiamo aprendo il file in scrittura.
Se il file non esiste, verrà creato; se esiste già, il suo contenuto verrà sostituito.

Come primo passo, scorreremo il file finché non troviamo la prima riga speciale.

```python
for line in reader:
    if is_special_line(line):
        break
```

L'istruzione `break` "rompe" il ciclo – cioè lo fa terminare immediatamente, prima di arrivare alla fine del file.

Quando il ciclo termina, `line` contiene la riga speciale che ha reso vera la condizione.

```python
line
```

Poiché `reader` tiene traccia di dove si trova nel file, possiamo usare un secondo ciclo per riprendere da dove avevamo lasciato.

Il ciclo seguente legge il resto del file, una riga alla volta.
Quando trova la riga speciale che indica la fine del testo, esce dal ciclo.
Altrimenti, scrive la riga nel file di output.

```python
for line in reader:
    if is_special_line(line):
        break
    writer.write(line)
```

Quando questo ciclo termina, `line` contiene la seconda riga speciale.

```python
line
```

A questo punto `reader` e `writer` sono ancora aperti, il che significa che potremmo continuare a leggere righe da `reader` o a scrivere righe in `writer`.
Per indicare che abbiamo finito, possiamo chiudere entrambi i file invocando il metodo `close`.

```python
reader.close()
writer.close()
```

Per verificare se il procedimento ha avuto successo, possiamo leggere le prime righe del nuovo file appena creato.

```python
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    line = line.strip()
    if len(line) > 0:
        print(line)
    if line.endswith('Stoker'):
        break
```

Il metodo `endswith` verifica se una stringa termina con una data sequenza di caratteri.

## Trova e sostituisci

Nella traduzione islandese di *Dracula* del 1901, il nome di uno dei personaggi fu cambiato da "Jonathan" a "Thomas".
Per fare questa modifica nella versione inglese, possiamo scorrere il libro, usare il metodo `replace` per sostituire un nome con un altro e scrivere il risultato in un nuovo file.

Inizieremo contando le righe nella versione ripulita del file.

```python
total = 0
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    total += 1
    
total
```

Per vedere se una riga contiene "Jonathan", possiamo usare l'operatore `in`, che verifica se questa sequenza di caratteri compare in un punto qualsiasi della riga.

```python
total = 0
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    if 'Jonathan' in line:
        total += 1

total
```

Ci sono 199 righe che contengono il nome, ma questo non è proprio il numero totale di volte in cui compare, perché può comparire più di una volta in una riga.
Per ottenere il totale, possiamo usare il metodo `count`, che restituisce il numero di volte in cui una sequenza compare in una stringa.

```python
total = 0
for line in open('pg345_cleaned.txt', encoding='utf-8'):
    total += line.count('Jonathan')

total
```

Ora possiamo sostituire `'Jonathan'` con `'Thomas'` così:

```python
writer = open('pg345_replaced.txt', 'w', encoding='utf-8')

for line in open('pg345_cleaned.txt', encoding='utf-8'):
    line = line.replace('Jonathan', 'Thomas')
    writer.write(line)
```

Il risultato è un nuovo file chiamato `pg345_replaced.txt` che contiene una versione di *Dracula* in cui Jonathan Harker si chiama Thomas.

```python
total = 0
for line in open('pg345_replaced.txt', encoding='utf-8'):
    total += line.count('Thomas')

total
```

## Espressioni regolari

Se sappiamo esattamente quale sequenza di caratteri cerchiamo, possiamo usare l'operatore `in` per trovarla e il metodo `replace` per sostituirla.
Ma c'è un altro strumento, chiamato **espressione regolare** (*regular expression*), che può eseguire anche queste operazioni – e molto altro.

Per dimostrarlo, partirò da un esempio semplice e saliremo man mano.
Supponiamo, di nuovo, di voler trovare tutte le righe che contengono una certa parola.
Per cambiare, cerchiamo i riferimenti al personaggio che dà il titolo al libro, il Conte Dracula.
Ecco una riga che lo menziona.

```python
text = "I am Dracula; and I bid you welcome, Mr. Harker, to my house."
```

Ed ecco lo **schema** (*pattern*) che useremo per la ricerca.

```python
pattern = 'Dracula'
```

Un modulo chiamato `re` fornisce funzioni legate alle espressioni regolari.
Possiamo importarlo così e usare la funzione `search` per verificare se lo schema compare nel testo.

```python
import re

result = re.search(pattern, text)
result
```

Se lo schema compare nel testo, `search` restituisce un oggetto `Match` che contiene i risultati della ricerca.
Tra le altre informazioni, ha una variabile chiamata `string` che contiene il testo in cui si è cercato.

```python
result.string
```

Fornisce anche un metodo chiamato `group` che restituisce la parte del testo che corrisponde allo schema.

```python
result.group()
```

E fornisce un metodo chiamato `span` che restituisce gli indici del testo in cui lo schema inizia e finisce.

```python
result.span()
```

Se lo schema non compare nel testo, il valore restituito da `search` è `None`.

```python
result = re.search('Count', text)
print(result)
```

Quindi possiamo verificare se la ricerca ha avuto successo controllando se il risultato è `None`.

```python
result == None
```

Mettendo tutto insieme, ecco una funzione che scorre le righe del libro finché non ne trova una che corrisponde allo schema dato, e restituisce l'oggetto `Match`.

```python
def find_first(pattern):
    for line in open('pg345_cleaned.txt', encoding='utf-8'):
        result = re.search(pattern, line)
        if result != None:
            return result
```

Possiamo usarla per trovare la prima menzione di un personaggio.

```python
result = find_first('Harker')
result.string
```

Per questo esempio non era necessario usare le espressioni regolari – avremmo potuto fare la stessa cosa più facilmente con l'operatore `in`.
Ma le espressioni regolari possono fare cose che l'operatore `in` non può.

Per esempio, se lo schema include il carattere barra verticale, `'|'`, può corrispondere alla sequenza a sinistra oppure a quella a destra.
Supponiamo di voler trovare la prima menzione di Mina Murray nel libro, ma di non essere sicuri se venga chiamata per nome o per cognome.
Possiamo usare lo schema seguente, che corrisponde a entrambi i nomi.

```python
pattern = 'Mina|Murray'
result = find_first(pattern)
result.string
```

Possiamo usare uno schema come questo per vedere quante volte un personaggio viene menzionato con l'uno o l'altro nome.
Ecco una funzione che scorre il libro e conta il numero di righe che corrispondono allo schema dato.

```python
def count_matches(pattern):
    count = 0
    for line in open('pg345_cleaned.txt', encoding='utf-8'):
        result = re.search(pattern, line)
        if result != None:
            count += 1
    return count
```

Ora vediamo quante volte viene menzionata Mina.

```python
count_matches('Mina|Murray')
```

Il carattere speciale `'^'` corrisponde all'inizio di una stringa, quindi possiamo trovare una riga che inizia con un dato schema.

```python
result = find_first('^Dracula')
result.string
```

E il carattere speciale `'$'` corrisponde alla fine di una stringa, quindi possiamo trovare una riga che termina con un dato schema (ignorando l'a capo finale).

```python
result = find_first('Harker$')
result.string
```

## Sostituzione di stringhe

Bram Stoker era nato in Irlanda, e quando *Dracula* fu pubblicato, nel 1897, viveva in Inghilterra.
Quindi ci aspetteremmo che usasse la grafia britannica di parole come "centre" e "colour".
Per verificarlo, possiamo usare lo schema seguente, che corrisponde sia a "centre" sia alla grafia americana "center".

```python
pattern = 'cent(er|re)'
```

In questo schema, le parentesi racchiudono la parte dello schema a cui si applica la barra verticale.
Quindi questo schema corrisponde a una sequenza che inizia con `'cent'` e termina con `'er'` oppure con `'re'`.

```python
result = find_first(pattern)
result.string
```

Come previsto, ha usato la grafia britannica.

Possiamo anche verificare se ha usato la grafia britannica di "colour".
Lo schema seguente usa il carattere speciale `'?'`, che significa che il carattere precedente è facoltativo.

```python
pattern = 'colou?r'
```

Questo schema corrisponde sia a "colour" con la `'u'` sia a "color" senza.

```python
result = find_first(pattern)
line = result.string
line
```

Di nuovo, come previsto, ha usato la grafia britannica.

Supponiamo ora di voler produrre un'edizione del libro con la grafia americana.
Possiamo usare la funzione `sub` del modulo `re`, che esegue la **sostituzione di stringhe** (*string substitution*).

```python
re.sub(pattern, 'color', line)
```

Il primo argomento è lo schema che vogliamo trovare e sostituire, il secondo è ciò con cui vogliamo sostituirlo e il terzo è la stringa in cui vogliamo cercare.
Nel risultato puoi vedere che "colour" è stato sostituito da "color".

```python
# I used this function to search for lines to use as examples

def all_matches(pattern):
    for line in open('pg345_cleaned.txt', encoding='utf-8'):
        result = re.search(pattern, line)
        if result:
            print(line.strip())
```

```python
# Here's the pattern I used (which uses some features we haven't seen)

names = r'(?<!\.\s)[A-Z][a-zA-Z]+'

all_matches(names)
```

## Glossario

**sequenza** (*sequence*):
Una raccolta ordinata di valori in cui ogni valore è identificato da un indice intero.

**carattere** (*character*):
Un elemento di una stringa, comprese lettere, numeri e simboli.

**indice** (*index*):
Un valore intero usato per selezionare un elemento di una sequenza, come un carattere di una stringa. In Python gli indici partono da `0`.

**slice**:
Una parte di una stringa specificata da un intervallo di indici.

**stringa vuota** (*empty string*):
Una stringa che non contiene caratteri e ha lunghezza `0`.

**oggetto** (*object*):
Qualcosa a cui una variabile può fare riferimento. Un oggetto ha un tipo e un valore.

**immutabile** (*immutable*):
Se gli elementi di un oggetto non possono essere modificati, l'oggetto è immutabile.

**invocazione** (*invocation*):
Un'espressione – o parte di un'espressione – che chiama un metodo.

**espressione regolare** (*regular expression*):
Una sequenza di caratteri che definisce uno schema di ricerca.

**schema** (*pattern*):
Una regola che specifica i requisiti che una stringa deve soddisfare per costituire una corrispondenza (*match*).

**sostituzione di stringhe** (*string substitution*):
La sostituzione di una stringa, o di parte di una stringa, con un'altra stringa.

**comando di shell** (*shell command*):
Un'istruzione in un linguaggio di shell, cioè un linguaggio usato per interagire con un sistema operativo.
