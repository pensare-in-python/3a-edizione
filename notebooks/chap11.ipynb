## Le tuple sono come le liste

Una tupla è una sequenza di valori. I valori possono essere di qualsiasi tipo e sono indicizzati da interi, quindi le tuple somigliano molto alle liste.
La differenza importante è che le tuple sono immutabili.

Per creare una tupla, puoi scrivere un elenco di valori separati da virgole.

```python
t = 'l', 'u', 'p', 'i', 'n'
type(t)
```

Anche se non è necessario, è comune racchiudere le tuple tra parentesi tonde.

```python
t = ('l', 'u', 'p', 'i', 'n')
type(t)
```

Per creare una tupla con un solo elemento, devi includere una virgola finale.

```python
t1 = 'p',
type(t1)
```

Un singolo valore tra parentesi non è una tupla.

```python
t2 = ('p')
type(t2)
```

Un altro modo per creare una tupla è la funzione predefinita `tuple`. Senza
argomenti, crea una tupla vuota.

```python
t = tuple()
t
```

Se l'argomento è una sequenza (stringa, lista o tupla), il risultato è una
tupla con gli elementi della sequenza.

```python
t = tuple('lupin')
t
```

Poiché `tuple` è il nome di una funzione predefinita, dovresti evitare di usarlo come nome di variabile.

La maggior parte degli operatori delle liste funziona anche con le tuple.
Per esempio, l'operatore parentesi quadre accede a un elemento tramite indice.

```python
t[0]
```

E l'operatore di slice seleziona un intervallo di elementi.

```python
t[1:3]
```

L'operatore `+` concatena le tuple.

```python
tuple('lup') + ('i', 'n')
```

E l'operatore `*` duplica una tupla un dato numero di volte.

```python
tuple('spam') * 2 
```

La funzione `sorted` funziona con le tuple – ma il risultato è una lista, non una tupla.

```python
sorted(t)
```

Anche la funzione `reversed` funziona con le tuple.

```python
reversed(t)
```

Il risultato è un oggetto `reversed`, che possiamo convertire in una lista o in una tupla.

```python
tuple(reversed(t))
```

In base agli esempi visti finora, potrebbe sembrare che le tuple siano uguali alle liste.

## Ma le tuple sono immutabili

Se provi a modificare una tupla con l'operatore parentesi quadre, ottieni un `TypeError`.

```python
%%expect TypeError
t[0] = 'L'
```

E le tuple non hanno nessuno dei metodi che modificano le liste, come `append` e `remove`.

```python
%%expect AttributeError

t.remove('l')
```

Ricorda che un "attributo" è una variabile o un metodo associato a un oggetto – questo messaggio di errore significa che le tuple non hanno un metodo chiamato `remove`.

Poiché le tuple sono immutabili, sono hashabili, il che significa che possono essere usate come chiavi in un dizionario.
Per esempio, il dizionario seguente contiene due tuple come chiavi, che corrispondono a degli interi.

```python
d = {}
d[1, 2] = 3
d[3, 4] = 7
```

Possiamo cercare una tupla in un dizionario così:

```python
d[1, 2]
```

Oppure, se abbiamo una variabile che fa riferimento a una tupla, possiamo usarla come chiave.

```python
t = (3, 4)
d[t]
```

Le tuple possono comparire anche come valori in un dizionario.

```python
t = tuple('abc')
d = {'key': t}
d
```

## Assegnazione di tuple

Puoi mettere una tupla di variabili a sinistra di un'assegnazione e una tupla di valori a destra.

```python
a, b = 1, 2
```

I valori vengono assegnati alle variabili da sinistra a destra – in questo esempio, `a` riceve il valore `1` e `b` riceve il valore `2`.
Possiamo visualizzare i risultati così:

```python
a, b
```

Più in generale, se il lato sinistro di un'assegnazione è una tupla, il lato destro può essere qualsiasi tipo di sequenza – stringa, lista o tupla.
Per esempio, per dividere un indirizzo email in nome utente e dominio, potresti scrivere:

```python
email = 'monty@python.org'
username, domain = email.split('@')
```

Il valore restituito da `split` è una lista con due elementi – il primo elemento viene assegnato a `username`, il secondo a `domain`.

```python
username, domain
```

Il numero di variabili a sinistra e il numero di valori a
destra devono essere uguali – altrimenti ottieni un `ValueError`.

```python
%%expect ValueError
a, b = 1, 2, 3
```

L'assegnazione di tuple è utile se vuoi scambiare i valori di due variabili.
Con le assegnazioni normali devi usare una variabile temporanea, così:

```python
temp = a
a = b
b = temp
```

Funziona, ma con l'assegnazione di tuple possiamo fare la stessa cosa senza una variabile temporanea.

```python
a, b = b, a
```

Questo funziona perché tutte le espressioni a destra vengono valutate prima di qualsiasi assegnazione.

Possiamo usare l'assegnazione di tuple anche in un'istruzione `for`.
Per esempio, per scorrere gli elementi di un dizionario, possiamo usare il metodo `items`.

```python
d = {'one': 1, 'two': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
```

A ogni giro del ciclo, a `item` viene assegnata una tupla che contiene una chiave e il valore corrispondente.

Possiamo scrivere questo ciclo in modo più conciso, così:

```python
for key, value in d.items():
    print(key, '->', value)
```

A ogni giro del ciclo, una chiave e il valore corrispondente vengono assegnati direttamente a `key` e `value`.

## Tuple come valori di ritorno

A rigore, una funzione può restituire un solo valore, ma se il
valore è una tupla, l'effetto è lo stesso che restituire più valori.
Per esempio, se vuoi dividere due interi e calcolare il quoziente
e il resto, è inefficiente calcolare `x//y` e poi `x%y`. È
meglio calcolarli entrambi allo stesso tempo.

La funzione predefinita `divmod` riceve due argomenti e restituisce una tupla
di due valori, il quoziente e il resto.

```python
divmod(7, 3)
```

Possiamo usare l'assegnazione di tuple per memorizzare gli elementi della tupla in due variabili.

```python
quotient, remainder = divmod(7, 3)
quotient
```

```python
remainder
```

Ecco un esempio di funzione che restituisce una tupla.

```python
def min_max(t):
    return min(t), max(t)
```

`max` e `min` sono funzioni predefinite che trovano l'elemento più grande e quello più piccolo di una sequenza.
`min_max` li calcola entrambi e restituisce una tupla di due valori.

```python
min_max([2, 4, 1, 3])
```

Possiamo assegnare i risultati a delle variabili così:

```python
low, high = min_max([2, 4, 1, 3])
low, high
```

## Impacchettamento degli argomenti

Le funzioni possono ricevere un numero variabile di argomenti.
Un nome di parametro che inizia con l'operatore `*` **impacchetta** (*packs*) gli argomenti in una tupla.
Per esempio, la funzione seguente riceve un numero qualsiasi di argomenti e ne calcola la media aritmetica – cioè la loro somma divisa per il numero di argomenti.

```python
def mean(*args):
    return sum(args) / len(args)
```

Il parametro può avere qualsiasi nome, ma `args` è quello convenzionale.
Possiamo chiamare la funzione così:

```python
mean(1, 2, 3)
```

Se hai una sequenza di valori e vuoi passarli a una funzione come argomenti separati, puoi usare l'operatore `*` per **spacchettare** (*unpack*) la tupla.
Per esempio, `divmod` riceve esattamente due argomenti – se passi una tupla come parametro, ottieni un errore.

```python
%%expect TypeError
t = (7, 3)
divmod(t)
```

Anche se la tupla contiene due elementi, conta come un solo argomento.
Ma se spacchetti la tupla, viene trattata come due argomenti.

```python
divmod(*t)
```

Impacchettamento e spacchettamento possono essere utili se vuoi adattare il comportamento di una funzione esistente.
Per esempio, questa funzione riceve un numero qualsiasi di argomenti, rimuove il più basso e il più alto e calcola la media dei restanti.

```python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
```

Prima usa `min_max` per trovare l'elemento più basso e quello più alto.
Poi converte `args` in una lista per poter usare il metodo `remove`.
Infine spacchetta la lista, così gli elementi vengono passati a `mean` come argomenti separati, invece che come un'unica lista.

Ecco un esempio che ne mostra l'effetto.

```python
mean(1, 2, 3, 10)
```

```python
trimmed_mean(1, 2, 3, 10)
```

Questo tipo di media "troncata" si usa in alcuni sport con giudizio soggettivo – come i tuffi e la ginnastica – per ridurre l'effetto di un giudice il cui punteggio si discosta dagli altri.

## Zip

Le tuple sono utili per scorrere gli elementi di due sequenze ed eseguire operazioni sugli elementi corrispondenti.
Per esempio, supponiamo che due squadre giochino una serie di sette partite, e che registriamo i loro punteggi in due liste, una per squadra.

```python
scores1 = [1, 2, 4, 5, 1, 5, 2]
scores2 = [5, 5, 2, 2, 5, 2, 3]
```

Vediamo quante partite ha vinto ciascuna squadra.
Useremo `zip`, una funzione predefinita che riceve due o più sequenze e restituisce un **oggetto zip** (*zip object*), così chiamato perché accoppia gli elementi delle sequenze come i denti di una cerniera lampo (*zipper*).

```python
zip(scores1, scores2)
```

Possiamo usare l'oggetto zip per scorrere a coppie i valori delle sequenze.

```python
for pair in zip(scores1, scores2):
     print(pair)
```

A ogni giro del ciclo, a `pair` viene assegnata una tupla di punteggi.
Quindi possiamo assegnare i punteggi a delle variabili e contare le vittorie della prima squadra, così:

```python
wins = 0
for team1, team2 in zip(scores1, scores2):
    if team1 > team2:
        wins += 1
        
wins
```

Purtroppo la prima squadra ha vinto solo tre partite e ha perso la serie.

Se hai due liste e vuoi una lista di coppie, puoi usare `zip` e `list`.

```python
t = list(zip(scores1, scores2))
t
```

Il risultato è una lista di tuple, quindi possiamo ottenere il risultato dell'ultima partita così:

```python
t[-1]
```

Se hai una lista di chiavi e una lista di valori, puoi usare `zip` e `dict` per creare un dizionario.
Per esempio, ecco come possiamo creare un dizionario che fa corrispondere a ogni lettera la sua posizione nell'alfabeto.

```python
letters = 'abcdefghijklmnopqrstuvwxyz'
numbers = range(len(letters))
letter_map = dict(zip(letters, numbers))
```

Ora possiamo cercare una lettera e ottenere il suo indice nell'alfabeto.

```python
letter_map['a'], letter_map['z']
```

In questa mappatura, l'indice di `'a'` è `0` e l'indice di `'z'` è `25`.

Se devi scorrere gli elementi di una sequenza insieme ai loro indici, puoi usare la funzione predefinita `enumerate`.

```python
enumerate('abc')
```

Il risultato è un **oggetto enumerate** (*enumerate object*) che scorre una sequenza di coppie, in cui ogni coppia contiene un indice (a partire da 0) e un elemento della sequenza data.

```python
for index, element in enumerate('abc'):
    print(index, element)
```

## Confronto e ordinamento

Gli operatori relazionali funzionano con le tuple e con le altre sequenze.
Per esempio, se usi l'operatore `<` con le tuple, inizia confrontando il primo elemento di ciascuna sequenza.
Se sono uguali, passa alla coppia di elementi successiva, e così via, finché non trova una coppia di elementi diversi.

```python
(0, 1, 2) < (0, 3, 4)
```

Gli elementi successivi non vengono considerati – anche se sono molto grandi.

```python
(0, 1, 2000000) < (0, 3, 4)
```

Questo modo di confrontare le tuple è utile per ordinare una lista di tuple, o per trovare il minimo o il massimo.
Come esempio, troviamo la lettera più comune in una parola.
Nel capitolo precedente abbiamo scritto `value_counts`, che riceve una stringa e restituisce un dizionario che fa corrispondere a ogni lettera il numero di volte in cui compare.

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

Ecco il risultato per la stringa `'banana'`.

```python
counter = value_counts('banana')
counter
```

Con soli tre elementi, possiamo vedere facilmente che la lettera più frequente è `'a'`, che compare tre volte.
Ma se ci fossero più elementi, sarebbe utile ordinarli automaticamente.

Possiamo ottenere gli elementi di `counter` così.

```python
items = counter.items()
items
```

Il risultato è un oggetto `dict_items` che si comporta come una lista di tuple, quindi possiamo ordinarlo così.

```python
sorted(items)
```

Il comportamento predefinito è usare il primo elemento di ogni tupla per ordinare la lista, e il secondo elemento per risolvere i casi di parità.

Tuttavia, per trovare gli elementi con i conteggi più alti, vogliamo usare il secondo elemento per ordinare la lista.
Possiamo farlo scrivendo una funzione che riceve una tupla e ne restituisce il secondo elemento.

```python
def second_element(t):
    return t[1]
```

Poi possiamo passare questa funzione a `sorted` come argomento facoltativo chiamato `key`, che indica che questa funzione va usata per calcolare la **chiave di ordinamento** (*sort key*) di ciascun elemento.

```python
sorted_items = sorted(items, key=second_element)
sorted_items
```

La chiave di ordinamento determina l'ordine degli elementi nella lista.
La lettera con il conteggio più basso compare per prima, e quella con il conteggio più alto per ultima.
Quindi possiamo trovare la lettera più comune così.

```python
sorted_items[-1]
```

Se vogliamo solo il massimo, non dobbiamo ordinare la lista.
Possiamo usare `max`, che riceve anch'essa `key` come argomento facoltativo.

```python
max(items, key=second_element)
```

Per trovare la lettera con il conteggio più basso, potremmo usare `min` nello stesso modo.

## Invertire un dizionario

Supponiamo di voler invertire un dizionario per poter cercare un valore e ottenere la chiave corrispondente.
Per esempio, se hai un contatore di parole che fa corrispondere a ogni parola il numero di volte in cui compare, potresti creare un dizionario che faccia corrispondere a degli interi le parole che compaiono quel numero di volte.

Ma c'è un problema – le chiavi di un dizionario devono essere uniche, mentre i valori no. Per esempio, in un contatore di parole potrebbero esserci molte parole con lo stesso conteggio.

Quindi un modo per invertire un dizionario è creare un nuovo dizionario in cui i valori sono liste di chiavi dell'originale.
Come esempio, contiamo le lettere di `parrot`.

```python
d =  value_counts('parrot')
d
```

Se invertiamo questo dizionario, il risultato dovrebbe essere `{1: ['p', 'a', 'o', 't'], 2: ['r']}`, che indica che le lettere che compaiono una volta sono `'p'`, `'a'`, `'o'` e `'t'`, e la lettera che compare due volte è `'r'`.

La funzione seguente riceve un dizionario e restituisce il suo inverso come nuovo dizionario.

```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

L'istruzione `for` scorre le chiavi e i valori di `d`.
Se il valore non è già nel nuovo dizionario, viene aggiunto e associato a una lista che contiene un solo elemento.
Altrimenti viene aggiunto alla lista esistente.

Possiamo testarla così:

```python
invert_dict(d)
```

E otteniamo il risultato che ci aspettavamo.

Questo è il primo esempio che vediamo in cui i valori del dizionario sono liste.

## Glossario

**impacchettare** (*pack*):
Raccogliere più argomenti in una tupla.

**spacchettare** (*unpack*):
Trattare una tupla (o un'altra sequenza) come più argomenti.

**oggetto zip** (*zip object*):
Il risultato della chiamata della funzione predefinita `zip`; si può usare per scorrere una sequenza di tuple.

**oggetto enumerate** (*enumerate object*):
Il risultato della chiamata della funzione predefinita `enumerate`; si può usare per scorrere una sequenza di tuple.

**chiave di ordinamento** (*sort key*):
Un valore, o una funzione che calcola un valore, usato per ordinare gli elementi di una raccolta.

**struttura dati** (*data structure*):
Una raccolta di valori, organizzata per eseguire certe operazioni in modo efficiente.
