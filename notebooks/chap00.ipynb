# Prefazione

## A chi è rivolto questo libro?

Se vuoi imparare a programmare, sei nel posto giusto.
Python è uno dei migliori linguaggi di programmazione per principianti – ed è anche una delle competenze più richieste.

Sei arrivato anche nel momento giusto, perché imparare a programmare oggi è probabilmente più facile che mai.
Con assistenti virtuali come ChatGPT, non devi imparare da solo.
Nel corso del libro ti suggerirò dei modi per usare questi strumenti per accelerare il tuo apprendimento.

Questo libro è pensato soprattutto per chi non ha mai programmato e per chi ha un po' di esperienza con un altro linguaggio di programmazione.
Se hai già una solida esperienza in Python, potresti trovare i primi capitoli troppo lenti.

Una delle difficoltà dell'imparare a programmare è che devi imparare *due* linguaggi: uno è il linguaggio di programmazione stesso; l'altro è il vocabolario che usiamo per parlare dei programmi.
Se impari solo il linguaggio di programmazione, è probabile che tu abbia problemi quando devi interpretare un messaggio di errore, leggere la documentazione, parlare con un'altra persona o usare gli assistenti virtuali.
Se hai già programmato un po', ma non hai imparato anche questo secondo linguaggio, spero che questo libro ti sia utile.

## Obiettivi del libro

Scrivendo questo libro, ho cercato di fare attenzione al vocabolario.
Definisco ogni termine la prima volta che compare.
E alla fine di ogni capitolo c'è un glossario che riassume i termini introdotti.

Ho anche cercato di essere conciso.
Meno sforzo mentale richiede la lettura del libro, più energie avrai per programmare.

Ma non si può imparare a programmare solo leggendo un libro – bisogna esercitarsi.
Per questo motivo, alla fine di ogni capitolo ci sono degli esercizi in cui puoi mettere in pratica ciò che hai imparato.

Se leggi con attenzione e fai gli esercizi con costanza, farai progressi.
Ma ti avverto fin da ora – imparare a programmare non è facile, e anche per i programmatori esperti può essere frustrante.
Strada facendo, ti suggerirò delle strategie per aiutarti a scrivere programmi corretti e a correggere quelli sbagliati.

## Come orientarsi nel libro

Ogni capitolo di questo libro si basa sui precedenti, quindi dovresti leggerli in ordine e prenderti il tempo di fare gli esercizi prima di andare avanti.

I primi sei capitoli introducono gli elementi di base come l'aritmetica, le istruzioni condizionali e i cicli.
Introducono anche il concetto più importante della programmazione, le funzioni, e un modo potente di usarle, la ricorsione.

I capitoli 7 e 8 introducono le stringhe – che possono rappresentare lettere, parole e frasi – e gli algoritmi per lavorarci.

I capitoli dal 9 al 12 introducono le strutture dati fondamentali di Python – liste, dizionari e tuple – che sono strumenti potenti per scrivere programmi efficienti.
Il capitolo 12 presenta algoritmi per analizzare testo e generarne di nuovo in modo casuale.
Algoritmi come questi sono alla base dei modelli linguistici di grandi dimensioni (LLM), quindi questo capitolo ti darà un'idea di come funzionano strumenti come ChatGPT.

Il capitolo 13 riguarda i modi per conservare i dati in modo permanente – file e database.
Come esercizio, potrai scrivere un programma che esplora un file system e trova i file duplicati.

I capitoli dal 14 al 17 introducono la programmazione orientata agli oggetti (OOP), che è un modo di organizzare i programmi e i dati su cui lavorano.
Molte librerie Python sono scritte in stile orientato agli oggetti, quindi questi capitoli ti aiuteranno a capirne la struttura – e a definire i tuoi oggetti.

L'obiettivo di questo libro non è coprire l'intero linguaggio Python.
Mi concentro piuttosto su un sottoinsieme del linguaggio che offre le maggiori possibilità con il minor numero di concetti.
Tuttavia, Python ha molte funzionalità che puoi usare per risolvere in modo efficiente problemi comuni.
Il capitolo 18 presenta alcune di queste funzionalità.

Infine, il capitolo 19 contiene le mie riflessioni conclusive e alcuni suggerimenti per proseguire il tuo percorso nella programmazione.

## Cosa c'è di nuovo nella terza edizione?

I cambiamenti più grandi di questa edizione sono stati guidati da due nuove tecnologie – i notebook Jupyter e gli assistenti virtuali.

Ogni capitolo di questo libro è un notebook Jupyter, cioè un documento che contiene sia testo normale sia codice.
Per me, questo rende più facile scrivere il codice, testarlo e mantenerlo coerente con il testo.
Per te, significa che puoi eseguire il codice, modificarlo e fare gli esercizi, tutto in un unico posto.
Le istruzioni per lavorare con i notebook si trovano nel primo capitolo.

L'altro grande cambiamento è che ho aggiunto consigli su come lavorare con assistenti virtuali come ChatGPT e come usarli per accelerare l'apprendimento.
Quando la precedente edizione di questo libro fu pubblicata nel 2016, i predecessori di questi strumenti erano molto meno utili e la maggior parte delle persone non ne conosceva l'esistenza.
Oggi sono uno strumento standard nell'ingegneria del software, e penso che diventeranno uno strumento rivoluzionario per imparare a programmare – e per imparare anche molte altre cose.

Gli altri cambiamenti nel libro nascono dai miei rimpianti riguardo alla seconda edizione.

Il primo è che non avevo dato abbastanza importanza al testing del software.
Era già una mancanza deplorevole nel 2016, ma con l'arrivo degli assistenti virtuali il testing automatico è diventato ancora più importante.
Così questa edizione presenta gli strumenti di testing più usati in Python, `doctest` e `unittest`, e include diversi esercizi in cui puoi esercitarti a usarli.

L'altro mio rimpianto è che gli esercizi della seconda edizione erano disomogenei – alcuni erano più interessanti di altri e alcuni erano troppo difficili.
Il passaggio ai notebook Jupyter mi ha aiutato a sviluppare e testare una sequenza di esercizi più coinvolgente ed efficace.

In questa revisione, la sequenza degli argomenti è quasi la stessa, ma ho riorganizzato alcuni capitoli e ne ho fusi due brevi in uno solo.
Inoltre, ho ampliato la trattazione delle stringhe includendo le espressioni regolari.

Alcuni capitoli usano la grafica della tartaruga (turtle graphics).
Nelle edizioni precedenti usavo il modulo `turtle` di Python, ma purtroppo non funziona nei notebook Jupyter.
Così l'ho sostituito con un nuovo modulo turtle che dovrebbe essere più facile da usare.

Infine, ho riscritto una parte consistente del testo, chiarendo i punti che ne avevano bisogno e sfoltendo quelli in cui non ero stato conciso quanto avrei potuto.

Sono molto orgoglioso di questa nuova edizione – spero che ti piaccia!

## Per iniziare

Per la maggior parte dei linguaggi di programmazione, Python compreso, esistono molti strumenti per scrivere ed eseguire programmi.
Questi strumenti si chiamano ambienti di sviluppo integrati (IDE).
In generale, esistono due tipi di IDE:

* Alcuni lavorano con file che contengono codice, quindi offrono strumenti per modificare ed eseguire questi file.

* Altri lavorano principalmente con i notebook, che sono documenti contenenti testo e codice.

Per i principianti, consiglio di iniziare con un ambiente di sviluppo a notebook come Jupyter.

I notebook di questo libro sono disponibili in un repository online all'indirizzo <https://allendowney.github.io/ThinkPython>.

Ci sono due modi per usarli:

* Puoi scaricare i notebook ed eseguirli sul tuo computer. In questo caso devi installare Python e Jupyter, il che non è difficile, ma se vuoi imparare Python può essere frustrante passare molto tempo a installare software.

* Un'alternativa è eseguire i notebook su Colab, un ambiente Jupyter che funziona nel browser, così non devi installare nulla. Colab è gestito da Google ed è gratuito.

Se stai appena iniziando, ti consiglio vivamente di cominciare con Colab.
