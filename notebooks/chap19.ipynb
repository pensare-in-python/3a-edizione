# Riflessioni finali

Imparare a programmare non è facile, ma se sei arrivato fin qui, sei partito con il piede giusto.
Ora ho alcuni suggerimenti su come continuare a imparare e mettere in pratica ciò che hai imparato.

Questo libro vuole essere un'introduzione generale alla programmazione, quindi non ci siamo concentrati su applicazioni specifiche.
A seconda dei tuoi interessi, ci sono moltissimi campi in cui puoi applicare le tue nuove competenze.

Se ti interessa la Data Science, ci sono tre miei libri che potrebbero piacerti:

* *Think Stats: Exploratory Data Analysis*, O'Reilly Media, 2014.

* *Think Bayes: Bayesian Statistics in Python*, O'Reilly Media, 2021.

* *Think DSP: Digital Signal Processing in Python*, O'Reilly Media, 2016.

Se ti interessano la modellazione fisica e i sistemi complessi, potrebbero piacerti:

* *Modeling and Simulation in Python: An Introduction for Scientists and Engineers*, No Starch Press, 2023.

* *Think Complexity: Complexity Science and Computational Modeling*, O'Reilly Media, 2018.

Questi usano NumPy, SciPy, pandas e altre librerie Python per la data science e il calcolo scientifico.

Questo libro cerca un equilibrio tra i principi generali della programmazione e i dettagli di Python.
Di conseguenza, non include tutte le funzionalità del linguaggio Python.
Per saperne di più su Python, e ricevere buoni consigli su come usarlo, ti consiglio *Fluent Python: Clear, Concise, and Effective Programming*, seconda edizione, di Luciano Ramalho, O'Reilly Media, 2022.

Dopo un'introduzione alla programmazione, un passo successivo comune è studiare strutture dati e algoritmi.
Ho un lavoro in corso su questo argomento, intitolato *Data Structures and Information Retrieval in Python*.
Una versione elettronica gratuita è disponibile da Green Tea Press su <https://greenteapress.com>.

Man mano che lavori a programmi più complessi, incontrerai nuove difficoltà.
Potrebbe esserti utile rileggere le sezioni di questo libro dedicate al debugging.
In particolare, ricorda le sei "R" del debugging del [capitolo 12](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap12.ipynb) (in inglese *reading, running, ruminating, rubber-ducking, retreating, resting*): leggere, eseguire, rimuginare, parlare alla paperella di gomma, ritirarsi e riposare.

Questo libro suggerisce strumenti utili per il debugging, tra cui le funzioni `print` e `repr`, la funzione `structshape` del [capitolo 11](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap11.ipynb) – e le funzioni predefinite `isinstance`, `hasattr` e `vars` del [capitolo 14](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap14.ipynb).

Suggerisce anche strumenti per testare i programmi, tra cui l'istruzione `assert`, il modulo `doctest` e il modulo `unittest`.
Includere test nei tuoi programmi è uno dei modi migliori per prevenire e individuare gli errori, e risparmiare tempo di debugging.

Ma il miglior debugging è quello che non devi fare.
Se usi un procedimento di sviluppo incrementale come quello descritto nel [capitolo 6](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap06.ipynb) – e testi man mano – farai meno errori e, quando li farai, li troverai più rapidamente.
Ricorda anche l'incapsulamento e la generalizzazione del [capitolo 4](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap04.ipynb), particolarmente utili quando sviluppi codice nei notebook Jupyter.

In tutto il libro ti ho suggerito modi per usare gli assistenti virtuali per aiutarti a imparare, programmare e fare debugging.
Spero che tu stia trovando utili questi strumenti.

Oltre ad assistenti virtuali come ChatGPT, potresti voler usare anche uno strumento come Copilot, che completa automaticamente il codice mentre scrivi.
All'inizio non ho consigliato di usare questi strumenti, perché possono essere disorientanti per i principianti.
Ma ora potresti volerli esplorare.

Usare in modo efficace gli strumenti di IA richiede un po' di sperimentazione e di riflessione per trovare un modo di lavorare adatto a te.
Se trovi scomodo copiare codice da ChatGPT a Jupyter, potresti preferire qualcosa come Copilot.
Ma il lavoro mentale che fai per comporre un prompt e interpretare la risposta può essere prezioso quanto il codice generato dallo strumento, nello stesso spirito del debugging con la paperella di gomma.

Man mano che acquisisci esperienza di programmazione, potresti voler esplorare altri ambienti di sviluppo.
Penso che i notebook Jupyter siano un buon punto di partenza, ma sono relativamente recenti e non diffusi quanto gli ambienti di sviluppo integrati (IDE) tradizionali.
Per Python, tra gli IDE più popolari ci sono PyCharm e Spyder – e Thonny, spesso consigliato ai principianti.
Altri IDE, come Visual Studio Code ed Eclipse, funzionano anche con altri linguaggi di programmazione.
Oppure, come alternativa più semplice, puoi scrivere programmi Python con qualsiasi editor di testo che ti piaccia.

Mentre prosegui il tuo percorso nella programmazione, non devi andare da solo!
Se vivi in una città o vicino a una città, è probabile che ci sia un gruppo di utenti Python a cui puoi unirti.
Questi gruppi di solito sono accoglienti con i principianti, quindi non aver paura.
Se non c'è un gruppo vicino a te, potresti riuscire a partecipare agli eventi a distanza.
Inoltre, tieni d'occhio le conferenze Python regionali.

Uno dei modi migliori per migliorare le tue capacità di programmazione è imparare un altro linguaggio.
Se ti interessano la statistica e la data science, potresti voler imparare R.
Ma ti consiglio in particolare di imparare un linguaggio funzionale come Racket o Elixir.
La programmazione funzionale richiede un modo diverso di pensare, che cambia il modo in cui pensi ai programmi.

Buona fortuna!
