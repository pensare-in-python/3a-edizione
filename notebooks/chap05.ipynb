## Divisione intera e modulo

Ricorda che l'operatore di divisione intera, `//`, divide due numeri e arrotonda
per difetto a un intero.
Per esempio, supponiamo che la durata di un film sia di 105 minuti.
Potresti voler sapere quanto dura in ore.
La divisione normale restituisce un numero in virgola mobile:

```python
minutes = 105
minutes / 60
```

Ma di solito non scriviamo le ore con i decimali.
La divisione intera restituisce il numero intero di ore, arrotondando per difetto:

```python
minutes = 105
hours = minutes // 60
hours
```

Per ottenere il resto, potresti sottrarre un'ora espressa in minuti:

```python
remainder = minutes - hours * 60
remainder
```

Oppure potresti usare l'**operatore modulo** (*modulus operator*), `%`, che divide due numeri e restituisce il resto.

```python
remainder = minutes % 60
remainder
```

L'operatore modulo è più utile di quanto possa sembrare.
Per esempio, può verificare se un numero è divisibile per un altro – se `x % y` è zero, allora `x` è divisibile per `y`.

Inoltre, può estrarre la cifra o le cifre più a destra di un numero.
Per esempio, `x % 10` dà la cifra più a destra di `x` (in base 10).
Allo stesso modo, `x % 100` dà le ultime due cifre.

```python
x = 123
x % 10
```

```python
x % 100
```

Infine, l'operatore modulo può fare l'"aritmetica dell'orologio".
Per esempio, se un evento inizia alle 11 di mattina e dura tre ore, possiamo usare l'operatore modulo per calcolare a che ora finisce.

```python
start = 11
duration = 3
end = (start + duration) % 12
end
```

L'evento finirebbe alle 2 del pomeriggio.

## Operatori logici

Per combinare valori booleani in espressioni, possiamo usare gli **operatori logici** (*logical operators*).
I più comuni sono `and`, `or` e `not`.
Il significato di questi operatori è simile a quello delle parole inglesi corrispondenti ("e", "o", "non").
Per esempio, il valore dell'espressione seguente è `True` solo se `x` è maggiore di `0` *e* minore di `10`.

Infine, l'operatore `not` nega un'espressione booleana, quindi l'espressione seguente è `True` se `x > y` è `False`.

A rigore, gli operandi di un operatore logico dovrebbero essere espressioni booleane, ma Python non è molto rigido.

## Istruzioni if

Per scrivere programmi utili, ci serve quasi sempre la possibilità di
verificare delle condizioni e cambiare di conseguenza il comportamento del programma.
Le **istruzioni condizionali** (*conditional statements*) ci danno questa possibilità. La forma più semplice è
l'istruzione `if`:

```python
if x > 0:
    print('x is positive')
```

`if` è una parola chiave di Python.
Le istruzioni `if` hanno la stessa struttura delle definizioni di funzione: un'intestazione
seguita da un'istruzione indentata o da una sequenza di istruzioni chiamata **blocco** (*block*).

L'espressione booleana dopo `if` si chiama **condizione** (*condition*).
Se è vera, le istruzioni del blocco indentato vengono eseguite. Altrimenti no.

Non c'è limite al numero di istruzioni che possono comparire nel blocco, ma ce ne deve essere almeno una.
A volte è utile avere un blocco che non fa niente – di solito come segnaposto per codice che non hai ancora scritto.
In quel caso puoi usare l'istruzione `pass`, che non fa nulla.

```python
if x < 0:
    pass          # TODO: need to handle negative values!
```

La parola `TODO` ("da fare") in un commento è un promemoria convenzionale che indica che c'è qualcosa da fare in seguito.

## La clausola `else`

Un'istruzione `if` può avere una seconda parte, chiamata clausola `else`.
La sintassi è questa:

```python
if x % 2 == 0:
    print('x is even')
else:
    print('x is odd')
```

Se la condizione è vera, viene eseguita la prima istruzione indentata; altrimenti viene eseguita la seconda.

In questo esempio, se `x` è pari, il resto della divisione di `x` per `2` è `0`, quindi la condizione è vera e il programma visualizza `x is even` ("x è pari").
Se `x` è dispari, il resto è `1`, quindi la condizione
è falsa e il programma visualizza `x is odd` ("x è dispari").

Poiché la condizione deve essere vera o falsa, verrà eseguita esattamente una delle alternative.
Le alternative si chiamano **rami** (*branches*).

## Condizionali concatenati

A volte le possibilità sono più di due e ci servono più di due rami.
Un modo per esprimere un calcolo del genere è un **condizionale concatenato** (*chained conditional*), che include una clausola `elif`.

```python
if x < y:
    print('x is less than y')
elif x > y:
    print('x is greater than y')
else:
    print('x and y are equal')
```

`elif` è un'abbreviazione di "else if" ("altrimenti se").
Non c'è limite al numero di clausole `elif`.
Se c'è una clausola `else`, deve trovarsi alla fine, ma non è obbligatorio
che ci sia.

Ogni condizione viene verificata in ordine.
Se la prima è falsa, si verifica la successiva, e così via.
Se una di esse è vera, il ramo corrispondente viene eseguito e l'istruzione `if` termina.
Anche se più di una condizione è vera, viene eseguito solo il primo ramo vero.

## Condizionali annidati

Un condizionale può anche essere annidato all'interno di un altro.
Avremmo potuto scrivere l'esempio della sezione precedente così:

```python
if x == y:
    print('x and y are equal')
else:
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
```

L'istruzione `if` esterna contiene due rami.
Il primo ramo contiene un'istruzione semplice. Il secondo ramo contiene un'altra istruzione `if`, che ha a sua volta due rami.
Questi due rami sono entrambi istruzioni semplici, anche se avrebbero potuto essere a loro volta istruzioni condizionali.

Anche se l'indentazione delle istruzioni rende evidente la struttura, i **condizionali annidati** (*nested conditionals*) possono essere difficili da leggere.
Ti suggerisco di evitarli quando puoi.

Gli operatori logici offrono spesso un modo per semplificare le istruzioni condizionali annidate.
Ecco un esempio con un condizionale annidato.

```python
if 0 < x:
    if x < 10:
        print('x is a positive single-digit number.')
```

L'istruzione `print` viene eseguita solo se superiamo entrambi i condizionali, quindi otteniamo lo stesso effetto con l'operatore `and`.

```python
if 0 < x and x < 10:
    print('x is a positive single-digit number.')
```

Per questo tipo di condizione, Python offre un'opzione più concisa:

```python
if 0 < x < 10:
    print('x is a positive single-digit number.')
```

## Ricorsione

È lecito che una funzione chiami sé stessa.
Potrebbe non essere ovvio perché sia una buona cosa, ma si rivela una delle cose più magiche che un programma possa fare.
Ecco un esempio.

```python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)
```

Se `n` è 0 o negativo, `countdown` stampa la parola "Blastoff!" ("Decollo!"). Altrimenti
stampa `n` e poi chiama sé stessa, passando `n-1` come argomento.

Ecco cosa succede quando chiamiamo questa funzione con l'argomento `3`.

```python
countdown(3)
```

L'esecuzione di `countdown` inizia con `n=3`, e poiché `n` è maggiore
di `0`, visualizza `3` e poi chiama sé stessa\...

> L'esecuzione di `countdown` inizia con `n=2`, e poiché `n` è
> maggiore di `0`, visualizza `2` e poi chiama sé stessa\...
>
> > L'esecuzione di `countdown` inizia con `n=1`, e poiché `n` è
> > maggiore di `0`, visualizza `1` e poi chiama sé stessa\...
> >
> > > L'esecuzione di `countdown` inizia con `n=0`, e poiché `n` non è
> > > maggiore di `0`, visualizza "Blastoff!" e ritorna.
> >
> > La `countdown` che ha ricevuto `n=1` ritorna.
>
> La `countdown` che ha ricevuto `n=2` ritorna.

La `countdown` che ha ricevuto `n=3` ritorna.

Una funzione che chiama sé stessa è **ricorsiva** (*recursive*).
Come altro esempio, possiamo scrivere una funzione che stampa una stringa `n` volte.

```python
def print_n_times(string, n):
    if n > 0:
        print(string)
        print_n_times(string, n-1)
```

Se `n` è positivo, `print_n_times` visualizza il valore di `string` e poi chiama sé stessa, passando come argomenti `string` e `n-1`.

Se `n` è `0` o negativo, la condizione è falsa e `print_n_times` non fa nulla.

Ecco come funziona.

```python
print_n_times('Spam ', 4)
```

Per esempi semplici come questo, probabilmente è più facile usare un ciclo `for`.
Ma più avanti vedremo esempi difficili da scrivere con un ciclo `for`
e facili da scrivere con la ricorsione, quindi è bene iniziare presto.

## Ricorsione infinita

Se una ricorsione non raggiunge mai un caso base, continua a fare chiamate ricorsive
all'infinito, e il programma non termina mai. Questo fenomeno si chiama
**ricorsione infinita** (*infinite recursion*), e in genere non è una buona idea.
Ecco una funzione minima con una ricorsione infinita.

```python
def recurse():
    recurse()
```

Ogni volta che `recurse` viene chiamata, chiama sé stessa, creando un altro frame.
In Python c'è un limite al numero di frame che possono trovarsi contemporaneamente sullo stack.
Se un programma supera il limite, si verifica un errore di runtime.

```python
%xmode Context
```

```python
%%expect RecursionError

recurse()
```

Il traceback indica che c'erano quasi 3000 frame sullo stack quando si è verificato l'errore.

## Glossario

**ricorsione** (*recursion*):
Il processo di chiamare la funzione attualmente in esecuzione.

**operatore modulo** (*modulus operator*):
Un operatore, `%`, che lavora sugli interi e restituisce il resto della divisione di un numero per un altro.

**espressione booleana** (*boolean expression*):
Un'espressione il cui valore è `True` oppure `False`.

**operatore relazionale** (*relational operator*):
Uno degli operatori che confrontano i propri operandi: `==`, `!=`, `>`, `<`, `>=` e `<=`.

**operatore logico** (*logical operator*):
Uno degli operatori che combinano espressioni booleane, tra cui `and`, `or` e `not`.

**istruzione condizionale** (*conditional statement*):
Un'istruzione che controlla il flusso di esecuzione in base a una condizione.

**condizione** (*condition*):
L'espressione booleana di un'istruzione condizionale che determina quale ramo viene eseguito.

**blocco** (*block*):
Una o più istruzioni indentate per indicare che fanno parte di un'altra istruzione.

**ramo** (*branch*):
Una delle sequenze alternative di istruzioni in un'istruzione condizionale.

**condizionale concatenato** (*chained conditional*):
Un'istruzione condizionale con una serie di rami alternativi.

**condizionale annidato** (*nested conditional*):
Un'istruzione condizionale che compare in uno dei rami di un'altra istruzione condizionale.

**ricorsiva** (*recursive*):
Una funzione che chiama sé stessa è ricorsiva.

**caso base** (*base case*):
Un ramo condizionale di una funzione ricorsiva che non fa una chiamata ricorsiva.

**ricorsione infinita** (*infinite recursion*):
Una ricorsione che non ha un caso base, o che non lo raggiunge mai.
Prima o poi, una ricorsione infinita provoca un errore di runtime.

**a capo** (*newline*):
Un carattere che crea un'interruzione di riga tra due parti di una stringa.
