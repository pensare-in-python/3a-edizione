# Ereditarietà

[Esegui questo notebook su Colab](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap17.ipynb)

La funzionalità del linguaggio più spesso associata alla programmazione orientata agli oggetti è l'**ereditarietà** (*inheritance*).
L'ereditarietà è la possibilità di definire una nuova classe che è una versione modificata di una classe esistente.
In questo capitolo illustro l'ereditarietà usando classi che rappresentano carte da gioco, mazzi di carte e mani di poker.
Se non giochi a poker, non preoccuparti – ti dirò quello che c'è da sapere.

## Rappresentare le carte

In un mazzo standard ci sono 52 carte da gioco – ognuna appartiene a uno dei quattro semi e a uno dei tredici valori.
I semi sono Picche (*Spades*), Cuori (*Hearts*), Quadri (*Diamonds*) e Fiori (*Clubs*).
I valori sono Asso, 2, 3, 4, 5, 6, 7, 8, 9, 10, Fante (*Jack*), Donna (*Queen*) e Re (*King*).
A seconda del gioco, un Asso può essere più alto del Re o più basso del 2.

Se vogliamo definire un nuovo oggetto che rappresenti una carta da gioco, è ovvio quali dovrebbero essere gli attributi: `rank` (valore) e `suit` (seme).
È meno ovvio di che tipo dovrebbero essere gli attributi.
Una possibilità è usare stringhe come `'Spade'` per i semi e `'Queen'` per i valori.
Un problema di questa implementazione è che non sarebbe facile confrontare le carte per vedere quale ha il valore o il seme più alto.

Un'alternativa è usare interi per **codificare** (*encode*) valori e semi.
In questo contesto, "codificare" significa che definiremo una corrispondenza tra numeri e semi, o tra numeri e valori.
Questo tipo di codifica non ha lo scopo di essere segreta (quella sarebbe "cifratura").

Per esempio, questa tabella mostra i semi e i corrispondenti codici interi:


| Seme | Codice |
| --- | --- |
|  Picche (Spades)     |   3  |
|  Cuori (Hearts)     |   2  |
|  Quadri (Diamonds)   |   1  |
|  Fiori (Clubs)      |   0  |

Con questa codifica, possiamo confrontare i semi confrontandone i codici.

Per codificare i valori, useremo l'intero `2` per rappresentare il valore `2`, `3` per rappresentare `3`, e così via fino a `10`.
La tabella seguente mostra i codici delle figure.


| Valore | Codice |
| --- | --- |
|  Fante (Jack)     |   11  |
|  Donna (Queen)   |   12  |
|  Re (King)      |   13  |

E possiamo usare `1` oppure `14` per rappresentare un Asso, a seconda che vogliamo considerarlo più basso o più alto degli altri valori.

Per rappresentare queste codifiche, useremo due liste di stringhe, una con i nomi dei semi e l'altra con i nomi dei valori.

Ecco la definizione di una classe che rappresenta una carta da gioco, con queste liste di stringhe come **variabili di classe** (*class variables*), cioè variabili definite all'interno di una definizione di classe, ma non all'interno di un metodo.

```python
class Card:
    """Represents a standard playing card."""

    suit_names = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
    rank_names = [None, 'Ace', '2', '3', '4', '5', '6', '7', 
                  '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']
```

Il primo elemento di `rank_names` è `None` perché non esiste una carta con valore zero. Includendo `None` come segnaposto, otteniamo una lista con la comoda proprietà che l'indice `2` corrisponde alla stringa `'2'`, e così via.

Le variabili di classe sono associate alla classe, e non a un'istanza della classe, quindi possiamo accedervi così.

```python
Card.suit_names
```

Possiamo usare `suit_names` per cercare un seme e ottenere la stringa corrispondente.

```python
Card.suit_names[0]
```

E `rank_names` per cercare un valore.

```python
Card.rank_names[11]
```

## Attributi delle carte

Ecco un metodo `__init__` per la classe `Card` – riceve `suit` e `rank` come parametri e li assegna ad attributi con gli stessi nomi.

```python
%%add_method_to Card

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
```

Ora possiamo creare un oggetto `Card` così.

```python
queen = Card(1, 12)
```

Possiamo usare la nuova istanza per accedere agli attributi.

```python
queen.suit, queen.rank
```

È lecito anche usare l'istanza per accedere alle variabili di classe.

```python
queen.suit_names
```

Ma se usi la classe, è più chiaro che si tratta di variabili di classe e non di attributi.

```python
%%add_method_to Card

    def __str__(self):
        rank_name = Card.rank_names[self.rank]
        suit_name = Card.suit_names[self.suit]
        return f'{rank_name} of {suit_name}' 
```

Quando stampiamo una `Card`, Python chiama il metodo `__str__` per ottenere una rappresentazione leggibile della carta.

```python
print(queen)
```

## Confrontare le carte

Supponiamo di creare un secondo oggetto `Card` con lo stesso seme e lo stesso valore.

```python
queen2 = Card(1, 12)
print(queen2)
```

Se usiamo l'operatore `==` per confrontarli, verifica se `queen` e `queen2` fanno riferimento allo stesso oggetto.

```python
queen == queen2
```

Non è così, quindi restituisce `False`.
Possiamo cambiare questo comportamento definendo il metodo speciale `__eq__`.

```python
%%add_method_to Card

    def __eq__(self, other):
        return self.suit == other.suit and self.rank == other.rank
```

`__eq__` riceve due oggetti `Card` come parametri e restituisce `True` se hanno lo stesso seme e lo stesso valore, anche se non sono lo stesso oggetto.
In altre parole, verifica se sono equivalenti, anche se non sono identici.

Quando usiamo l'operatore `==` con oggetti `Card`, Python chiama il metodo `__eq__`.

```python
queen == queen2
```

Come secondo test, creiamo una carta con lo stesso seme e un valore diverso.

```python
six = Card(1, 6)
print(six)
```

Possiamo confermare che `queen` e `six` non sono equivalenti.

```python
queen == six
```

Se usiamo l'operatore `!=`, Python invoca un metodo speciale chiamato `__ne__`, se esiste.
Altrimenti invoca `__eq__` e inverte il risultato – quindi se `__eq__` restituisce `True`, il risultato dell'operatore `!=` è `False`.

```python
queen != queen2
```

```python
queen != six
```

Supponiamo ora di voler confrontare due carte per vedere quale è più alta.
Se usiamo uno degli operatori relazionali, otteniamo un `TypeError`.

```python
%%expect TypeError

queen < queen2
```

Per cambiare il comportamento dell'operatore `<`, possiamo definire un metodo speciale chiamato `__lt__`, abbreviazione di "less than" (minore di).
Ai fini di questo esempio, assumiamo che il seme sia più importante del valore – quindi tutte le Picche battono tutti i Cuori, che battono tutti i Quadri, e così via.
Se due carte hanno lo stesso seme, vince quella con il valore più alto.

Per implementare questa logica, useremo il metodo seguente, che restituisce una tupla contenente il seme e il valore di una carta, in quest'ordine.

```python
%%add_method_to Card

    def to_tuple(self):
        return (self.suit, self.rank)
```

Possiamo usare questo metodo per scrivere `__lt__`.

```python
%%add_method_to Card

    def __lt__(self, other):
        return self.to_tuple() < other.to_tuple()
```

Il confronto tra tuple confronta i primi elementi di ciascuna tupla, che rappresentano i semi.
Se sono uguali, confronta i secondi elementi, che rappresentano i valori.

Ora, se usiamo l'operatore `<`, questo invoca il metodo `__lt__`.

```python
six < queen
```

Se usiamo l'operatore `>`, invoca un metodo speciale chiamato `__gt__`, se esiste.
Altrimenti invoca `__lt__` con gli argomenti in ordine inverso.

```python
queen < queen2
```

```python
queen > queen2
```

Infine, se usiamo l'operatore `<=`, invoca un metodo speciale chiamato `__le__`.

```python
%%add_method_to Card

    def __le__(self, other):
        return self.to_tuple() <= other.to_tuple()
```

Quindi possiamo verificare se una carta è minore o uguale a un'altra.

```python
queen <= queen2
```

```python
queen <= six
```

Se usiamo l'operatore `>=`, usa `__ge__` se esiste. Altrimenti invoca `__le__` con gli argomenti in ordine inverso.

```python
queen >= six
```

Così come li abbiamo definiti, questi metodi sono completi, nel senso che possiamo confrontare due oggetti `Card` qualsiasi, e coerenti, nel senso che i risultati di operatori diversi non si contraddicono.
Con queste due proprietà, possiamo dire che gli oggetti `Card` sono **totalmente ordinati** (*totally ordered*).
E questo significa, come vedremo presto, che si possono ordinare.

## Mazzi

Ora che abbiamo oggetti che rappresentano carte, definiamo oggetti che rappresentano mazzi.
Quella che segue è una definizione di classe per `Deck` (mazzo) con
un metodo `__init__` che riceve una lista di oggetti `Card` come parametro e la assegna a un attributo chiamato `cards`.

```python
class Deck:

    def __init__(self, cards):
        self.cards = cards
```

Per creare una lista che contiene le 52 carte di un mazzo standard, useremo il seguente metodo statico.

```python
%%add_method_to Deck

    def make_cards():
        cards = []
        for suit in range(4):
            for rank in range(2, 15):
                card = Card(suit, rank)
                cards.append(card)
        return cards
```

In `make_cards`, il ciclo esterno enumera i semi da `0` a `3`.
Il ciclo interno enumera i valori da `2` a `14` – dove `14` rappresenta un Asso che batte il Re.
Ogni iterazione crea una nuova `Card` con il seme e il valore correnti e la aggiunge a `cards`.

Ecco come creiamo una lista di carte e un oggetto `Deck` che la contiene.

```python
cards = Deck.make_cards()
deck = Deck(cards)
len(deck.cards)
```

Contiene 52 carte, come previsto.

## Stampare il mazzo

Ecco un metodo `__str__` per `Deck`.

```python
%%add_method_to Deck

    def __str__(self):
        res = []
        for card in self.cards:
            res.append(str(card))
        return '\n'.join(res)
```

Questo metodo mostra un modo efficiente di accumulare una stringa grande – costruendo una lista di stringhe e poi usando il metodo delle stringhe `join`.

Testeremo questo metodo con un mazzo che contiene solo due carte.

```python
small_deck = Deck([queen, six])
```

Se chiamiamo `str`, questa invoca `__str__`.

```python
str(small_deck)
```

Quando Jupyter visualizza una stringa, mostra la forma "rappresentativa" della stringa, che rappresenta l'a capo con la sequenza `\n`.

Tuttavia, se stampiamo il risultato, Jupyter mostra la forma "stampabile" della stringa, che stampa l'a capo come spaziatura.

```python
print(small_deck)
```

Quindi le carte compaiono su righe separate.

## Aggiungere, togliere, mescolare e ordinare

e la restituisce. Il metodo delle liste `pop` offre un modo comodo per

```python
%%add_method_to Deck

    def take_card(self):
        return self.cards.pop()
```

```python
len(deck.cards)
```

Per aggiungere una carta, possiamo usare il metodo delle liste `append`.

```python
%%add_method_to Deck

    def put_card(self, card):
        self.cards.append(card)
```

Come esempio, possiamo rimettere nel mazzo la carta che abbiamo appena tolto.

```python
deck.put_card(card)
len(deck.cards)
```

## Genitori e figli

L'ereditarietà è la possibilità di definire una nuova classe che è una versione modificata di una classe esistente.
Come esempio, diciamo che vogliamo una classe che rappresenti una "mano", cioè le carte tenute da un giocatore.

* Una mano è simile a un mazzo – entrambi sono formati da una raccolta di carte, ed entrambi richiedono operazioni come aggiungere e togliere carte.

* Una mano è anche diversa da un mazzo – ci sono operazioni che vogliamo per le mani che non hanno senso per un mazzo. Per esempio, nel poker potremmo confrontare due mani per vedere quale vince. Nel bridge, potremmo calcolare un punteggio per una mano per fare una dichiarazione.

Questa relazione tra classi – in cui una è una versione specializzata dell'altra – si presta all'ereditarietà.

Per definire una nuova classe basata su una classe esistente, mettiamo il nome della classe esistente tra parentesi.

```python
class Hand(Deck):
    """Represents a hand of playing cards."""
```

Questa definizione indica che `Hand` eredita da `Deck`, il che significa che gli oggetti `Hand` possono accedere ai metodi definiti in `Deck`, come `take_card` e `put_card`.

`Hand` eredita anche `__init__` da `Deck`, ma se definiamo `__init__` nella classe `Hand`, questo sovrascrive quello della classe `Deck`.

```python
%%add_method_to Hand

    def __init__(self, label=''):
        self.label = label
        self.cards = []
```

Questa versione di `__init__` riceve come parametro una stringa facoltativa e parte sempre con una lista di carte vuota.
Quando creiamo una `Hand`, Python invoca questo metodo, non quello di `Deck` – come possiamo confermare verificando che il risultato abbia un attributo `label`.

```python
hand = Hand('player 1')
hand.label
```

Per distribuire una carta, possiamo usare `take_card` per togliere una carta da un `Deck`, e `put_card` per aggiungere la carta a una `Hand`.

```python
deck = Deck(cards)
card = deck.take_card()
hand.put_card(card)
print(hand)
```

Incapsuliamo questo codice in un metodo di `Deck` chiamato `move_cards`.

```python
%%add_method_to Deck

    def move_cards(self, other, num):
        for i in range(num):
            card = self.take_card()
            other.put_card(card)
```

Questo metodo è polimorfico – cioè funziona con più di un tipo: `self` e `other` possono essere sia una `Hand` sia un `Deck`.
Quindi possiamo usare questo metodo per distribuire una carta da un `Deck` a una `Hand`, da una `Hand` a un'altra, o da una `Hand` di nuovo a un `Deck`.

Quando una nuova classe eredita da una esistente, quella esistente si chiama **genitore** (*parent*) e la nuova classe si chiama **figlia** (*child*). In generale:

* Le istanze della classe figlia dovrebbero avere tutti gli attributi della classe genitore, ma possono avere attributi aggiuntivi.

* La classe figlia dovrebbe avere tutti i metodi della classe genitore, ma può avere metodi aggiuntivi.

* Se una classe figlia sovrascrive un metodo della classe genitore, il nuovo metodo dovrebbe ricevere gli stessi parametri e restituire un risultato compatibile.

Questo insieme di regole si chiama "principio di sostituzione di Liskov", dal nome dell'informatica Barbara Liskov.

Se segui queste regole, qualsiasi funzione o metodo progettato per lavorare con un'istanza di una classe genitore, come un `Deck`, funzionerà anche con le istanze di una classe figlia, come `Hand`.
Se violi queste regole, il tuo codice crollerà come un castello di carte (scusa).

## Specializzazione

Creiamo una classe chiamata `BridgeHand` che rappresenta una mano di bridge – un gioco di carte molto diffuso.
Erediteremo da `Hand` e aggiungeremo un nuovo metodo chiamato `high_card_point_count` che valuta una mano usando il metodo dei "punti onori" (*high card points*), che somma i punti delle carte alte nella mano.

Ecco una definizione di classe che contiene come variabile di classe un dizionario che fa corrispondere ai nomi delle carte il loro valore in punti.

```python
class BridgeHand(Hand):
    """Represents a bridge hand."""

    hcp_dict = {
        'Ace': 4,
        'King': 3,
        'Queen': 2,
        'Jack': 1,
    }
```

Dato il valore di una carta, come `12`, possiamo usare `Card.rank_names` per ottenere la rappresentazione in stringa del valore, e poi usare `hcp_dict` per ottenerne il punteggio.

```python
rank = 12
rank_name = Card.rank_names[rank]
score = BridgeHand.hcp_dict.get(rank_name, 0)
rank_name, score
```

Il metodo seguente scorre le carte di una `BridgeHand` e ne somma i punteggi.

```python
%%add_method_to BridgeHand

    def high_card_point_count(self):
        count = 0
        for card in self.cards:
            rank_name = Card.rank_names[card.rank]
            count += BridgeHand.hcp_dict.get(rank_name, 0)
        return count
```

```python
# This cell makes a fresh Deck and 
# initializes the random number generator

cards = Deck.make_cards()
deck = Deck(cards)
random.seed(3)
```

Per testarlo, distribuiremo una mano di cinque carte – una mano di bridge di solito ne ha tredici, ma è più facile testare il codice con esempi piccoli.

```python
hand = BridgeHand('player 2')

deck.shuffle()
deck.move_cards(hand, 5)
print(hand)
```

Ed ecco il punteggio totale per il Re e la Donna.

```python
hand.high_card_point_count()
```

`BridgeHand` eredita le variabili e i metodi di `Hand` e aggiunge una variabile di classe e un metodo specifici del bridge.
Questo modo di usare l'ereditarietà si chiama **specializzazione** (*specialization*) perché definisce una nuova classe specializzata per un uso particolare, come giocare a bridge.

## Debugging

L'ereditarietà è una funzionalità utile.
Alcuni programmi che sarebbero ripetitivi senza ereditarietà si possono scrivere in modo più conciso con essa.
Inoltre, l'ereditarietà può facilitare il riuso del codice, perché puoi personalizzare il comportamento di una classe genitore senza doverla modificare.
In alcuni casi, la struttura dell'ereditarietà rispecchia la struttura naturale del problema, il che rende il progetto più facile da capire.

D'altra parte, l'ereditarietà può rendere i programmi difficili da leggere.
Quando viene invocato un metodo, a volte non è chiaro dove trovarne la definizione – il codice pertinente potrebbe essere sparso su diversi moduli.

Ogni volta che non sei sicuro del flusso di esecuzione del tuo programma, la soluzione più semplice è aggiungere istruzioni print all'inizio dei metodi pertinenti.
Se `Deck.shuffle` stampa un messaggio come `Running Deck.shuffle`, allora, mentre il programma viene eseguito, traccia il flusso di esecuzione.

In alternativa, potresti usare la funzione seguente, che riceve un oggetto e un nome di metodo (come stringa) e restituisce la classe che fornisce la definizione del metodo.

```python
def find_defining_class(obj, method_name):
    """Find the class where the given method is defined."""
    for typ in type(obj).mro():
        if method_name in vars(typ):
            return typ
    return f'Method {method_name} not found.'
```

`find_defining_class` usa il metodo `mro` per ottenere la lista degli oggetti classe (tipi) in cui verranno cercati i metodi.
"MRO" sta per "method resolution order" (ordine di risoluzione dei metodi), cioè la sequenza di classi in cui Python cerca per "risolvere" il nome di un metodo – cioè per trovare l'oggetto funzione a cui il nome fa riferimento.

Come esempio, istanziamo una `BridgeHand` e poi troviamo la classe che definisce `shuffle`.

```python
hand = BridgeHand('player 3')
find_defining_class(hand, 'shuffle')
```

Il metodo `shuffle` dell'oggetto `BridgeHand` è quello di `Deck`.

## Glossario

**ereditarietà** (*inheritance*):
La possibilità di definire una nuova classe che è una versione modificata di una classe definita in precedenza.

**codificare** (*encode*):
Rappresentare un insieme di valori usando un altro insieme di valori, costruendo una corrispondenza tra loro.

**variabile di classe** (*class variable*):
Una variabile definita all'interno di una definizione di classe, ma non all'interno di un metodo.

**totalmente ordinato** (*totally ordered*):
Un insieme di oggetti è totalmente ordinato se possiamo confrontare due elementi qualsiasi e i risultati sono coerenti.

**delega** (*delegation*):
Quando un metodo passa a un altro metodo la responsabilità di svolgere la maggior parte del lavoro, o tutto.

**classe genitore** (*parent class*):
Una classe da cui si eredita.

I prossimi esercizi ti chiedono di scrivere funzioni che classificano le mani di poker.
Se non conosci il poker, ti spiegherò quello che c'è da sapere.
Useremo la classe seguente per rappresentare le mani di poker.

```python
class PokerHand(Hand):
    """Represents a poker hand."""

    def get_suit_counts(self):
        counter = {}
        for card in self.cards:
            key = card.suit
            counter[key] = counter.get(key, 0) + 1
        return counter
    
    def get_rank_counts(self):
        counter = {}
        for card in self.cards:
            key = card.rank
            counter[key] = counter.get(key, 0) + 1
        return counter    
```

`PokerHand` fornisce due metodi che ti aiuteranno negli esercizi.

* `get_suit_counts` scorre le carte della `PokerHand`, conta il numero di carte di ciascun seme e restituisce un dizionario che fa corrispondere a ogni codice di seme il numero di volte in cui compare.

* `get_rank_counts` fa la stessa cosa con i valori delle carte, restituendo un dizionario che fa corrispondere a ogni codice di valore il numero di volte in cui compare.

```python
%%add_method_to PokerHand

    def has_straight(self, n=5):
        """Checks whether this hand has a straight with at least `n` cards."""
        counter = self.get_rank_counts()
        aces = counter.get(1, 0) + counter.get(14, 0)
        counter[1] = aces
        counter[14] = aces
        
        return False
```

Questo esercizio è un racconto ammonitore su un errore comune che può essere difficile da correggere.
Considera la seguente definizione di classe.

```python
class Kangaroo:
    """A Kangaroo is a marsupial."""
    
    def __init__(self, name, contents=[]):
        """Initialize the pouch contents.

        name: string
        contents: initial pouch contents.
        """
        self.name = name
        self.contents = contents

    def __str__(self):
        """Return a string representaion of this Kangaroo.
        """
        t = [ self.name + ' has pouch contents:' ]
        for obj in self.contents:
            s = '    ' + object.__str__(obj)
            t.append(s)
        return '\n'.join(t)

    def put_in_pouch(self, item):
        """Adds a new item to the pouch contents.

        item: object to be added
        """
        self.contents.append(item)
```

`__init__` riceve due parametri: `name` è obbligatorio, ma `contents` è facoltativo – se non viene fornito, il valore predefinito è una lista vuota.

`__str__` restituisce una rappresentazione in stringa dell'oggetto che include il nome e il contenuto del marsupio.

`put_in_pouch` riceve un oggetto qualsiasi e lo aggiunge a `contents`.

Ora vediamo come funziona questa classe.
Creeremo due oggetti `Kangaroo` con i nomi `'Kanga'` e `'Roo'`.

```python
kanga = Kangaroo('Kanga')
roo = Kangaroo('Roo')
```

Nel marsupio di Kanga aggiungeremo due stringhe e Roo.

```python
kanga.put_in_pouch('wallet')
kanga.put_in_pouch('car keys')
kanga.put_in_pouch(roo)
```

Se stampiamo `kanga`, sembra che tutto abbia funzionato.

```python
print(kanga)
```

Ma cosa succede se stampiamo `roo`?

```python
print(roo)
```

Il marsupio di Roo ha lo stesso contenuto di quello di Kanga, compreso un riferimento a `roo`!
