## Variabili

Una **variabile** (*variable*) è un nome che fa riferimento a un valore.
Per creare una variabile, possiamo scrivere un'**istruzione di assegnazione** (*assignment statement*) come questa.

````python
n = 17
````

Un'istruzione di assegnazione ha tre parti: il nome della variabile a sinistra, l'operatore di uguaglianza, `=`, e un'espressione a destra.
In questo esempio, l'espressione è un intero.
Nell'esempio seguente, l'espressione è un numero in virgola mobile.

````python
pi = 3.141592653589793
````

E nell'esempio seguente, l'espressione è una stringa.

````python
message = 'And now for something completely different'
````

Quando esegui un'istruzione di assegnazione, non c'è alcun output.
Python crea la variabile e le assegna un valore, ma l'istruzione di assegnazione non ha alcun effetto visibile.
Tuttavia, dopo aver creato una variabile, puoi usarla come espressione.
Quindi possiamo visualizzare il valore di `message` così:

````python
message
````

Puoi anche usare una variabile come parte di un'espressione con operatori aritmetici.

````python
n + 25
````

````python
2 * pi
````

E puoi usare una variabile quando chiami una funzione.

````python
round(pi)
````

````python
len(message)
````

## Nomi delle variabili

I nomi delle variabili possono essere lunghi quanto vuoi. Possono contenere sia lettere sia numeri, ma non possono iniziare con un numero.
È lecito usare lettere maiuscole, ma per convenzione si usano solo lettere minuscole per
i nomi delle variabili.

L'unico segno di punteggiatura che può comparire in un nome di variabile è il carattere di sottolineatura (underscore), `_`. Si usa spesso nei nomi composti da più parole, come `your_name` o `airspeed_of_unladen_swallow`.

Se dai a una variabile un nome non valido, ottieni un errore di sintassi.
Il nome `million!` non è valido perché contiene un segno di punteggiatura.

````python
%%expect SyntaxError

million! = 1000000
````

`76trombones` non è valido perché inizia con un numero.

````python
%%expect SyntaxError

76trombones = 'big parade'
````

Anche `class` non è valido, ma il motivo potrebbe non essere evidente.

````python
%%expect SyntaxError

class = 'Self-Defence Against Fresh Fruit'
````

Il fatto è che `class` è una **parola chiave** (*keyword*), cioè una parola speciale usata per specificare la struttura di un programma.
Le parole chiave non si possono usare come nomi di variabili.


## L'istruzione import

Per usare alcune funzionalità di Python, devi **importarle** (*import*).
Per esempio, l'istruzione seguente importa il modulo `math`.

````python
import math
````

Un **modulo** (*module*) è una raccolta di variabili e funzioni.
Il modulo math fornisce una variabile chiamata `pi` che contiene il valore della costante matematica indicata con $\pi$.
Possiamo visualizzarne il valore così.

````python
math.pi
````

Per usare una variabile di un modulo, devi usare l'**operatore punto** (*dot operator*, `.`) tra il nome del modulo e il nome della variabile.

Il modulo math contiene anche delle funzioni.
Per esempio, `sqrt` calcola le radici quadrate.

````python
math.sqrt(25)
````

E `pow` eleva un numero alla potenza di un secondo numero.

````python
math.pow(5, 2)
````

A questo punto abbiamo visto due modi per elevare un numero a potenza: possiamo usare la funzione `math.pow` oppure l'operatore di elevamento a potenza, `**`.
Vanno bene entrambi, ma l'operatore si usa più spesso della funzione.

## Espressioni e istruzioni

Finora abbiamo visto alcuni tipi di espressioni.
Un'espressione può essere un singolo valore, come un intero, un numero in virgola mobile o una stringa.
Può anche essere un insieme di valori e operatori.
E può includere nomi di variabili e chiamate di funzione.
Ecco un'espressione che include molti di questi elementi.

````python
19 + n + round(math.pi) * 2
````

Abbiamo anche visto alcuni tipi di istruzioni.
Un'**istruzione** (*statement*) è un'unità di codice che ha un effetto, ma non ha un valore.
Per esempio, un'istruzione di assegnazione crea una variabile e le assegna un valore, ma l'istruzione in sé non ha alcun valore.

````python
n = 17
````

Allo stesso modo, un'istruzione import ha un effetto – importa un modulo così che possiamo usare le variabili e le funzioni che contiene – ma non ha un effetto visibile.

````python
import math
````

Calcolare il valore di un'espressione si chiama **valutazione** (*evaluation*).
Eseguire un'istruzione si chiama **esecuzione** (*execution*).

## La funzione print

Quando valuti un'espressione, il risultato viene visualizzato.

````python
n + 1
````

Ma se valuti più di un'espressione, viene visualizzato solo il valore dell'ultima.

````python
n + 2
n + 3
````

Per visualizzare più di un valore, puoi usare la funzione `print`.

````python
print(n+2)
print(n+3)
````

Funziona anche con i numeri in virgola mobile e con le stringhe.

````python
print('The value of pi is approximately')
print(math.pi)
````

Puoi anche usare una sequenza di espressioni separate da virgole.

````python
print('The value of pi is approximately', math.pi)
````

Nota che la funzione `print` mette uno spazio tra i valori.

## Argomenti

Quando chiami una funzione, l'espressione tra parentesi si chiama **argomento** (*argument*).
Di solito spiegherei il perché, ma in questo caso il significato tecnico del termine non ha quasi nulla a che fare con il significato comune della parola, quindi non ci proverò nemmeno.

Alcune delle funzioni che abbiamo visto finora ricevono un solo argomento, come `int`.

````python
int('101')
````

Alcune ne ricevono due, come `math.pow`.

````python
math.pow(5, 2)
````

Alcune possono ricevere argomenti aggiuntivi facoltativi.
Per esempio, `int` può ricevere un secondo argomento che specifica la base del numero.

````python
int('101', 2)
````

La sequenza di cifre `101` in base 2 rappresenta il numero 5 in base 10.

Anche `round` riceve un secondo argomento facoltativo, che è il numero di cifre decimali a cui arrotondare.

````python
round(math.pi, 3)
````

Alcune funzioni possono ricevere un numero qualsiasi di argomenti, come `print`.

````python
print('Any', 'number', 'of', 'arguments')
````

Se chiami una funzione e fornisci troppi argomenti, ottieni un `TypeError`.

````python
%%expect TypeError

float('123.0', 2)
````

Se fornisci troppo pochi argomenti, ottieni ancora un `TypeError`.

````python
%%expect TypeError

math.pow(2)
````

E se fornisci un argomento di un tipo che la funzione non sa gestire, anche questo è un `TypeError`.

````python
%%expect TypeError

math.sqrt('123')
````

Questo tipo di controllo può essere fastidioso quando inizi, ma ti aiuta a individuare e correggere gli errori.

## Commenti

Man mano che i programmi diventano più grandi e complicati, diventano più difficili da leggere.
I linguaggi formali sono densi, e spesso è difficile guardare un pezzo di codice e capire cosa fa e perché.

Per questo motivo è una buona idea aggiungere note ai programmi per spiegare in linguaggio naturale cosa sta facendo il programma.
Queste note si chiamano **commenti** (*comments*) e iniziano con il simbolo `#`.

````python
# number of seconds in 42:42
seconds = 42 * 60 + 42
````

In questo caso, il commento si trova da solo su una riga. Puoi anche mettere
i commenti alla fine di una riga:

````python
miles = 10 / 1.61     # 10 kilometers in miles
````

Tutto ciò che va dal `#` alla fine della riga viene ignorato – non ha
alcun effetto sull'esecuzione del programma.

I commenti sono più utili quando documentano aspetti non ovvi del codice.
È ragionevole supporre che il lettore possa capire *cosa* fa il codice; è più utile spiegare *perché*.

Questo commento è ridondante rispetto al codice e inutile:

````python
v = 8     # assign 8 to v
````

Questo commento contiene informazioni utili che non sono nel codice:

````python
v = 8     # velocity in miles per hour 
````

Dei buoni nomi di variabile possono ridurre il bisogno di commenti, ma i nomi lunghi possono
rendere difficili da leggere le espressioni complesse, quindi c'è un compromesso da trovare.

## Debugging

In un programma possono verificarsi tre tipi di errori: errori di sintassi, errori di runtime ed errori semantici.
È utile distinguerli per scovarli più rapidamente.

* **Errore di sintassi** (*syntax error*): "sintassi" si riferisce alla struttura di un programma e alle regole su quella struttura. Se c'è un errore di sintassi in un punto qualsiasi del programma, Python non esegue il programma. Visualizza subito un messaggio di errore.

* **Errore di runtime** (*runtime error*): se non ci sono errori di sintassi nel programma, questo può iniziare l'esecuzione. Ma se qualcosa va storto, Python visualizza un messaggio di errore e si ferma. Questo tipo di errore si chiama errore di runtime (errore in fase di esecuzione). Si chiama anche **eccezione** (*exception*) perché indica che è successo qualcosa di eccezionale.

* **Errore semantico** (*semantic error*): il terzo tipo di errore è "semantico", cioè legato al significato. Se c'è un errore semantico nel programma, questo viene eseguito senza generare messaggi di errore, ma non fa quello che intendevi. Individuare gli errori semantici può essere difficile, perché richiede di ragionare a ritroso guardando l'output del programma e cercando di capire cosa sta facendo.

Come abbiamo visto, un nome di variabile non valido è un errore di sintassi.

````python
%%expect SyntaxError

million! = 1000000
````

Se usi un operatore con un tipo che non supporta, è un errore di runtime.

````python
%%expect TypeError

'126' / 3
````

Infine, ecco un esempio di errore semantico.
Supponiamo di voler calcolare la media di `1` e `3`, ma di dimenticare l'ordine delle operazioni e scrivere questo:

````python
1 + 3 / 2
````

Quando questa espressione viene valutata non produce alcun messaggio di errore, quindi non c'è un errore di sintassi né un errore di runtime.
Ma il risultato non è la media di `1` e `3`, quindi il programma non è corretto.
È un errore semantico perché il programma viene eseguito ma non fa ciò che si voleva.

## Glossario

**variabile** (*variable*):
Un nome che fa riferimento a un valore.

**istruzione di assegnazione** (*assignment statement*):
Un'istruzione che assegna un valore a una variabile.

**diagramma di stato** (*state diagram*):
Una rappresentazione grafica di un insieme di variabili e dei valori a cui fanno riferimento.

**parola chiave** (*keyword*):
Una parola speciale usata per specificare la struttura di un programma.

**istruzione import** (*import statement*):
Un'istruzione che legge il file di un modulo così che possiamo usare le variabili e le funzioni che contiene.

**modulo** (*module*):
Un file che contiene codice Python, incluse definizioni di funzioni e a volte altre istruzioni.

**operatore punto** (*dot operator*):
L'operatore `.`, usato per accedere a una funzione di un altro modulo specificando il nome del modulo seguito da un punto e dal nome della funzione.

**valutare** (*evaluate*):
Eseguire le operazioni di un'espressione per calcolarne il valore.

**istruzione** (*statement*):
Una o più righe di codice che rappresentano un comando o un'azione.

**eseguire** (*execute*):
Mandare in esecuzione un'istruzione e fare ciò che dice.

**argomento** (*argument*):
Un valore fornito a una funzione quando la funzione viene chiamata.

**commento** (*comment*):
Testo incluso in un programma che fornisce informazioni sul programma ma non ha alcun effetto sulla sua esecuzione.

**errore di runtime** (*runtime error*):
Un errore che fa sì che un programma visualizzi un messaggio di errore e termini.

**eccezione** (*exception*):
Un errore rilevato mentre il programma è in esecuzione.

**errore semantico** (*semantic error*):
Un errore che fa sì che un programma faccia la cosa sbagliata, ma senza visualizzare un messaggio di errore.
