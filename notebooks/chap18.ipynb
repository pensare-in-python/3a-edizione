## Insiemi

Python fornisce una classe chiamata `set` che rappresenta una raccolta di elementi unici (un insieme).
Per creare un insieme vuoto, possiamo usare l'oggetto classe come una funzione.

```python
s1 = set()
s1
```

Possiamo usare il metodo `add` per aggiungere elementi.

```python
s1.add('a')
s1.add('b')
s1
```

Oppure possiamo passare a `set` qualsiasi tipo di sequenza.

```python
s2 = set('acd')
s2
```

Un elemento può comparire in un `set` una sola volta.
Se aggiungi un elemento che c'è già, non ha alcun effetto.

```python
s1.add('a')
s1
```

Oppure, se crei un insieme con una sequenza che contiene duplicati, il risultato contiene solo elementi unici.

```python
set('banana')
```

Alcuni degli esercizi di questo libro si possono risolvere in modo conciso ed efficiente con gli insiemi.
Per esempio, ecco la soluzione di un esercizio del capitolo 11 che usa un dizionario per verificare se ci sono elementi duplicati in una sequenza.

```python
def has_duplicates(t):
    d = {}
    for x in t:
        d[x] = True
    return len(d) < len(t)
```

Questa versione aggiunge gli elementi di `t` come chiavi in un dizionario, e poi verifica se ci sono meno chiavi che elementi.
Usando gli insiemi, possiamo scrivere la stessa funzione così.

```python
def has_duplicates(t):
    s = set(t)
    return len(s) < len(t)
```

```python
has_duplicates('abba')
```

Un elemento può comparire in un insieme una sola volta, quindi se un elemento di `t` compare più di una volta, l'insieme sarà più piccolo di `t`.
Se non ci sono duplicati, l'insieme avrà la stessa dimensione di `t`.

Gli oggetti `set` forniscono metodi che eseguono operazioni insiemistiche.

Alcuni operatori aritmetici funzionano con gli insiemi.
Per esempio, l'operatore `-` esegue la sottrazione di insiemi – il risultato è un nuovo insieme che contiene tutti gli elementi del primo insieme che _non_ sono nel secondo.

```python
s1 - s2
```

Nel [capitolo 12](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap12.ipynb) abbiamo usato i dizionari per trovare le parole che compaiono in un documento ma non in un elenco di parole.
Abbiamo usato la funzione seguente, che riceve due dizionari e restituisce un nuovo dizionario che contiene solo le chiavi del primo che non compaiono nel secondo.

```python
def subtract(d1, d2):
    res = {}
    for key in d1:
        if key not in d2:
            res[key] = d1[key]
    return res
```

```python
def uses_only(word, available):
    for letter in word: 
        if letter not in available:
            return False
    return True
```

## Counter

Un `Counter` è come un insieme, tranne che, se un elemento compare più di una volta, il `Counter` tiene traccia di quante volte compare.
Se conosci il concetto matematico di "multiinsieme", un `Counter` è un
modo naturale di rappresentare un multiinsieme.

La classe `Counter` è definita in un modulo chiamato `collections`, quindi devi importarla.
Poi puoi usare l'oggetto classe come una funzione e passare come argomento una stringa, una lista o qualsiasi altro tipo di sequenza.

```python
from collections import Counter

counter = Counter('banana')
counter
```

Un oggetto `Counter` è come un dizionario che fa corrispondere a ogni chiave il numero di volte in cui compare.
Come nei dizionari, le chiavi devono essere hashabili.

A differenza dei dizionari, gli oggetti `Counter` non sollevano un'eccezione se accedi a un
elemento che non compare.
Restituiscono invece `0`.

```python
counter['d']
```

Possiamo usare gli oggetti `Counter` per risolvere uno degli esercizi del capitolo 10, che chiede una funzione che riceve due parole e verifica se sono anagrammi – cioè se le lettere dell'una possono essere riordinate per comporre l'altra.

Ecco una soluzione che usa oggetti `Counter`.

```python
def is_anagram(word1, word2):
    return Counter(word1) == Counter(word2)
```

Se due parole sono anagrammi, contengono le stesse lettere con gli stessi conteggi, quindi i loro oggetti `Counter` sono equivalenti.

`Counter` fornisce un metodo chiamato `most_common` che restituisce una lista di coppie valore-frequenza, ordinate dalla più comune alla meno comune.

```python
counter.most_common()
```

I `Counter` forniscono anche metodi e operatori per eseguire operazioni simili a quelle sugli insiemi, tra cui addizione, sottrazione, unione e intersezione.
Per esempio, l'operatore `+` combina due oggetti `Counter` e crea un nuovo `Counter` che contiene le chiavi di entrambi e le somme dei conteggi.

Possiamo testarlo creando un `Counter` con le lettere di `'bans'` e sommandolo alle lettere di `'banana'`.

```python
counter2 = Counter('bans')
counter + counter2
```

## defaultdict

Il modulo `collections` fornisce anche `defaultdict`, che è come un dizionario tranne che, se accedi a una chiave che non esiste, genera automaticamente un nuovo valore.

Quando crei un `defaultdict`, fornisci una funzione che viene usata per creare i nuovi valori.
Una funzione che crea oggetti a volte si chiama **fabbrica** (*factory*).
Le funzioni predefinite che creano liste, insiemi e altri tipi si possono usare come fabbriche.

Per esempio, ecco un `defaultdict` che crea una nuova `list` quando serve.

```python
from collections import defaultdict

d = defaultdict(list)
d
```

Nota che l'argomento è `list`, che è un oggetto classe, e non `list()`, che è una chiamata di funzione che crea una nuova lista.
La funzione fabbrica non viene chiamata a meno che non accediamo a una chiave che non esiste.

```python
t = d['new key']
t
```

La nuova lista, che stiamo chiamando `t`, viene anche aggiunta al dizionario.
Quindi se modifichiamo `t`, la modifica compare in `d`:

```python
t.append('new value')
d['new key']
```

Se stai creando un dizionario di liste, spesso puoi scrivere codice più
semplice usando `defaultdict`.

In uno degli esercizi del [capitolo 11](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap11.ipynb), ho creato un dizionario che fa corrispondere a una stringa di lettere ordinate la lista delle parole che si possono comporre con quelle lettere.
Per esempio, la stringa `'opst'` corrisponde alla lista `['opts', 'post', 'pots', 'spot', 'stop', 'tops']`.
Ecco il codice originale.

```python
def all_anagrams(filename):
    d = {}
    for line in open(filename):
        word = line.strip().lower()
        t = signature(word)
        if t not in d:
            d[t] = [word]
        else:
            d[t].append(word)
    return d
```

Ed ecco una versione più semplice che usa un `defaultdict`.

```python
def all_anagrams(filename):
    d = defaultdict(list)
    for line in open(filename):
        word = line.strip().lower()
        t = signature(word)
        d[t].append(word)
    return d
```

## Espressioni condizionali

Le istruzioni condizionali si usano spesso per scegliere uno tra due valori, così:

```python
import math
x = 5
```

```python
if x > 0:
    y = math.log(x)
else:
    y = float('nan')
```

```python
y
```

Questa istruzione verifica se `x` è positivo. In tal caso ne calcola il logaritmo.
Altrimenti, `math.log` solleverebbe un ValueError.
Per evitare di fermare il programma, generiamo un `NaN`, che è un valore speciale in virgola mobile che rappresenta "Not a Number" (non un numero).

Possiamo scrivere questa istruzione in modo più conciso usando un'**espressione condizionale** (*conditional expression*).

```python
y = math.log(x) if x > 0 else float('nan')
```

```python
y
```

In inglese questa riga si legge quasi come una frase: "`y` riceve log-`x` se `x` è maggiore di 0; altrimenti riceve `NaN`".

Le funzioni ricorsive a volte si possono scrivere in modo conciso usando espressioni condizionali.
Per esempio, ecco una versione di `factorial` con un'_istruzione_ condizionale.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

Ed ecco una versione con un'_espressione_ condizionale.

```python
def factorial(n):
    return 1 if n == 0 else n * factorial(n-1)
```

Un altro uso delle espressioni condizionali è la gestione degli argomenti facoltativi.
Per esempio, ecco una definizione di classe con un metodo `__init__` che usa un'istruzione condizionale per controllare un parametro con un valore predefinito.

```python
class Kangaroo:
    def __init__(self, name, contents=None):
        self.name = name
        if contents is None:
            contents = []
        self.contents = contents
```

Ecco una versione che usa un'espressione condizionale.

```python
def __init__(self, name, contents=None):
    self.name = name
    self.contents = [] if contents is None else contents 
```

In generale, puoi sostituire un'istruzione condizionale con un'espressione condizionale se entrambi i rami contengono un'unica espressione e nessuna istruzione.

## List comprehension

Nei capitoli precedenti abbiamo visto alcuni esempi in cui partiamo da una lista vuota e aggiungiamo elementi, uno alla volta, usando il metodo `append`.
Per esempio, supponiamo di avere una stringa che contiene il titolo di un film, e di voler rendere maiuscola l'iniziale di tutte le parole.

```python
title = 'monty python and the holy grail'
```

Possiamo dividerla in una lista di stringhe, scorrere le stringhe, renderle maiuscole e aggiungerle a una lista.

```python
t = []
for word in title.split():
    t.append(word.capitalize())

' '.join(t)
```

Possiamo fare la stessa cosa in modo più conciso usando una **list comprehension** (costruzione di lista):

```python
t = [word.capitalize() for word in title.split()]

' '.join(t)
```

Gli operatori parentesi quadre indicano che stiamo costruendo una nuova lista.
L'espressione tra parentesi quadre specifica gli elementi della lista, e la clausola `for` indica quale sequenza stiamo scorrendo.

La sintassi di una list comprehension potrebbe sembrare strana, perché la variabile di ciclo – `word` in questo esempio – compare nell'espressione prima che arriviamo alla sua definizione.
Ma ci si abitua.

Come altro esempio, nel [capitolo 9](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap09.ipynb) abbiamo usato questo ciclo per leggere le parole da un file e aggiungerle a una lista.

```python
download('https://raw.githubusercontent.com/AllenDowney/ThinkPython2/master/code/words.txt');
```

```python
word_list = []

for line in open('words.txt'):
    word = line.strip()
    word_list.append(word)
```

```python
len(word_list)
```

Ecco come possiamo scriverlo come list comprehension.

```python
word_list = [line.strip() for line in open('words.txt')]
```

```python
len(word_list)
```

Una list comprehension può avere anche una clausola `if` che determina quali elementi vengono inclusi nella lista.
Per esempio, ecco un ciclo `for` che abbiamo usato nel [capitolo 10](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap10.ipynb) per creare una lista delle sole parole di `word_list` che sono palindromi.

```python
def is_palindrome(word):
    return list(reversed(word)) == list(word)
```

```python
palindromes = []

for word in word_list:
    if is_palindrome(word):
        palindromes.append(word)
```

```python
palindromes[:10]
```

Ecco come possiamo fare la stessa cosa con una list comprehension.

```python
palindromes = [word for word in word_list if is_palindrome(word)]
```

```python
palindromes[:10]
```

Quando una list comprehension è usata come argomento di una funzione, spesso possiamo omettere le parentesi quadre.
Per esempio, supponiamo di voler sommare $1 / 2^n$ per i valori di $n$ da 0 a 9.
Possiamo usare una list comprehension così.

```python
sum([1/2**n for n in range(10)])
```

Oppure possiamo omettere le parentesi quadre così.

```python
sum(1/2**n for n in range(10))
```

In questo esempio, l'argomento è tecnicamente un'**espressione generatrice** (*generator expression*), non una list comprehension, e non crea mai davvero una lista.
Ma a parte questo, il comportamento è lo stesso.

Le list comprehension e le espressioni generatrici sono concise e facili da leggere, almeno per le espressioni semplici.
E di solito sono più veloci dei cicli for equivalenti, a volte molto più veloci.

## `any` e `all`

Python fornisce una funzione predefinita, `any`, che riceve una sequenza di valori booleani e restituisce `True` se almeno uno dei valori è `True`.

```python
any([False, False, True])
```

`any` si usa spesso con le espressioni generatrici.

```python
any(letter == 't' for letter in 'monty')
```

Quell'esempio non è molto utile perché fa la stessa cosa dell'operatore `in`.
Ma potremmo usare `any` per scrivere soluzioni concise ad alcuni degli esercizi del [capitolo 7](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap07.ipynb). Per esempio, possiamo scrivere `uses_none` così.

```python
def uses_none(word, forbidden):
    """Checks whether a word avoids forbidden letters."""
    return not any(letter in forbidden for letter in word)
```

```python
uses_none('banana', 'xyz')
```

```python
uses_none('apple', 'efg')
```

Questa funzione scorre le lettere di `word` e verifica se qualcuna è in `forbidden`.
Usare `any` con un'espressione generatrice è efficiente perché si ferma subito se trova un valore `True`, quindi non deve scorrere l'intera sequenza.

Python fornisce un'altra funzione predefinita, `all`, che restituisce `True` se ogni elemento della sequenza è `True`.
Possiamo usarla per scrivere una versione concisa di `uses_all`.

```python
def uses_all(word, required):
    """Check whether a word uses all required letters."""
    return all(letter in word for letter in required)
```

```python
uses_all('banana', 'ban')
```

```python
uses_all('apple', 'api')
```

Le espressioni che usano `any` e `all` possono essere concise, efficienti e facili da leggere.

## Tuple con nome

Il modulo `collections` fornisce una funzione chiamata `namedtuple` che si può usare per creare classi semplici.
Per esempio, l'oggetto `Point` del [capitolo 16](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap16.ipynb) ha solo due attributi, `x` e `y`.
Ecco come lo avevamo definito.

```python
class Point:
    """Represents a point in 2-D space."""
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f'({self.x}, {self.y})'
```

È molto codice per trasmettere poche informazioni.
`namedtuple` offre un modo più conciso di definire classi come questa.

```python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
```

Il primo argomento è il nome della classe che vuoi creare. Il
secondo è una lista degli attributi che gli oggetti `Point` devono avere.
Il risultato è un oggetto classe, ed è per questo che viene assegnato a un nome di variabile con l'iniziale maiuscola.

Una classe creata con `namedtuple` fornisce un metodo `__init__` che assegna i valori agli attributi e un metodo `__str__` che visualizza l'oggetto in forma leggibile.
Quindi possiamo creare e visualizzare un oggetto `Point` così.

```python
p = Point(1, 2)
p
```

`Point` fornisce anche un metodo `__eq__` che verifica se due oggetti `Point` sono equivalenti – cioè se i loro attributi sono uguali.

```python
p == Point(1, 2)
```

Puoi accedere agli elementi di una tupla con nome per nome o per indice.

```python
p.x, p.y
```

```python
p[0], p[1]
```

Puoi anche trattare una tupla con nome come una tupla, come in questa assegnazione.

```python
x, y = p
x, y
```

Ma gli oggetti `namedtuple` sono immutabili.
Una volta inizializzati, gli attributi non si possono cambiare.

```python
%%expect TypeError

p[0] = 3
```

```python
%%expect AttributeError

p.x = 3
```

`namedtuple` offre un modo rapido di definire classi semplici.
Lo svantaggio è che le classi semplici non sempre restano semplici.
Potresti decidere in seguito di voler aggiungere metodi a una tupla con nome.
In quel caso, puoi definire una nuova classe che eredita dalla tupla con nome.

```python
class Pointier(Point):
    """This class inherits from Point"""
```

## Impacchettare gli argomenti keyword

Nel [capitolo 11](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap11.ipynb) abbiamo scritto una funzione che impacchetta i suoi argomenti in una tupla.

```python
def mean(*args):
    return sum(args) / len(args)
```

Puoi chiamare questa funzione con un numero qualsiasi di argomenti.

```python
mean(1, 2, 3)
```

Ma l'operatore `*` non impacchetta gli argomenti keyword.
Quindi chiamare questa funzione con un argomento keyword provoca un errore.

```python
%%expect TypeError

mean(1, 2, start=3)
```

Per impacchettare gli argomenti keyword, possiamo usare l'operatore `**`:

```python
def mean(*args, **kwargs):
    print(kwargs)
    return sum(args) / len(args)
```

Il parametro che impacchetta gli argomenti keyword può avere qualsiasi nome, ma `kwargs` è una scelta comune.
Il risultato è un dizionario che fa corrispondere le keyword ai valori.

```python
mean(1, 2, start=3)
```

In questo esempio il valore di `kwargs` viene stampato, ma per il resto non ha alcun effetto.

Ma l'operatore `**` si può usare anche in un elenco di argomenti per spacchettare un dizionario.
Per esempio, ecco una versione di `mean` che impacchetta gli eventuali argomenti keyword che riceve e poi li spacchetta come argomenti keyword per `sum`.

```python
def mean(*args, **kwargs):
    return sum(args, **kwargs) / len(args)
```

Ora, se chiamiamo `mean` con `start` come argomento keyword, questo viene passato a sum, che lo usa come punto di partenza della somma.
Nell'esempio seguente, `start=3` aggiunge `3` alla somma prima di calcolare la media, quindi la somma è `6` e il risultato è `3`.

```python
mean(1, 2, start=3)
```

Come altro esempio, se abbiamo un dizionario con chiavi `x` e `y`, possiamo usarlo con l'operatore di spacchettamento per creare un oggetto `Point`.

```python
d = dict(x=1, y=2)
Point(**d)
```

Senza l'operatore di spacchettamento, `d` viene trattato come un unico argomento posizionale, quindi viene assegnato a `x`, e otteniamo un `TypeError` perché non c'è un secondo argomento da assegnare a `y`.

```python
%%expect TypeError

d = dict(x=1, y=2)
Point(d)
```

Quando lavori con funzioni che hanno molti argomenti keyword, spesso è utile creare e passare dizionari che specificano le opzioni usate più spesso.

```python
def pack_and_print(**kwargs):
    print(kwargs)
    
pack_and_print(a=1, b=2)
```

Nei capitoli precedenti abbiamo usato `doctest` per testare le funzioni.
Per esempio, ecco una funzione chiamata `add` che riceve due numeri e ne restituisce la somma.
Include un doctest che verifica se `2 + 2` fa `4`.

```python
def add(a, b):
    '''Add two numbers.
    
    >>> add(2, 2)
    4
    '''
    return a + b
```

Questa funzione riceve un oggetto funzione ed esegue i suoi doctest.

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

Quindi possiamo testare `add` così.

```python
run_doctests(add)
```

Non c'è output, il che significa che tutti i test sono passati.

Python fornisce un altro strumento per eseguire test automatici, chiamato `unittest`.
È un po' più complicato da usare, ma ecco un esempio.

```python
from unittest import TestCase

class TestExample(TestCase):

    def test_add(self):
        result = add(2, 2)
        self.assertEqual(result, 4)
```

Per prima cosa importiamo `TestCase`, che è una classe del modulo `unittest`.
Per usarla, dobbiamo definire una nuova classe che eredita da `TestCase` e fornisce almeno un metodo di test.
Il nome del metodo di test deve iniziare con `test` e dovrebbe indicare quale funzione testa.

In questo esempio, `test_add` testa la funzione `add` chiamandola, salvando il risultato e invocando `assertEqual`, che è ereditato da `TestCase`.
`assertEqual` riceve due argomenti e verifica se sono uguali.

Per eseguire questo metodo di test, dobbiamo eseguire una funzione di `unittest` chiamata `main` e fornire diversi argomenti keyword.
La funzione seguente mostra i dettagli – se sei curioso, puoi chiedere a un assistente virtuale di spiegarti come funziona.

```python
import unittest

def run_unittest():
    unittest.main(argv=[''], verbosity=0, exit=False)
```

`run_unittest` non riceve `TestExample` come argomento – cerca invece le classi che ereditano da `TestCase`.
Poi cerca i metodi che iniziano con `test` e li esegue.
Questo processo si chiama **scoperta dei test** (*test discovery*).

Ecco cosa succede quando chiamiamo `run_unittest`.

```python
run_unittest()
```

`unittest.main` riporta il numero di test eseguiti e i risultati.
In questo caso `OK` indica che i test sono passati.

Per vedere cosa succede quando un test fallisce, aggiungeremo un metodo di test sbagliato a `TestExample`.

```python
%%add_method_to TestExample

    def test_add_broken(self):
        result = add(2, 2)
        self.assertEqual(result, 100)
```

Ecco cosa succede quando eseguiamo i test.

```python
run_unittest()
```

Il rapporto include il metodo di test che è fallito e un messaggio di errore che mostra dove.
Il riepilogo indica che sono stati eseguiti due test e uno è fallito.

## Glossario

**fabbrica** (*factory*):
Una funzione usata per creare oggetti, spesso passata come parametro a una funzione.

**espressione condizionale** (*conditional expression*):
Un'espressione che usa una condizione per scegliere uno tra due valori.

**list comprehension**:
Un modo conciso di scorrere una sequenza e creare una lista.

**espressione generatrice** (*generator expression*):
Simile a una list comprehension, tranne che non crea una lista.

Un processo usato per trovare ed eseguire i test.

```python
def binomial_coeff(n, k):
    """Compute the binomial coefficient "n choose k".

    n: number of trials
    k: number of successes

    returns: int
    """
    if k == 0:
        return 1
    
    if n == 0:
        return 0

    return binomial_coeff(n-1, k) + binomial_coeff(n-1, k-1)
```
