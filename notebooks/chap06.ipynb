## Alcune funzioni hanno valori di ritorno

Quando chiami una funzione come `math.sqrt`, il risultato si chiama **valore di ritorno** (*return value*).
Se la chiamata di funzione si trova alla fine di una cella, Jupyter visualizza subito il valore di ritorno.

Oppure puoi usare il valore di ritorno come parte di un'espressione.

Ecco un esempio di funzione che restituisce un valore.

```python
def circle_area(radius):
    area = math.pi * radius**2
    return area
```

`circle_area` riceve `radius` come parametro e calcola l'area di un cerchio con quel raggio.

L'ultima riga è un'istruzione `return` che restituisce il valore di `area`.

```python
circle_area(radius) + 2 * circle_area(radius / 2)
```

```python
a
```

```python
%%expect NameError

area
```

## E alcune restituiscono None

Se una funzione non ha un'istruzione `return`, restituisce `None`, che è un valore speciale come `True` e `False`.

## Valori di ritorno e condizionali

Se Python non fornisse `abs`, potremmo scriverla così.

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x
```

Se `x` è negativo, la prima istruzione `return` restituisce `-x` e la funzione termina immediatamente.
Altrimenti, la seconda istruzione `return` restituisce `x` e la funzione termina.
Quindi questa funzione è corretta.

Tuttavia, se metti delle istruzioni `return` in un condizionale, devi assicurarti che ogni possibile percorso nel programma incontri un'istruzione `return`.
Per esempio, ecco una versione sbagliata di `absolute_value`.

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

Ecco cosa succede se chiamiamo questa funzione con `0` come argomento.

```python
absolute_value_wrong(0)
```

Non otteniamo nulla! Il problema è questo: quando `x` è `0`, nessuna delle due condizioni è vera e la funzione termina senza incontrare un'istruzione `return`, il che significa che il valore di ritorno è `None`, quindi Jupyter non visualizza nulla.

Come altro esempio, ecco una versione di `absolute_value` con un'istruzione `return` in più alla fine.

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x
    
    return 'This is dead code'
```

Se `x` è negativo, viene eseguita la prima istruzione `return` e la funzione termina.
Altrimenti viene eseguita la seconda istruzione `return` e la funzione termina.
In entrambi i casi non arriviamo mai alla terza istruzione `return` – che quindi non può mai essere eseguita.

Il codice che non può mai essere eseguito si chiama **codice morto** (*dead code*).
In generale il codice morto non fa danni, ma spesso indica un'incomprensione e può confondere chi cerca di capire il programma.

## Sviluppo incrementale

Man mano che scrivi funzioni più grandi, potresti ritrovarti a passare più
tempo a fare debugging.
Per gestire programmi sempre più complessi, potresti provare lo **sviluppo incrementale** (*incremental development*), un modo di aggiungere e testare solo una piccola quantità di codice alla volta.

Come esempio, supponiamo di voler trovare la distanza tra due punti rappresentati dalle coordinate $(x_1, y_1)$ e $(x_2, y_2)$.
Per il teorema di Pitagora, la distanza è:

$$\mathrm{distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

Il primo passo è chiedersi come dovrebbe essere una funzione `distance` in Python – cioè, quali sono gli input (parametri) e quale l'output (valore di ritorno)?

Per questa funzione, gli input sono le coordinate dei punti.
Il valore di ritorno è la distanza.
Puoi scrivere subito uno schema della funzione:

```python
def distance(x1, y1, x2, y2):
    return 0.0
```

Questa versione non calcola ancora le distanze – restituisce sempre zero.
Ma è una funzione completa con un valore di ritorno, il che significa che puoi testarla prima di renderla più complicata.

Per testare la nuova funzione, la chiameremo con degli argomenti di esempio:

```python
distance(1, 2, 4, 6)
```

Ho scelto questi valori in modo che la distanza orizzontale sia `3` e la
distanza verticale sia `4`.
In questo modo il risultato è `5`, l'ipotenusa di un triangolo rettangolo `3-4-5`. Quando testi una funzione, è utile conoscere la risposta giusta.

A questo punto abbiamo confermato che la funzione viene eseguita e restituisce un valore, e possiamo iniziare ad aggiungere codice al corpo.
Un buon passo successivo è trovare le differenze `x2 - x1` e `y2 - y1`.
Ecco una versione che memorizza quei valori in variabili temporanee e li visualizza.

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    print('dx is', dx)
    print('dy is', dy)
    return 0.0
```

Se la funzione funziona, dovrebbe visualizzare `dx is 3` e `dy is 4`.
In tal caso sappiamo che la funzione riceve gli argomenti giusti ed
esegue correttamente il primo calcolo. Altrimenti, ci sono solo poche
righe da controllare.

```python
distance(1, 2, 4, 6)
```

Fin qui tutto bene. Ora calcoliamo la somma dei quadrati di `dx` e `dy`:

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    print('dsquared is: ', dsquared)
    return 0.0
```

Di nuovo, possiamo eseguire la funzione e controllare l'output, che dovrebbe essere `25`.

```python
distance(1, 2, 4, 6)
```

Infine, possiamo usare `math.sqrt` per calcolare la distanza:

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    print("result is", result)
```

E testarla.

```python
distance(1, 2, 4, 6)
```

Il risultato è corretto, ma questa versione della funzione visualizza il risultato invece di restituirlo, quindi il valore di ritorno è `None`.

Possiamo correggere sostituendo la funzione `print` con un'istruzione `return`.

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    return result
```

Questa versione di `distance` è una funzione pura.
Se la chiamiamo così, viene visualizzato solo il risultato.

```python
distance(1, 2, 4, 6)
```

E se assegniamo il risultato a una variabile, non viene visualizzato nulla.

```python
d = distance(1, 2, 4, 6)
```

Le istruzioni `print` che abbiamo scritto sono utili per il debugging, ma una volta che la funzione funziona possiamo toglierle.
Codice di questo tipo si chiama **impalcatura** (*scaffolding*) perché è utile per costruire il programma ma non fa parte del prodotto finale.

Questo esempio illustra lo sviluppo incrementale.
Gli aspetti chiave di questo procedimento sono:

1.  Parti da un programma funzionante, fai piccole modifiche e testa dopo ogni modifica.

2.  Usa variabili per contenere i valori intermedi, così puoi visualizzarli e controllarli.

3.  Quando il programma funziona, rimuovi l'impalcatura.

In qualsiasi momento, se c'è un errore, dovresti avere una buona idea di dove si trovi.
Lo sviluppo incrementale può farti risparmiare molto tempo di debugging.

## Funzioni booleane

Le funzioni possono restituire i valori booleani `True` e `False`, il che spesso è comodo per incapsulare un test complesso in una funzione.
Per esempio, `is_divisible` verifica se `x` è divisibile per `y` senza resto.

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```

Ecco come la usiamo.

```python
is_divisible(6, 4)
```

```python
is_divisible(6, 3)
```

All'interno della funzione, il risultato dell'operatore `==` è un booleano, quindi possiamo scrivere la
funzione in modo più conciso restituendolo direttamente.

```python
def is_divisible(x, y):
    return x % y == 0
```

Le funzioni booleane si usano spesso nelle istruzioni condizionali.

```python
if is_divisible(6, 2):
    print('divisible')
```

Si potrebbe essere tentati di scrivere qualcosa del genere:

```python
if is_divisible(6, 2) == True:
    print('divisible')
```

Ma il confronto non è necessario.

## Ricorsione con valori di ritorno

Ora che sappiamo scrivere funzioni con valori di ritorno, possiamo scrivere funzioni ricorsive con valori di ritorno, e con questa capacità abbiamo superato una soglia importante – il sottoinsieme di Python che conosciamo è ora **Turing-completo** (*Turing complete*), il che significa che possiamo eseguire qualsiasi calcolo che possa essere descritto da un algoritmo.

Per illustrare la ricorsione con valori di ritorno, valuteremo alcune funzioni matematiche definite ricorsivamente.
Una definizione ricorsiva è simile a una definizione circolare, nel senso che la definizione fa riferimento alla cosa che viene definita. Una definizione davvero circolare non è molto utile:

> vorpal: un aggettivo usato per descrivere qualcosa che è vorpal.

Se trovassi quella definizione nel dizionario, potresti irritarti.
D'altra parte, se cercassi la definizione della funzione fattoriale, indicata con il simbolo $!$, potresti trovare qualcosa del genere:

$$\begin{aligned}
0! &= 1 \\
n! &= n~(n-1)!
\end{aligned}$$

Questa definizione dice che il fattoriale di $0$ è $1$, e il fattoriale di qualsiasi altro valore, $n$, è $n$ moltiplicato per il fattoriale di $n-1$.

Se riesci a scrivere una definizione ricorsiva di qualcosa, puoi scrivere un programma Python che la valuti.
Seguendo un procedimento di sviluppo incrementale, inizieremo con una funzione che riceve `n` come parametro e restituisce sempre `0`.

```python
def factorial(n):
    return 0
```

Ora aggiungiamo la prima parte della definizione – se l'argomento è `0`, tutto ciò che dobbiamo fare è restituire `1`:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return 0
```

Ora completiamo la seconda parte – se `n` non è `0`, dobbiamo fare una chiamata ricorsiva
per trovare il fattoriale di `n-1` e poi moltiplicare il risultato per `n`:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        recurse = factorial(n-1)
        return n * recurse
```

Il flusso di esecuzione di questo programma è simile al flusso di `countdown` nel capitolo 5.
Se chiamiamo `factorial` con il valore `3`:

Poiché `3` non è `0`, prendiamo il secondo ramo e calcoliamo il fattoriale
di `n-1`\...

> Poiché `2` non è `0`, prendiamo il secondo ramo e calcoliamo il
> fattoriale di `n-1`\...
>
> > Poiché `1` non è `0`, prendiamo il secondo ramo e calcoliamo il
> > fattoriale di `n-1`\...
> >
> > > Poiché `0` è uguale a `0`, prendiamo il primo ramo e restituiamo `1` senza
> > > fare altre chiamate ricorsive.
> >
> > Il valore di ritorno, `1`, viene moltiplicato per `n`, che è `1`, e il
> > risultato viene restituito.
>
> Il valore di ritorno, `1`, viene moltiplicato per `n`, che è `2`, e il risultato
> viene restituito.

Il valore di ritorno `2` viene moltiplicato per `n`, che è `3`, e il risultato,
`6`, diventa il valore di ritorno della chiamata di funzione che ha dato inizio all'intero
processo.

La figura seguente mostra il diagramma dello stack per questa sequenza di chiamate di funzione.

```python
from diagram import Frame, Stack, make_binding

main = Frame([], name='__main__', loc='left')
frames = [main]

ns = 3, 2, 1
recurses = 2, 1, 1
results = 6, 2, 1

for n, recurse, result in zip(ns, recurses, results):
    binding1 = make_binding('n', n)
    binding2 = make_binding('recurse', recurse)
    frame = Frame([binding1, binding2], 
                  name='factorial', value=result,
                  loc='left', dx=1.2)
    frames.append(frame)
    
binding1 = make_binding('n', 0)
frame = Frame([binding1], name='factorial', value=1, 
              shim=1.2, loc='left', dx=1.4)
frames.append(frame)

stack = Stack(frames, dy=-0.45)
```

```python
from diagram import diagram, adjust

width, height, x, y = [2.74, 2.26, 0.73, 2.05]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)
```

I valori di ritorno sono mostrati mentre risalgono lo stack.
In ogni frame, il valore di ritorno è il prodotto di `n` e `recurse`.

Nell'ultimo frame, la variabile locale `recurse` non esiste perché il ramo che la crea non viene eseguito.

## Atto di fede

Seguire il flusso di esecuzione è un modo per leggere i programmi, ma può diventare rapidamente opprimente. Un'alternativa è quello che chiamo l'"atto di fede" (*leap of faith*). Quando arrivi a una chiamata di funzione, invece di seguire il flusso di esecuzione, *assumi* che la funzione funzioni correttamente e restituisca il risultato giusto.

In effetti, pratichi già questo atto di fede quando usi le funzioni predefinite.
Quando chiami `abs` o `math.sqrt`, non esamini il corpo di quelle funzioni – assumi semplicemente che funzionino.

Lo stesso vale quando chiami una delle tue funzioni. Per esempio, prima abbiamo scritto una funzione chiamata `is_divisible` che determina se un numero è divisibile per un altro. Una volta convinti che questa funzione è corretta, possiamo usarla senza riguardarne il corpo.

Lo stesso vale per i programmi ricorsivi.
Quando arrivi alla chiamata ricorsiva, invece di seguire il flusso di esecuzione, dovresti assumere che la chiamata ricorsiva funzioni e poi chiederti: "Supponendo di saper calcolare il fattoriale di $n-1$, so calcolare il fattoriale di $n$?"
La definizione ricorsiva del fattoriale implica che puoi, moltiplicando per $n$.

Certo, è un po' strano assumere che la funzione funzioni correttamente quando non hai ancora finito di scriverla, ma è per questo che si chiama atto di fede!

## Fibonacci

Dopo `factorial`, l'esempio più comune di funzione ricorsiva è `fibonacci`, che ha la seguente definizione:

$$\begin{aligned}
\mathrm{fibonacci}(0) &= 0 \\
\mathrm{fibonacci}(1) &= 1 \\
\mathrm{fibonacci}(n) &= \mathrm{fibonacci}(n-1) + \mathrm{fibonacci}(n-2)
\end{aligned}$$

Tradotta in Python, appare così:

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif  n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

Se provi a seguire il flusso di esecuzione qui, anche per piccoli valori di $n$, ti scoppia la testa.
Ma secondo l'atto di fede, se assumi che le due chiamate ricorsive funzionino correttamente, puoi essere sicuro che l'ultima istruzione `return` sia corretta.

Per inciso, questo modo di calcolare i numeri di Fibonacci è molto inefficiente.
Nel [capitolo 10](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap10.ipynb) spiegherò perché e suggerirò un modo per migliorarlo.

## Controllare i tipi

Cosa succede se chiamiamo `factorial` passandogli `1.5` come argomento?

```python
%%expect RecursionError

factorial(1.5)
```

Sembra una ricorsione infinita. Com'è possibile? La funzione ha casi base quando `n == 1` o `n == 0`.
Ma se `n` non è un intero, possiamo *mancare* il caso base e procedere con la ricorsione all'infinito.

In questo esempio, il valore iniziale di `n` è `1.5`.
Nella prima chiamata ricorsiva, il valore di `n` è `0.5`.
Nella successiva è `-0.5`.
Da lì diventa sempre più piccolo (più negativo), ma non sarà mai `0`.

Per evitare la ricorsione infinita possiamo usare la funzione predefinita `isinstance` per controllare il tipo dell'argomento.
Ecco come verifichiamo se un valore è un intero.

```python
isinstance(3, int)
```

```python
isinstance(1.5, int)
```

Ecco ora una versione di `factorial` con il controllo degli errori.

```python
def factorial(n):
    if not isinstance(n, int):
        print('factorial is only defined for integers.')
        return None
    elif n < 0:
        print('factorial is not defined for negative numbers.')
        return None
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

Per prima cosa verifica se `n` è un intero.
Se non lo è, visualizza un messaggio di errore e restituisce `None`.

```python
factorial('crunchy frog')
```

Poi verifica se `n` è negativo.
Se lo è, visualizza un messaggio di errore e restituisce `None`.

```python
factorial(-2)
```

Se superiamo entrambi i controlli, sappiamo che `n` è un intero non negativo, quindi possiamo essere sicuri che la ricorsione terminerà.
Controllare i parametri di una funzione per assicurarsi che abbiano i tipi e i valori corretti si chiama **validazione dell'input** (*input validation*).

## Debugging

Dividere un programma grande in funzioni più piccole crea dei punti di controllo naturali per il debugging.
Se una funzione non funziona, ci sono tre possibilità da considerare:

-   C'è qualcosa di sbagliato negli argomenti che la funzione riceve – cioè, una precondizione è violata.

-   C'è qualcosa di sbagliato nella funzione – cioè, una postcondizione è violata.

-   Il chiamante fa qualcosa di sbagliato con il valore di ritorno.

Per escludere la prima possibilità, puoi aggiungere un'istruzione `print` all'inizio della funzione che visualizzi i valori dei parametri (e magari i loro tipi).
Oppure puoi scrivere codice che controlli esplicitamente le precondizioni.

Se i parametri sembrano a posto, puoi aggiungere un'istruzione `print` prima di ogni istruzione `return` e visualizzare il valore di ritorno.
Se possibile, chiama la funzione con argomenti che rendano facile controllare il risultato.

Se la funzione sembra funzionare, guarda la chiamata di funzione per assicurarti che il valore di ritorno venga usato correttamente – o che venga usato affatto!

Aggiungere istruzioni `print` all'inizio e alla fine di una funzione può aiutare a rendere più visibile il flusso di esecuzione.
Per esempio, ecco una versione di `factorial` con delle istruzioni print:

```python
def factorial(n):
    space = ' ' * (4 * n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
```

`space` è una stringa di caratteri spazio che controlla l'indentazione
dell'output. Ecco il risultato di `factorial(3)`:

```python
factorial(3)
```

Se il flusso di esecuzione ti confonde, questo tipo di output può essere utile.
Ci vuole un po' di tempo per sviluppare un'impalcatura efficace, ma un po' di impalcatura può far risparmiare molto debugging.

## Glossario

**valore di ritorno** (*return value*):
Il risultato di una funzione. Se una chiamata di funzione è usata come espressione, il valore di ritorno è il valore dell'espressione.

**funzione pura** (*pure function*):
Una funzione che non visualizza nulla e non ha altri effetti, oltre a restituire un valore di ritorno.

**codice morto** (*dead code*):
Parte di un programma che non può mai essere eseguita, spesso perché si trova dopo un'istruzione `return`.

**sviluppo incrementale** (*incremental development*):
Un piano di sviluppo dei programmi pensato per evitare il debugging aggiungendo e testando solo una piccola quantità di codice alla volta.

**impalcatura** (*scaffolding*):
Codice usato durante lo sviluppo del programma ma che non fa parte della versione finale.

**Turing-completo** (*Turing complete*):
