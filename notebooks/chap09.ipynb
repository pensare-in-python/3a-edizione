## Una lista è una sequenza

Come una stringa, una **lista** (*list*) è una sequenza di valori. In una stringa i
valori sono caratteri; in una lista possono essere di qualsiasi tipo.
I valori di una lista si chiamano **elementi** (*elements*).

Ci sono diversi modi per creare una nuova lista; il più semplice è racchiudere gli elementi tra parentesi quadre (`[` e `]`).
Per esempio, ecco una lista di due interi.

```python
numbers = [42, 123]
```

Ed ecco una lista di tre stringhe.

```python
cheeses = ['Cheddar', 'Edam', 'Gouda']
```

Gli elementi di una lista non devono essere dello stesso tipo.
La lista seguente contiene una stringa, un float, un intero e persino un'altra lista.

```python
t = ['spam', 2.0, 5, [10, 20]]
```

Una lista all'interno di un'altra lista è **annidata** (*nested*).

Una lista che non contiene elementi si chiama lista vuota; puoi crearne
una con le parentesi quadre vuote, `[]`.

```python
empty = []
```

La funzione `len` restituisce la lunghezza di una lista.

```python
len(cheeses)
```

La lunghezza di una lista vuota è `0`.

```python
len(empty)
```

La figura seguente mostra il diagramma di stato per `cheeses`, `numbers` ed `empty`.

```python
from diagram import make_list, Binding, Value

list1 = make_list(cheeses, dy=-0.3, offsetx=0.17)
binding1 = Binding(Value('cheeses'), list1)

list2 = make_list(numbers, dy=-0.3, offsetx=0.17)
binding2 = Binding(Value('numbers'), list2)

list3 = make_list(empty, dy=-0.3, offsetx=0.1)
binding3 = Binding(Value('empty'), list3)
```

Le liste sono rappresentate da riquadri con la parola "list" all'esterno e gli elementi numerati della lista all'interno.

## Le liste sono mutabili

Per leggere un elemento di una lista, possiamo usare l'operatore parentesi quadre.
L'indice del primo elemento è `0`.

```python
cheeses[0]
```

A differenza delle stringhe, le liste sono mutabili. Quando l'operatore parentesi quadre compare
a sinistra di un'assegnazione, identifica l'elemento della lista
che verrà assegnato.

```python
numbers[1] = 17
numbers
```

Il secondo elemento di `numbers`, che prima era `123`, ora è `17`.

Gli indici delle liste funzionano come quelli delle stringhe:

-   Qualsiasi espressione intera può essere usata come indice.

-   Se provi a leggere o scrivere un elemento che non esiste, ottieni
    un `IndexError`.

-   Se un indice ha valore negativo, conta all'indietro dalla fine
    della lista.

L'operatore `in` funziona sulle liste – verifica se un dato elemento compare in un punto qualsiasi della lista.

```python
'Edam' in cheeses
```

```python
'Wensleydale' in cheeses
```

Anche se una lista può contenere un'altra lista, la lista annidata conta comunque come un solo elemento – quindi nella lista seguente ci sono solo quattro elementi.

```python
t = ['spam', 2.0, 5, [10, 20]]
len(t)
```

E `10` non è considerato un elemento di `t` perché è un elemento di una lista annidata, non di `t`.

```python
10 in t
```

## Slice di liste

L'operatore di slice funziona sulle liste come sulle stringhe.
L'esempio seguente seleziona il secondo e il terzo elemento di una lista di quattro lettere.

```python
letters = ['a', 'b', 'c', 'd']
letters[1:3]
```

Se ometti il primo indice, la slice parte dall'inizio.

```python
letters[:2]
```

Se ometti il secondo, la slice arriva fino alla fine.

```python
letters[2:]
```

Quindi, se li ometti entrambi, la slice è una copia dell'intera lista.

```python
letters[:]
```

Un altro modo per copiare una lista è usare la funzione `list`.

```python
list(letters)
```

Poiché `list` è il nome di una funzione predefinita, dovresti evitare di usarlo come nome di variabile.

## Liste e stringhe

Una stringa è una sequenza di caratteri e una lista è una sequenza di valori,
ma una lista di caratteri non è la stessa cosa di una stringa.
Per convertire una stringa in una lista di caratteri, puoi usare la funzione `list`.

```python
s = 'spam'
t = list(s)
t
```

La funzione `list` spezza una stringa nelle singole lettere.
Se vuoi spezzare una stringa in parole, puoi usare il metodo `split`:

```python
s = 'pining for the fjords'
t = s.split()
t
```

Un argomento facoltativo chiamato **delimitatore** (*delimiter*) specifica quali caratteri usare come confini tra le parole. L'esempio seguente usa un trattino come delimitatore.

```python
s = 'ex-parrot'
t = s.split('-')
t
```

Se hai una lista di stringhe, puoi concatenarle in un'unica stringa usando `join`.
`join` è un metodo delle stringhe, quindi devi invocarlo sul delimitatore e passare la lista come argomento.

```python
delimiter = ' '
t = ['pining', 'for', 'the', 'fjords']
s = delimiter.join(t)
s
```

In questo caso il delimitatore è un carattere spazio, quindi `join` mette uno spazio
tra le parole.
Per unire stringhe senza spazi, puoi usare la stringa vuota, `''`, come delimitatore.

## Oggetti e valori

Se eseguiamo queste istruzioni di assegnazione:

```python
a = 'banana'
b = 'banana'
```

Sappiamo che `a` e `b` fanno entrambe riferimento a una stringa, ma non sappiamo se fanno riferimento alla *stessa* stringa.
Ci sono due stati possibili, mostrati nella figura seguente.

```python
from diagram import Frame, Stack

s = 'banana'
bindings = [Binding(Value(name), Value(repr(s))) for name in 'ab']
frame1 = Frame(bindings, dy=-0.25)

binding1 = Binding(Value('a'), Value(repr(s)), dy=-0.11)
binding2 = Binding(Value('b'), draw_value=False, dy=0.11)
frame2 = Frame([binding1, binding2], dy=-0.25)

stack = Stack([frame1, frame2], dx=1.7, dy=0)
```

```python
width, height, x, y = [2.85, 0.76, 0.17, 0.51]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)
```

Nel diagramma a sinistra, `a` e `b` fanno riferimento a due oggetti diversi che hanno lo
stesso valore. Nel diagramma a destra, fanno riferimento allo stesso oggetto.
Per verificare se due variabili fanno riferimento allo stesso oggetto, puoi usare l'operatore `is`.

```python
a = 'banana'
b = 'banana'
a is b
```

In questo esempio, Python ha creato un solo oggetto stringa, e sia `a`
sia `b` fanno riferimento a esso.
Ma quando crei due liste, ottieni due oggetti.

```python
a = [1, 2, 3]
b = [1, 2, 3]
a is b
```

Quindi il diagramma di stato appare così.

```python
t = [1, 2, 3]
binding1 = Binding(Value('a'), Value(repr(t)))
binding2 = Binding(Value('b'), Value(repr(t)))
frame = Frame([binding1, binding2], dy=-0.25)
```

```python
width, height, x, y = [1.16, 0.76, 0.21, 0.51]
ax = diagram(width, height)
bbox = frame.draw(ax, x, y)
# adjust(x, y, bbox)
```

In questo caso diremmo che le due liste sono **equivalenti** (*equivalent*), perché hanno gli stessi elementi, ma non **identiche** (*identical*), perché non sono lo stesso oggetto.
Se due oggetti sono identici, sono anche equivalenti, ma se sono equivalenti non sono necessariamente identici.

## Aliasing

Se `a` fa riferimento a un oggetto e assegni `b = a`, allora entrambe le variabili fanno riferimento allo stesso oggetto.

```python
a = [1, 2, 3]
b = a
b is a
```

Quindi il diagramma di stato appare così.

```python
t = [1, 2, 3]
binding1 = Binding(Value('a'), Value(repr(t)), dy=-0.11)
binding2 = Binding(Value('b'), draw_value=False, dy=0.11)
frame = Frame([binding1, binding2], dy=-0.25)
```

```python
width, height, x, y = [1.11, 0.81, 0.17, 0.56]
ax = diagram(width, height)
bbox = frame.draw(ax, x, y)
# adjust(x, y, bbox)
```

L'associazione di una variabile con un oggetto si chiama **riferimento** (*reference*).
In questo esempio ci sono due riferimenti allo stesso oggetto.

Un oggetto con più di un riferimento ha più di un nome, quindi diciamo che l'oggetto ha degli **alias** (*aliased*).
Se l'oggetto con alias è mutabile, le modifiche fatte con un nome si riflettono sull'altro.
In questo esempio, se modifichiamo l'oggetto a cui fa riferimento `b`, stiamo modificando anche l'oggetto a cui fa riferimento `a`.

```python
b[0] = 5
a
```

Quindi diremmo che `a` "vede" questa modifica.
Anche se questo comportamento può essere utile, è soggetto a errori.
In generale, è più sicuro evitare l'aliasing quando lavori con oggetti mutabili.

Per gli oggetti immutabili come le stringhe, l'aliasing non è un gran problema.
In questo esempio:

```python
a = 'banana'
b = 'banana'
```

Non fa quasi mai differenza se `a` e `b` fanno riferimento alla stessa
stringa o no.

## Liste come argomenti

Quando passi una lista a una funzione, la funzione riceve un riferimento alla
lista. Se la funzione modifica la lista, il chiamante vede la modifica. Per
esempio, `pop_first` usa il metodo delle liste `pop` per rimuovere il primo elemento da una lista.

```python
def pop_first(lst):
    return lst.pop(0)
```

Possiamo usarla così.

```python
letters = ['a', 'b', 'c']
pop_first(letters)
```

Il valore di ritorno è il primo elemento, che è stato rimosso dalla lista – come possiamo vedere visualizzando la lista modificata.

```python
letters
```

In questo esempio, il parametro `lst` e la variabile `letters` sono alias dello stesso oggetto, quindi il diagramma di stato appare così:

```python
lst = make_list('abc', dy=-0.3, offsetx=0.1)
binding1 = Binding(Value('letters'), draw_value=False)
frame1 = Frame([binding1], name='__main__', loc='left')

binding2 = Binding(Value('lst'), draw_value=False, dx=0.61, dy=0.35)
frame2 = Frame([binding2], name='pop_first', loc='left', offsetx=0.08)

stack = Stack([frame1, frame2], dx=-0.3, dy=-0.5)
```

```python
width, height, x, y = [2.04, 1.24, 1.06, 0.85]
ax = diagram(width, height)
bbox1 = stack.draw(ax, x, y)
bbox2 = lst.draw(ax, x+0.5, y)
bbox = Bbox.union([bbox1, bbox2])
adjust(x, y, bbox)
```

Passare un riferimento a un oggetto come argomento di una funzione crea una forma di aliasing.
Se la funzione modifica l'oggetto, le modifiche persistono dopo che la funzione è terminata.

## Debugging

Nota che la maggior parte dei metodi delle liste modifica l'argomento e restituisce `None`.
È il contrario dei metodi delle stringhe, che restituiscono una nuova stringa e lasciano intatta l'originale.

Se sei abituato a scrivere codice con le stringhe come questo:

```python
word = 'plumage!'
word = word.strip('!')
word
```

Si potrebbe essere tentati di scrivere codice con le liste come questo:

```python
t = [1, 2, 3]
t = t.remove(3)           # WRONG!
```

`remove` modifica la lista e restituisce `None`, quindi è probabile che la prossima operazione che esegui con `t` fallisca.

```python
%%expect AttributeError

t.remove(2)
```

Questo messaggio di errore richiede qualche spiegazione.
Un **attributo** (*attribute*) di un oggetto è una variabile o un metodo associato a esso.
In questo caso, il valore di `t` è `None`, che è un oggetto di tipo `NoneType`, che non ha un attributo chiamato `remove`, quindi il risultato è un `AttributeError`.

Se vedi un messaggio di errore come questo, dovresti ripercorrere all'indietro il programma e vedere se per caso hai chiamato un metodo delle liste in modo sbagliato.

## Glossario

**lista** (*list*):
Un oggetto che contiene una sequenza di valori.

**elemento** (*element*):
Uno dei valori di una lista o di un'altra sequenza.

**lista annidata** (*nested list*):
Una lista che è un elemento di un'altra lista.

**delimitatore** (*delimiter*):
Un carattere o una stringa usata per indicare dove una stringa va spezzata.

**equivalenti** (*equivalent*):
Che hanno lo stesso valore.

**identici** (*identical*):
Che sono lo stesso oggetto (il che implica l'equivalenza).

**riferimento** (*reference*):
L'associazione tra una variabile e il suo valore.

**con alias** (*aliased*):
Se più di una variabile fa riferimento a un oggetto, l'oggetto ha degli alias.

**attributo** (*attribute*):
Uno dei valori con nome associati a un oggetto.
