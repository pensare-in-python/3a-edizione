## Che cos'è un notebook?

Un notebook Jupyter è formato da celle, e ogni cella contiene testo oppure codice.
Questa cella contiene testo.

La cella seguente contiene codice.

```python
print('Hello')
```

Fai clic sulla cella precedente per selezionarla.
Dovresti vedere a sinistra un pulsante con un triangolo dentro un cerchio, che è l'icona di "Play" (esegui).
Se premi questo pulsante, Jupyter esegue il codice della cella e visualizza il risultato.

Quando esegui codice in un notebook per la prima volta, l'avvio potrebbe richiedere qualche secondo.
E se è un notebook che non hai scritto tu, potresti ricevere un messaggio di avviso.
Se stai eseguendo un notebook da una fonte di cui ti fidi, che spero includa me, puoi premere "Run Anyway" (esegui comunque).

Invece di fare clic sul pulsante "Play", puoi eseguire il codice di una cella anche tenendo premuto `Shift` e premendo `Invio`.

Se stai eseguendo questo notebook su Colab, dovresti vedere in alto a sinistra dei pulsanti con scritto "+ Code" e "+ Text". Il primo aggiunge una cella di codice e il secondo una cella di testo.
Se vuoi provarli, seleziona questa cella facendo clic su di essa, poi premi il pulsante "+ Text".
Sotto questa dovrebbe comparire una nuova cella.

Aggiungi del testo alla cella.
Puoi usare i pulsanti per formattarlo, oppure puoi marcare il testo usando [Markdown](https://www.markdownguide.org/basic-syntax/).
Quando hai finito, tieni premuto `Shift` e premi `Invio`: il testo appena scritto verrà formattato e passerai alla cella successiva.

In ogni momento Jupyter si trova in una di due modalità:

* In **modalità comando** (*command mode*) puoi eseguire operazioni sulle celle, come aggiungere e rimuovere celle intere.

* In **modalità modifica** (*edit mode*) puoi modificare il contenuto di una cella.

Con le celle di testo è evidente in quale modalità ti trovi.
In modalità modifica la cella è divisa verticalmente, con il testo che stai modificando a sinistra e il testo formattato a destra.
E in alto vedrai gli strumenti di modifica del testo.
In modalità comando vedi solo il testo formattato.

Con le celle di codice la differenza è meno evidente, ma se nella cella c'è un cursore, sei in modalità modifica.

Per passare dalla modalità modifica alla modalità comando, premi `ESC`.
Per passare dalla modalità comando alla modalità modifica, premi `Invio`.

Quando hai finito di lavorare su un notebook, puoi chiudere la finestra, ma le modifiche che hai fatto andranno perse.
Se hai fatto modifiche che vuoi conservare, apri il menu File in alto a sinistra.
Vedrai diversi modi per salvare il notebook.

* Se hai un account Google, puoi salvare il notebook nel tuo Drive.

* Se hai un account GitHub, puoi salvarlo su GitHub.

* Oppure, se vuoi salvare il notebook sul tuo computer, seleziona "Download" e poi "Download .ipynb". Il suffisso ".ipynb" indica che si tratta di un file notebook, a differenza di ".py", che indica un file che contiene solo codice Python.

## Il codice di *Think Python*

All'inizio di ogni notebook vedrai una cella con codice come questo:

```python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/thinkpython.py')

import thinkpython
```

Non devi sapere come funziona questo codice, ma quando arriverai alla fine del libro la maggior parte ti sarà chiara.
Come puoi immaginare, scarica un file – in particolare scarica `thinkpython.py`, che contiene codice Python fornito apposta per questo libro.
L'ultima riga "importa" questo codice, il che significa che possiamo usarlo nel notebook.

In altri capitoli vedrai codice che scarica `diagram.py`, che serve a generare i diagrammi del libro, e `jupyturtle.py`, che si usa in diversi capitoli per creare la grafica della tartaruga.

In alcuni punti vedrai una cella come questa, che inizia con `%%expect`.

```python
%%expect SyntaxError

abs 42
```

`%%expect` non fa parte di Python – è un "comando magico" di Jupyter che indica che ci aspettiamo che la cella produca un errore.
Quando vedi questo comando, significa che l'errore è voluto, di solito per avvertirti di una trappola comune.
