## Operatori aritmetici

Un **operatore aritmetico** (*arithmetic operator*) è un simbolo che rappresenta un calcolo aritmetico. Per esempio, il segno più, `+`, esegue l'addizione.

```python
30 + 12
```

Il segno meno, `-`, è l'operatore che esegue la sottrazione.

```python
43 - 1
```

L'asterisco, `*`, esegue la moltiplicazione.

```python
6 * 7
```

E la barra, `/`, esegue la divisione:

```python
84 / 2
```

Nota che il risultato della divisione è `42.0` e non `42`. Questo perché in Python esistono due tipi di numeri:

* gli **interi** (*integers*), che rappresentano numeri senza parte frazionaria o decimale, e

* i **numeri in virgola mobile** (*floating-point numbers*), che rappresentano interi e numeri con la virgola (in Python si scrive il punto decimale).

Se sommi, sottrai o moltiplichi due interi, il risultato è un intero.
Ma se dividi due interi, il risultato è un numero in virgola mobile.
Python fornisce un altro operatore, `//`, che esegue la **divisione intera** (*integer division*).
Il risultato della divisione intera è sempre un intero.

```python
84 // 2
```

La divisione intera è chiamata anche "floor division" (divisione "verso il pavimento") perché arrotonda sempre per difetto (verso il "pavimento").

```python
85 // 2
```

Infine, l'operatore `**` esegue l'elevamento a potenza; cioè eleva un
numero a una potenza:

```python
7 ** 2
```

In alcuni altri linguaggi si usa l'accento circonflesso, `^`, per l'elevamento a potenza, ma in Python
è un operatore bit a bit chiamato XOR.
Se non conosci gli operatori bit a bit, il risultato potrebbe sorprenderti:

```python
7 ^ 2
```

In questo libro non tratterò gli operatori bit a bit, ma puoi leggerne
qualcosa su <http://wiki.python.org/moin/BitwiseOperators>.

## Espressioni

Un insieme di operatori e numeri si chiama **espressione** (*expression*).
Un'espressione può contenere un numero qualsiasi di operatori e numeri.
Per esempio, ecco un'espressione che contiene due operatori.

```python
6 + 6 ** 2
```

Nota che l'elevamento a potenza avviene prima dell'addizione.
Python segue l'ordine delle operazioni che forse hai imparato a lezione di matematica: l'elevamento a potenza viene prima della moltiplicazione e della divisione, che vengono prima dell'addizione e della sottrazione.

Nell'esempio seguente, la moltiplicazione avviene prima dell'addizione.

```python
12 + 5 * 6
```

Se vuoi che l'addizione avvenga per prima, puoi usare le parentesi.

```python
(12 + 5) * 6
```

Ogni espressione ha un **valore** (*value*).
Per esempio, l'espressione `6 * 7` ha il valore `42`.

## Funzioni aritmetiche

Oltre agli operatori aritmetici, Python fornisce alcune **funzioni** (*functions*) che lavorano con i numeri.
Per esempio, la funzione `round` prende un numero in virgola mobile e lo arrotonda all'intero più vicino.

```python
round(42.4)
```

```python
round(42.6)
```

La funzione `abs` calcola il valore assoluto di un numero.
Per un numero positivo, il valore assoluto è il numero stesso.

```python
abs(42)
```

Per un numero negativo, il valore assoluto è positivo.

```python
abs(-42)
```

Quando usiamo una funzione in questo modo, diciamo che stiamo **chiamando** la funzione.
Un'espressione che chiama una funzione è una **chiamata di funzione** (*function call*).

Quando chiami una funzione, le parentesi sono obbligatorie.
Se le ometti, ottieni un messaggio di errore.

## Stringhe

Oltre ai numeri, Python può rappresentare anche sequenze di lettere, chiamate **stringhe** (*strings*) perché le lettere sono infilate una dopo l'altra come le perle di una collana (in inglese *string* è anche il filo).
Per scrivere una stringa, possiamo mettere una sequenza di lettere tra apici dritti.

```python
'Hello'
```

È valido anche usare le virgolette doppie.

```python
"world"
```

Le virgolette doppie rendono facile scrivere una stringa che contiene un apostrofo, che è lo stesso simbolo dell'apice dritto.

```python
"it's a small "
```

Le stringhe possono contenere anche spazi, punteggiatura e cifre.

```python
'Well, '
```

L'operatore `+` funziona con le stringhe; unisce due stringhe in un'unica stringa, operazione che si chiama **concatenazione** (*concatenation*).

```python
'Well, ' + "it's a small " + 'world.'
```

Anche l'operatore `*` funziona con le stringhe; crea più copie di una stringa e le concatena.

```python
'Spam, ' * 4
```

Gli altri operatori aritmetici non funzionano con le stringhe.

Python fornisce una funzione chiamata `len` che calcola la lunghezza di una stringa.

```python
len('Spam')
```

Nota che `len` conta le lettere tra gli apici, ma non gli apici.

Quando crei una stringa, assicurati di usare apici dritti.
L'apice inverso, detto anche backtick, causa un errore di sintassi.

```python
%%expect SyntaxError

`Hello`
```

Anche le virgolette "intelligenti" (tipografiche, o curve) non sono valide.

```python
%%expect SyntaxError

‘Hello’
```

## Valori e tipi

Finora abbiamo visto tre tipi di valori:

* `2` è un intero,

* `42.0` è un numero in virgola mobile, e

* `'Hello'` è una stringa.

Un genere di valore si chiama **tipo** (*type*).
Ogni valore ha un tipo – o, come a volte diciamo, "appartiene a" un tipo.

Python fornisce una funzione chiamata `type` che ti dice il tipo di qualsiasi valore.
Il tipo di un intero è `int`.

```python
type(2)
```

Il tipo di un numero in virgola mobile è `float`.

```python
type(42.0)
```

E il tipo di una stringa è `str`.

```python
type('Hello, World!')
```

I tipi `int`, `float` e `str` si possono usare come funzioni.
Per esempio, `int` può prendere un numero in virgola mobile e convertirlo in un intero (arrotondando sempre per difetto).

```python
int(42.9)
```

E `float` può convertire un intero in un valore in virgola mobile.

```python
float(42)
```

Ora, ecco una cosa che può creare confusione.
Che cosa ottieni se metti una sequenza di cifre tra apici?

```python
'126'
```

Sembra un numero, ma in realtà è una stringa.

```python
type('126')
```

Se provi a usarla come un numero, potresti ottenere un errore.

```python
%%expect TypeError

'126' / 3
```

Questo esempio genera un `TypeError`, il che significa che i valori nell'espressione, chiamati **operandi** (*operands*), hanno il tipo sbagliato.
Il messaggio di errore indica che l'operatore `/` non supporta i tipi di questi valori, che sono `str` e `int`.

Se hai una stringa che contiene cifre, puoi usare `int` per convertirla in un intero.

```python
int('126') / 3
```

Se hai una stringa che contiene cifre e un punto decimale, puoi usare `float` per convertirla in un numero in virgola mobile.

```python
float('12.6')
```

Quando scrivi un intero grande, potresti essere tentato di usare le virgole
tra i gruppi di cifre, come in `1,000,000` (all'inglese).
In Python è un'espressione valida, ma il risultato non è un intero.

```python
1,000,000
```

Python interpreta `1,000,000` come una sequenza di interi separati da virgole.
Impareremo di più su questo tipo di sequenza più avanti.

Puoi usare i trattini bassi (underscore) per rendere più leggibili i numeri grandi.

```python
1_000_000
```

## Linguaggi formali e naturali

I **linguaggi naturali** (*natural languages*) sono le lingue che le persone parlano, come l'inglese, lo spagnolo e il francese. Non sono stati progettati da qualcuno; si sono evoluti naturalmente.

I **linguaggi formali** (*formal languages*) sono linguaggi progettati dalle persone per applicazioni specifiche.
Per esempio, la notazione usata dai matematici è un linguaggio formale particolarmente adatto a esprimere relazioni tra numeri e simboli.
Allo stesso modo, i linguaggi di programmazione sono linguaggi formali progettati per esprimere calcoli.

Anche se i linguaggi formali e quelli naturali hanno alcune caratteristiche
in comune, ci sono differenze importanti:

* Ambiguità: i linguaggi naturali sono pieni di ambiguità, che le persone gestiscono
    usando indizi dal contesto e altre informazioni. I linguaggi formali sono
    progettati per essere quasi o del tutto privi di ambiguità, il che significa che
    ogni programma ha esattamente un significato, indipendentemente dal contesto.

* Ridondanza: per compensare l'ambiguità e ridurre i fraintendimenti,
    i linguaggi naturali usano la ridondanza. Di conseguenza, sono
    spesso prolissi. I linguaggi formali sono meno ridondanti e più concisi.

* Letteralità: i linguaggi naturali sono pieni di modi di dire e metafore. I linguaggi formali significano esattamente ciò che dicono.

Poiché tutti cresciamo parlando linguaggi naturali, a volte è difficile abituarsi ai linguaggi formali.
I linguaggi formali sono più densi di quelli naturali, quindi ci vuole più tempo per leggerli.
Inoltre, la struttura è importante, quindi non sempre conviene leggere dall'alto in basso e da sinistra a destra.
Infine, i dettagli contano. Piccoli errori di ortografia e
punteggiatura, che nei linguaggi naturali passano inosservati, possono fare
una grande differenza in un linguaggio formale.

## Debugging

Programmare, e soprattutto fare debugging, a volte suscita emozioni forti. Se stai lottando con un bug difficile, potresti sentirti arrabbiato, triste o in imbarazzo.

Prepararsi a queste reazioni può aiutarti a gestirle. Un approccio è pensare al computer come a un dipendente con certi punti di forza, come la velocità e la precisione, e particolari debolezze, come la mancanza di empatia e l'incapacità di vedere il quadro generale.

## Glossario

**operatore aritmetico** (*arithmetic operator*):
Un simbolo, come `+` e `*`, che indica un'operazione aritmetica come l'addizione o la moltiplicazione.

**intero** (*integer*):
Un tipo che rappresenta numeri senza parte frazionaria o decimale.

**virgola mobile** (*floating-point*):
Un tipo che rappresenta interi e numeri con parte decimale.

**divisione intera** (*integer division*):
Un operatore, `//`, che divide due numeri e arrotonda per difetto a un intero.

**espressione** (*expression*):
Una combinazione di variabili, valori e operatori.

**valore** (*value*):
Un intero, un numero in virgola mobile o una stringa – o uno degli altri tipi di valori che vedremo più avanti.

**funzione** (*function*):
Una sequenza di istruzioni, dotata di un nome, che esegue qualche operazione utile.
Le funzioni possono ricevere o meno degli argomenti e possono produrre o meno un risultato.

**chiamata di funzione** (*function call*):
Un'espressione – o parte di un'espressione – che esegue una funzione.
È formata dal nome della funzione seguito da un elenco di argomenti tra parentesi.

**errore di sintassi** (*syntax error*):
Un errore in un programma che ne rende impossibile l'analisi (parsing) – e quindi impossibile l'esecuzione.

**stringa** (*string*):
Un tipo che rappresenta sequenze di caratteri.

**concatenazione** (*concatenation*):
L'unione di due stringhe una dopo l'altra.

**tipo** (*type*):
Una categoria di valori.
I tipi visti finora sono gli interi (tipo `int`), i numeri in virgola mobile (tipo `float`) e le stringhe (tipo `str`).

**operando** (*operand*):
Uno dei valori su cui agisce un operatore.

**linguaggio naturale** (*natural language*):
Una qualsiasi delle lingue parlate dalle persone, che si è evoluta naturalmente.

**linguaggio formale** (*formal language*):
Uno qualsiasi dei linguaggi progettati dalle persone per scopi specifici, come rappresentare idee matematiche o programmi per computer.
Tutti i linguaggi di programmazione sono linguaggi formali.

**bug**:
Un errore in un programma.

**debugging**:
Il processo di individuazione e correzione degli errori.
