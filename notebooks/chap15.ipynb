# Classi e metodi

[Esegui questo notebook su Colab](https://colab.research.google.com/github/pensare-in-python/3a-edizione/blob/main/notebooks/chap15.ipynb)

Python è un **linguaggio orientato agli oggetti** (*object-oriented language*) – cioè fornisce funzionalità che supportano la programmazione orientata agli oggetti, che ha queste caratteristiche distintive:

-   La maggior parte del calcolo è espressa in termini di operazioni sugli oggetti.

-   Gli oggetti spesso rappresentano cose del mondo reale, e i metodi spesso corrispondono ai modi in cui le cose del mondo reale interagiscono.

-   I programmi includono definizioni di classi e di metodi.

Per esempio, nel capitolo precedente abbiamo definito una classe `Time` che corrisponde al modo in cui le persone registrano l'ora del giorno, e abbiamo definito funzioni che corrispondono alle cose che le persone fanno con gli orari.
Ma non c'era alcun collegamento esplicito tra la definizione della classe `Time` e le definizioni di funzione che la seguivano.
Possiamo rendere esplicito il collegamento riscrivendo una funzione come **metodo** (*method*), che viene definito all'interno di una definizione di classe.

## Definire metodi

Nel capitolo precedente abbiamo definito una classe chiamata `Time` e scritto una funzione chiamata `print_time` che visualizza un'ora del giorno.

```python
class Time:
    """Represents the time of day."""

def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

Per trasformare `print_time` in un metodo, tutto ciò che dobbiamo fare è spostare la definizione
della funzione all'interno della definizione della classe. Nota il cambiamento
di indentazione.

Allo stesso tempo, cambieremo il nome del parametro da `time` a `self`.
Questo cambiamento non è necessario, ma per convenzione il primo parametro di un metodo si chiama `self`.

```python
class Time:
    """Represents the time of day."""    

    def print_time(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        print(s)
```

Per chiamare questo metodo, devi passare un oggetto `Time` come argomento.
Ecco la funzione che useremo per creare un oggetto `Time`.

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

Ed ecco un'istanza di `Time`.

```python
start = make_time(9, 40, 0)
```

Ora ci sono due modi per chiamare `print_time`. Il primo (e meno comune)
è usare la sintassi da funzione.

```python
Time.print_time(start)
```

In questa versione, `Time` è il nome della classe, `print_time` è il nome del metodo e `start` viene passato come parametro.
Il secondo modo (più idiomatico) è usare la sintassi da metodo:

```python
start.print_time()
```

In questa versione, `start` è l'oggetto su cui viene invocato il metodo, che si chiama **ricevente** (*receiver*), in base all'analogia secondo cui invocare un metodo è come inviare un messaggio a un oggetto.

Indipendentemente dalla sintassi, il comportamento del metodo è lo stesso.
Il ricevente viene assegnato al primo parametro, quindi all'interno del metodo `self` fa riferimento allo stesso oggetto di `start`.

## Un altro metodo

Ecco la funzione `time_to_int` del capitolo precedente.

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

Ed ecco una versione riscritta come metodo.

```python
%%add_method_to Time

    def time_to_int(self):
        minutes = self.hour * 60 + self.minute
        seconds = minutes * 60 + self.second
        return seconds
```

La prima riga usa il comando speciale `add_method_to`, che aggiunge un metodo a una classe definita in precedenza.
Questo comando funziona in un notebook Jupyter, ma non fa parte di Python, quindi non funzionerà in altri ambienti.
Normalmente, tutti i metodi di una classe si trovano all'interno della definizione della classe, quindi vengono definiti insieme alla classe.
Ma per questo libro è utile definire un metodo alla volta.

Come nell'esempio precedente, la definizione del metodo è indentata e il nome del parametro è `self`.
A parte questo, il metodo è identico alla funzione.
Ecco come lo invochiamo.

```python
start.time_to_int()
```

Si dice comunemente che si "chiama" una funzione e si "invoca" un metodo, ma significano la stessa cosa.

## Metodi statici

Come altro esempio, consideriamo la funzione `int_to_time`.
Ecco la versione del capitolo precedente.

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

Questa funzione riceve `seconds` come parametro e restituisce un nuovo oggetto `Time`.
Se la trasformiamo in un metodo della classe `Time`, dobbiamo invocarla su un oggetto `Time`.
Ma se stiamo cercando di creare un nuovo oggetto `Time`, su cosa dovremmo invocarla?

Possiamo risolvere questo problema dell'uovo e della gallina usando un **metodo statico** (*static method*), cioè un metodo che non richiede un'istanza della classe per essere invocato.
Ecco come riscriviamo questa funzione come metodo statico.

```python
%%add_method_to Time

    def int_to_time(seconds):
        minute, second = divmod(seconds, 60)
        hour, minute = divmod(minute, 60)
        return make_time(hour, minute, second)
```

Poiché è un metodo statico, non ha `self` come parametro.
Per invocarlo, usiamo `Time`, che è l'oggetto classe.

```python
start = Time.int_to_time(34800)
```

Il risultato è un nuovo oggetto che rappresenta le 9:40.

```python
start.print_time()
```

Ora che abbiamo `Time.int_to_time`, possiamo usarlo per scrivere `add_time` come metodo.
Ecco la funzione del capitolo precedente.

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

Ed ecco una versione riscritta come metodo.

```python
%%add_method_to Time

    def add_time(self, hours, minutes, seconds):
        duration = make_time(hours, minutes, seconds)
        seconds = Time.time_to_int(self) + Time.time_to_int(duration)
        return Time.int_to_time(seconds)
```

`add_time` ha `self` come parametro perché non è un metodo statico.
È un metodo ordinario – chiamato anche **metodo d'istanza** (*instance method*).
Per invocarlo, ci serve un'istanza di `Time`.

```python
end = start.add_time(1, 32, 0)
end.print_time()
```

## Confrontare oggetti Time

Come ulteriore esempio, scriviamo `is_after` come metodo.
Ecco la funzione `is_after`, che è la soluzione di un esercizio del capitolo precedente.

```python
def is_after(t1, t2):
    return time_to_int(t1) > time_to_int(t2)
```

Ed eccola come metodo.

```python
%%add_method_to Time

    def is_after(self, other):
        return self.time_to_int() > other.time_to_int()
```

Poiché stiamo confrontando due oggetti e il primo parametro è `self`, chiameremo il secondo parametro `other` (altro).
Per usare questo metodo, dobbiamo invocarlo su un oggetto e passare
l'altro come argomento.

```python
end.is_after(start)
```

Una cosa carina di questa sintassi è che in inglese si legge quasi come una domanda:
"`end` is after `start`?" ("`end` viene dopo `start`?")

## Il metodo `__str__`

Quando scrivi un metodo, puoi scegliere quasi qualsiasi nome.
Tuttavia, alcuni nomi hanno significati speciali.
Per esempio, se un oggetto ha un metodo chiamato `__str__`, Python usa quel metodo per convertire l'oggetto in stringa.
Per esempio, ecco un metodo `__str__` per un oggetto Time.

```python
%%add_method_to Time

    def __str__(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        return s
```

Questo metodo è simile a `print_time` del capitolo precedente, tranne che restituisce la stringa invece di stamparla.

Puoi invocare questo metodo nel modo consueto.

```python
end.__str__()
```

Ma Python può anche invocarlo al posto tuo.
Se usi la funzione predefinita `str` per convertire un oggetto `Time` in stringa, Python usa il metodo `__str__` della classe `Time`.

```python
str(end)
```

E fa lo stesso se stampi un oggetto `Time`.

```python
print(end)
```

Metodi come `__str__` si chiamano **metodi speciali** (*special methods*).
Puoi riconoscerli perché i loro nomi iniziano e finiscono con due trattini bassi.

## Il metodo init

Il più speciale dei metodi speciali è `__init__`, così chiamato perché inizializza gli attributi di un nuovo oggetto.
Un metodo `__init__` per la classe `Time` potrebbe essere così:

```python
%%add_method_to Time

    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second
```

Ora, quando istanziamo un oggetto `Time`, Python invoca `__init__` e gli passa gli argomenti.
Quindi possiamo creare un oggetto e inizializzarne gli attributi allo stesso tempo.

```python
time = Time(9, 40, 0)
print(time)
```

In questo esempio i parametri sono facoltativi, quindi se chiami `Time` senza argomenti,
ottieni i valori predefiniti.

```python
time = Time()
print(time)
```

Se fornisci un argomento, sovrascrive `hour`:

```python
time = Time(9)
print(time)
```

Se fornisci due argomenti, sovrascrivono `hour` e `minute`.

```python
time = Time(9, 45)
print(time)
```

E se fornisci tre argomenti, sovrascrivono tutti e tre i valori
predefiniti.

Quando scrivo una nuova classe, inizio quasi sempre scrivendo `__init__`, che rende più facile creare oggetti, e `__str__`, che è utile per il debugging.

## Sovraccarico degli operatori

Definendo altri metodi speciali, puoi specificare il comportamento
degli operatori sui tipi definiti dal programmatore. Per esempio, se definisci un
metodo chiamato `__add__` per la classe `Time`, puoi usare l'operatore `+`
sugli oggetti Time.

Ecco un metodo `__add__`.

```python
%%add_method_to Time

    def __add__(self, other):
        seconds = self.time_to_int() + other.time_to_int()
        return Time.int_to_time(seconds)
```

Possiamo usarlo così.

```python
duration = Time(1, 32)
end = start + duration
print(end)
```

Succedono molte cose quando eseguiamo queste tre righe di codice:

* Quando istanziamo un oggetto `Time`, viene invocato il metodo `__init__`.

* Quando usiamo l'operatore `+` con un oggetto `Time`, viene invocato il suo metodo `__add__`.

* E quando stampiamo un oggetto `Time`, viene invocato il suo metodo `__str__`.

Cambiare il comportamento di un operatore in modo che funzioni con i tipi definiti dal programmatore si chiama **sovraccarico degli operatori** (*operator overloading*).
Per ogni operatore, come `+`, c'è un metodo speciale corrispondente, come `__add__`.

## Debugging

Un oggetto `Time` è valido se i valori di `minute` e `second` sono compresi tra `0` e `60` – incluso `0` ma non `60` – e se `hour` è positivo.
Inoltre, `hour` e `minute` dovrebbero essere valori interi, ma potremmo ammettere che `second` abbia una parte frazionaria.
Requisiti come questi si chiamano **invarianti** (*invariants*) perché dovrebbero essere sempre veri.
Detto in altro modo, se non sono veri, qualcosa è andato storto.

Scrivere codice che verifichi gli invarianti può aiutare a individuare gli errori e a trovarne le cause.
Per esempio, potresti avere un metodo come `is_valid` che riceve un oggetto Time e restituisce `False` se viola un invariante.

```python
%%add_method_to Time

    def is_valid(self):
        if self.hour < 0 or self.minute < 0 or self.second < 0:
            return False
        if self.minute >= 60 or self.second >= 60:
            return False
        if not isinstance(self.hour, int):
            return False
        if not isinstance(self.minute, int):
            return False
        return True
```

Poi, all'inizio di ogni metodo, puoi controllare gli argomenti per assicurarti che siano validi.

```python
%%add_method_to Time

    def is_after(self, other):
        assert self.is_valid(), 'self is not a valid Time'
        assert other.is_valid(), 'other is not a valid Time'
        return self.time_to_int() > other.time_to_int()
```

L'istruzione `assert` valuta l'espressione che segue. Se il risultato è `True`, non fa nulla; se il risultato è `False`, provoca un `AssertionError`.
Ecco un esempio.

```python
duration = Time(minute=132)
print(duration)
```

```python
%%expect AssertionError

start.is_after(duration)
```

Le istruzioni `assert` sono utili perché distinguono il codice che gestisce le condizioni normali dal codice che controlla gli errori.

## Glossario

**linguaggio orientato agli oggetti** (*object-oriented language*):
Un linguaggio che fornisce funzionalità a supporto della programmazione orientata agli oggetti, in particolare i tipi definiti dall'utente.

**metodo** (*method*):
Una funzione definita all'interno di una definizione di classe e invocata sulle istanze di quella classe.

**ricevente** (*receiver*):
L'oggetto su cui viene invocato un metodo.

**metodo statico** (*static method*):
Un metodo che può essere invocato senza un oggetto come ricevente.

**metodo d'istanza** (*instance method*):
Un metodo che deve essere invocato con un oggetto come ricevente.

**metodo speciale** (*special method*):
Un metodo che cambia il modo in cui operatori e alcune funzioni lavorano con un oggetto.

**sovraccarico degli operatori** (*operator overloading*):
Il processo di usare metodi speciali per cambiare il modo in cui gli operatori lavorano con i tipi definiti dall'utente.

**invariante** (*invariant*):
Una condizione che dovrebbe essere sempre vera durante l'esecuzione di un programma.
