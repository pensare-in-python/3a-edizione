## Il modulo jupyturtle

Per usare il modulo `jupyturtle`, possiamo importarlo così.

````python
import jupyturtle
````

Ora possiamo usare le funzioni definite nel modulo, come `make_turtle` e `forward`.

````python
jupyturtle.make_turtle()
jupyturtle.forward(100)
````

`make_turtle` crea una **tela** (*canvas*), cioè uno spazio sullo schermo in cui possiamo disegnare, e una tartaruga, rappresentata da un guscio circolare e da una testa triangolare.
Il cerchio mostra la posizione della tartaruga e il triangolo indica la direzione verso cui è rivolta.

`forward` sposta la tartaruga di una data distanza nella direzione verso cui è rivolta, disegnando un segmento lungo il percorso.
La distanza è in unità arbitrarie – la dimensione effettiva dipende dallo schermo del tuo computer.

Useremo molte volte le funzioni definite nel modulo `jupyturtle`, quindi sarebbe comodo non dover scrivere ogni volta il nome del modulo.
È possibile se importiamo il modulo così.

````python
from jupyturtle import make_turtle, forward
````

Questa versione dell'istruzione import importa `make_turtle` e `forward` dal modulo `jupyturtle`, così possiamo chiamarle in questo modo.

````python
make_turtle()
forward(100)
````

`jupyturtle` fornisce altre due funzioni che useremo, chiamate `left` e `right`.
Le importeremo così.

````python
from jupyturtle import left, right
````

`left` fa girare la tartaruga a sinistra. Riceve un argomento, che è l'angolo della rotazione in gradi.
Per esempio, possiamo fare una svolta a sinistra di 90 gradi così.

````python
make_turtle()
forward(50)
left(90)
forward(50)
````

Questo programma sposta la tartaruga verso est e poi verso nord, lasciando dietro di sé due segmenti.
Prima di andare avanti, vedi se riesci a modificare il programma precedente per disegnare un quadrato.

## Disegnare un quadrato

Ecco un modo per disegnare un quadrato.

````python
make_turtle()

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)
````

Poiché questo programma ripete quattro volte la stessa coppia di righe, possiamo fare la stessa cosa in modo più conciso con un ciclo `for`.

````python
make_turtle()
for i in range(4):
    forward(50)
    left(90)
````

## Incapsulamento e generalizzazione

Prendiamo il codice che disegna il quadrato della sezione precedente e mettiamolo in una funzione chiamata `square`.

````python
def square():
    for i in range(4):
        forward(50)
        left(90)
````

Ora possiamo chiamare la funzione così.

````python
make_turtle()
square()
````

Racchiudere un pezzo di codice in una funzione si chiama **incapsulamento** (*encapsulation*).
Uno dei vantaggi dell'incapsulamento è che associa un nome al codice, e questo funge da documentazione. Un altro vantaggio è che, se riusi il codice, è più conciso chiamare una funzione due volte che copiare e incollare il corpo!

Nella versione attuale, la dimensione del quadrato è sempre `50`.
Se vogliamo disegnare quadrati di dimensioni diverse, possiamo ricevere la lunghezza dei lati come parametro.

````python
def square(length):
    for i in range(4):
        forward(length)
        left(90)
````

Ora possiamo disegnare quadrati di dimensioni diverse.

````python
make_turtle()
square(30)
square(60)
````

Aggiungere un parametro a una funzione si chiama **generalizzazione** (*generalization*) perché rende la funzione più generale: con la versione precedente il quadrato ha sempre la stessa dimensione; con questa versione può avere qualsiasi dimensione.

Se aggiungiamo un altro parametro, possiamo renderla ancora più generale.
La funzione seguente disegna poligoni regolari con un dato numero di lati.

````python
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
````

In un poligono regolare con `n` lati, l'angolo tra lati adiacenti è di `360 / n` gradi.

L'esempio seguente disegna un poligono di `7` lati con lati di lunghezza `30`.

````python
make_turtle()
polygon(7, 30)
````

Quando una funzione ha più di qualche argomento numerico, è facile dimenticare cosa rappresentano o in che ordine vanno messi.
Può essere una buona idea includere i nomi dei parametri nell'elenco degli argomenti.

````python
make_turtle()
polygon(n=7, length=30)
````

A volte vengono chiamati "argomenti con nome" (*named arguments*) perché includono i nomi dei parametri.
Ma in Python vengono chiamati più spesso **argomenti keyword** (*keyword arguments*; da non confondere con le parole chiave di Python come `for` e `def`).

Questo uso dell'operatore di assegnazione, `=`, ci ricorda come funzionano argomenti e parametri – quando chiami una funzione, gli argomenti vengono assegnati ai parametri.

## Approssimare un cerchio

Supponiamo ora di voler disegnare un cerchio.
Possiamo farlo, in modo approssimato, disegnando un poligono con un gran numero di lati, così ogni lato è abbastanza piccolo da essere difficile da vedere.
Ecco una funzione che usa `polygon` per disegnare un poligono di `30` lati che approssima un cerchio.

````python
import math

def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

`circle` riceve come parametro il raggio del cerchio.
Calcola `circumference`, che è la circonferenza di un cerchio con quel raggio.
`n` è il numero di lati, quindi `circumference / n` è la lunghezza di ciascun lato.

Questa funzione potrebbe impiegare molto tempo a essere eseguita.
Possiamo velocizzarla chiamando `make_turtle` con un argomento keyword chiamato `delay` che imposta il tempo, in secondi, che la tartaruga aspetta dopo ogni passo.
Il valore predefinito è `0.2` secondi – se lo impostiamo a `0.02` l'esecuzione è circa 10 volte più veloce.

````python
make_turtle(delay=0.02)
circle(30)
````

Un limite di questa soluzione è che `n` è una costante, il che significa
che per cerchi molto grandi i lati sono troppo lunghi, e per cerchi piccoli
sprechiamo tempo a disegnare lati molto corti.
Un'opzione è generalizzare la funzione ricevendo `n` come parametro.
Ma per ora teniamo le cose semplici.

## Refactoring

Scriviamo ora una versione più generale di `circle`, chiamata `arc`, che riceve un secondo parametro, `angle`, e disegna un arco di cerchio che copre l'angolo indicato.
Per esempio, se `angle` è `360` gradi, disegna un cerchio completo. Se `angle` è `180` gradi, disegna un semicerchio.

Per scrivere `circle` abbiamo potuto riusare `polygon`, perché un poligono con molti lati è una buona approssimazione di un cerchio.
Ma non possiamo usare `polygon` per scrivere `arc`.

Creeremo invece la versione più generale di `polygon`, chiamata `polyline`.

````python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)
````

`polyline` riceve come parametri il numero di segmenti da disegnare, `n`, la lunghezza dei segmenti, `length`, e l'angolo tra di essi, `angle`.

Ora possiamo riscrivere `polygon` in modo che usi `polyline`.

````python
def polygon(n, length):
    angle = 360.0 / n
    polyline(n, length, angle)
````

E possiamo usare `polyline` per scrivere `arc`.

````python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)
````

`arc` è simile a `circle`, tranne che calcola `arc_length`, che è una frazione della circonferenza del cerchio.

Infine, possiamo riscrivere `circle` in modo che usi `arc`.

````python
def circle(radius):
    arc(radius,  360)
````

Per verificare che queste funzioni si comportino come previsto, le useremo per disegnare qualcosa di simile a una chiocciola.
Con `delay=0`, la tartaruga va il più veloce possibile.

````python
make_turtle(delay=0)
polygon(n=20, length=9)
arc(radius=70, angle=70)
circle(radius=10)
````

In questo esempio siamo partiti da codice funzionante e lo abbiamo riorganizzato in funzioni diverse.
Modifiche come questa, che migliorano il codice senza cambiarne il comportamento, si chiamano **refactoring**.

Se avessimo pianificato in anticipo, avremmo potuto scrivere prima `polyline` ed evitare il refactoring, ma spesso all'inizio di un progetto non ne sai abbastanza per progettare tutte le funzioni.
Quando inizi a scrivere codice, capisci meglio il problema.
A volte il refactoring è segno che hai imparato qualcosa.

## Un piano di sviluppo

Un **piano di sviluppo** (*development plan*) è un procedimento per scrivere programmi.
Il procedimento che abbiamo usato in questo capitolo è "incapsulamento e generalizzazione".
I passi di questo procedimento sono:

1.  Inizia scrivendo un piccolo programma senza definizioni di funzione.

2.  Quando il programma funziona, individua una sua parte coerente,
    incapsulala in una funzione e dalle un nome.

3.  Generalizza la funzione aggiungendo parametri appropriati.

4.  Ripeti i passi da 1 a 3 finché non hai un insieme di funzioni funzionanti.

5.  Cerca occasioni per migliorare il programma con il refactoring. Per
    esempio, se hai codice simile in diversi punti, valuta di
    raccoglierlo in una funzione opportunamente generale.

Questo procedimento ha alcuni svantaggi – vedremo delle alternative più avanti – ma può essere utile se non sai in anticipo come dividere il programma in funzioni.
Questo approccio ti permette di progettare man mano.

La progettazione di una funzione ha due parti:

* L'**interfaccia** (*interface*) è il modo in cui si usa la funzione, compresi il suo nome, i parametri che riceve e ciò che la funzione deve fare.

* L'**implementazione** (*implementation*) è il modo in cui la funzione fa ciò che deve fare.

Per esempio, ecco la prima versione di `circle` che abbiamo scritto, che usa `polygon`.

````python
def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

Ed ecco la versione dopo il refactoring, che usa `arc`.

````python
def circle(radius):
    arc(radius,  360)
````

Queste due funzioni hanno la stessa interfaccia – ricevono gli stessi parametri e fanno la stessa cosa – ma hanno implementazioni diverse.

## Docstring

Una **docstring** è una stringa all'inizio di una funzione che ne spiega l'interfaccia ("doc" sta per "documentazione").
Ecco un esempio:

````python
def polyline(n, length, angle):
    """Draws line segments with the given length and angle between them.
    
    n: integer number of line segments
    length: length of the line segments
    angle: angle between segments (in degrees)
    """    
    for i in range(n):
        forward(length)
        left(angle)
````

Per convenzione, le docstring sono stringhe racchiuse tra tripli apici, dette anche **stringhe multiriga** (*multiline strings*) perché i tripli apici permettono alla stringa di estendersi su più righe.

Una docstring dovrebbe:

* Spiegare in modo conciso cosa fa la funzione, senza entrare nei dettagli di come funziona,

* Spiegare che effetto ha ciascun parametro sul comportamento della funzione, e

* Indicare di che tipo dovrebbe essere ciascun parametro, se non è ovvio.

Scrivere questo tipo di documentazione è una parte importante della progettazione delle interfacce.
Un'interfaccia ben progettata dovrebbe essere semplice da spiegare; se fai fatica a spiegare una delle tue funzioni, forse l'interfaccia può essere migliorata.

## Debugging

Un'interfaccia è come un contratto tra una funzione e chi la chiama. Il
chiamante si impegna a fornire certi argomenti e la funzione si impegna a
svolgere un certo lavoro.

Per esempio, `polyline` richiede tre argomenti: `n` deve essere un intero; `length` dovrebbe essere un numero positivo; e `angle` deve essere un numero, inteso in gradi.

Questi requisiti si chiamano **precondizioni** (*preconditions*) perché dovrebbero essere veri prima che la funzione inizi l'esecuzione. Al contrario, le condizioni alla fine della funzione sono **postcondizioni** (*postconditions*).
Le postcondizioni includono l'effetto voluto della funzione (come disegnare segmenti) e gli eventuali effetti collaterali (come spostare la tartaruga o fare altre modifiche).

Le precondizioni sono responsabilità del chiamante. Se il chiamante viola una precondizione e la funzione non funziona correttamente, il bug è nel chiamante, non nella funzione.

Se le precondizioni sono soddisfatte e le postcondizioni no, il bug è nella funzione. Se le tue pre- e postcondizioni sono chiare, possono aiutarti nel debugging.

## Glossario

**progettazione delle interfacce** (*interface design*):
Un procedimento per progettare l'interfaccia di una funzione, che comprende i parametri che deve ricevere.

**tela** (*canvas*):
Una finestra usata per visualizzare elementi grafici come linee, cerchi, rettangoli e altre forme.

**incapsulamento** (*encapsulation*):
Il processo di trasformare una sequenza di istruzioni in una definizione di funzione.

**generalizzazione** (*generalization*):
Il processo di sostituire qualcosa di inutilmente specifico (come un numero) con qualcosa di opportunamente generale (come una variabile o un parametro).

**argomento keyword** (*keyword argument*):
Un argomento che include il nome del parametro.

**refactoring**:
Il processo di modificare un programma funzionante per migliorare le interfacce delle funzioni e altre qualità del codice.

**piano di sviluppo** (*development plan*):
Un procedimento per scrivere programmi.

**docstring**:
Una stringa che compare all'inizio di una definizione di funzione per documentarne l'interfaccia.

**stringa multiriga** (*multiline string*):
Una stringa racchiusa tra tripli apici che può estendersi su più righe di un programma.

**precondizione** (*precondition*):
Un requisito che dovrebbe essere soddisfatto dal chiamante prima che una funzione inizi.

**postcondizione** (*postcondition*):
Un requisito che dovrebbe essere soddisfatto dalla funzione prima che termini.

La funzione seguente usa `penup` e `pendown` per spostare la tartaruga senza lasciare traccia.

````python
from jupyturtle import penup, pendown

def jump(length):
    """Move forward length units without leaving a trail.
    
    Postcondition: Leaves the pen down.
    """
    penup()
    forward(length)
    pendown()
````
